### 🧾 Input

* `n`: Total number of stairs

### 📤 Output

* Number of distinct ways to climb to the top

### ✅ Example

```python
Input: n = 3
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def climbStairs(self, n: int) -> int:
        if n <= 2:
            return n
        a, b = 1, 2
        for _ in range(3, n+1):
            a, b = b, a + b
        return b
```

### 🧾 Input

* `nums`: Amount of money in each house

### 📤 Output

* Maximum amount that can be robbed without robbing adjacent houses

### ✅ Example

```python
Input: nums = [2,7,9,3,1]
Output: 12
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def rob(self, nums: List[int]) -> int:
        prev1, prev2 = 0, 0
        for num in nums:
            prev1, prev2 = max(prev2 + num, prev1), prev1
        return prev1
```

  <summary><span style="font-size:19px;"><strong>Maximum Subarray - </strong><a href="https://leetcode.com/problems/maximum-subarray/" target="_blank">#53</a></span></summary>

### 🧾 Input

* `nums`: List of integers

### 📤 Output

* Maximum sum of any contiguous subarray

### ✅ Example

```python
Input: nums = [-2,1,-3,4,-1,2,1,-5,4]
Output: 6
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        current_sum = max_sum = nums[0]
        for num in nums[1:]:
            current_sum = max(num, current_sum + num)
            max_sum = max(max_sum, current_sum)
        return max_sum
```

**Time:** O(n)

### 🧾 Input

* `n`: An integer

### 📤 Output

* List of number of 1's in binary for each number `0` to `n`

### ✅ Example

```python
Input: n = 5
Output: [0,1,1,2,1,2]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def countBits(self, n: int) -> List[int]:
        dp = [0] * (n + 1)
        for i in range(1, n + 1):
            dp[i] = dp[i >> 1] + (i & 1)
        return dp
```

  <summary><span style="font-size:19px;"><strong>Unique Paths - </strong><a href="https://leetcode.com/problems/unique-paths/" target="_blank">#62</a></span></summary>

### 🧾 Input

* `m`, `n`: Grid dimensions

### 📤 Output

* Number of unique paths from top-left to bottom-right

### ✅ Example

```python
Input: m = 3, n = 2
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def uniquePaths(self, m: int, n: int) -> int:
        dp = [[1]*n for _ in range(m)]
        for i in range(1, m):
            for j in range(1, n):
                dp[i][j] = dp[i-1][j] + dp[i][j-1]
        return dp[-1][-1]
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(m × n)
**Space:** O(m × n)

</details>

### 🧾 Input

* `coins`: List of coin denominations
* `amount`: Target amount

### 📤 Output

* Minimum number of coins to make up the amount or -1

### ✅ Example

```python
Input: coins = [1, 2, 5], amount = 11
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def coinChange(self, coins: List[int], amount: int) -> int:
        dp = [amount + 1] * (amount + 1)
        dp[0] = 0
        for i in range(1, amount + 1):
            for coin in coins:
                if i - coin >= 0:
                    dp[i] = min(dp[i], dp[i - coin] + 1)
        return dp[amount] if dp[amount] != amount + 1 else -1
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n × m), where n = amount and m = number of coins
**Space:** O(n)

</details>

  <summary><span style="font-size:19px;"><strong>Longest Palindromic Substring - </strong><a href="https://leetcode.com/problems/longest-palindromic-substring/" target="_blank">#5</a></span></summary>

### 🧾 Input

* `s`: A string

### 📤 Output

* The longest palindromic substring in `s`

### ✅ Example

```python
Input: s = "babad"
Output: "bab"
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def longestPalindrome(self, s: str) -> str:
        n = len(s)
        dp = [[False] * n for _ in range(n)]
        start, max_len = 0, 1

        for i in range(n): dp[i][i] = True

        for end in range(1, n):
            for begin in range(end):
                if s[begin] == s[end]:
                    if end - begin == 1 or dp[begin + 1][end - 1]:
                        dp[begin][end] = True
                        if end - begin + 1 > max_len:
                            start = begin
                            max_len = end - begin + 1
        return s[start:start + max_len]
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n²)
**Space:** O(n²)

</details>

### 🧾 Input

* `nums`: List of positive integers

### 📤 Output

* Boolean indicating if the array can be partitioned into two subsets with equal sum

### ✅ Example

```python
Input: nums = [1,5,11,5]
Output: true
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def canPartition(self, nums: List[int]) -> bool:
        total = sum(nums)
        if total % 2 != 0:
            return False
        target = total // 2
        dp = set([0])
        for num in nums:
            dp |= {x + num for x in dp}
        return target in dp
```

**Time:** O(n × sum/2)

* Number of ways to decode the string

  <summary><span style="font-size:19px;"><strong>Unique Paths II - </strong><a href="https://leetcode.com/problems/unique-paths-ii/" target="_blank">#63</a></span></summary>

### 🧾 Input

* `obstacleGrid`: 2D grid with `1` as obstacle and `0` as free space

### 📤 Output

* Number of unique paths from top-left to bottom-right avoiding obstacles

### ✅ Example

```python
Input: obstacleGrid = [[0,0,0],[0,1,0],[0,0,0]]
Output: 2
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def uniquePathsWithObstacles(self, obstacleGrid: List[List[int]]) -> int:
        m, n = len(obstacleGrid), len(obstacleGrid[0])
        if obstacleGrid[0][0] == 1:
            return 0
        dp = [[0]*n for _ in range(m)]
        dp[0][0] = 1
        for i in range(m):
            for j in range(n):
                if obstacleGrid[i][j] == 1:
                    dp[i][j] = 0
                else:
                    if i > 0:
                        dp[i][j] += dp[i-1][j]
                    if j > 0:
                        dp[i][j] += dp[i][j-1]
        return dp[-1][-1]
```

</details>

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(m × n)

  <summary><span style="font-size:19px;"><strong>Target Sum - </strong><a href="https://leetcode.com/problems/target-sum/" target="_blank">#494</a></span></summary>

### 🧾 Input

* `nums`: List of integers
* `target`: Target sum

### 📤 Output

* Number of ways to assign + and - to reach target

### ✅ Example

```python
Input: nums = [1,1,1,1,1], target = 3
Output: 5
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def findTargetSumWays(self, nums: List[int], target: int) -> int:
        total = sum(nums)
        if (total + target) % 2 == 1 or total < abs(target):
            return 0
        s = (total + target) // 2
        dp = [0] * (s + 1)
        dp[0] = 1
        for num in nums:
            for i in range(s, num - 1, -1):
                dp[i] += dp[i - num]
        return dp[s]
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n × sum)
**Space:** O(sum)

</details>

  <summary><span style="font-size:19px;"><strong>Edit Distance - </strong><a href="https://leetcode.com/problems/edit-distance/" target="_blank">#72</a></span></summary>

### 🧾 Input

* `word1`, `word2`: Two strings

### 📤 Output

* Minimum number of operations to convert `word1` to `word2`

### ✅ Example

```python
Input: word1 = "horse", word2 = "ros"
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def minDistance(self, word1: str, word2: str) -> int:
        m, n = len(word1), len(word2)
        dp = [[0] * (n+1) for _ in range(m+1)]
        for i in range(m+1): dp[i][0] = i
        for j in range(n+1): dp[0][j] = j
        for i in range(1, m+1):
            for j in range(1, n+1):
                if word1[i-1] == word2[j-1]:
                    dp[i][j] = dp[i-1][j-1]
                else:
                    dp[i][j] = 1 + min(dp[i-1][j], dp[i][j-1], dp[i-1][j-1])
        return dp[m][n]
```

</details>

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(m × n)
**Space:** O(m × n)

  <summary><span style="font-size:19px;"><strong>Maximum Product Subarray - </strong><a href="https://leetcode.com/problems/maximum-product-subarray/" target="_blank">#152</a></span></summary>

### 🧾 Input

* `nums`: A list of integers

### 📤 Output

* Maximum product of any contiguous subarray

### ✅ Example

```python
Input: nums = [2,3,-2,4]
Output: 6
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def maxProduct(self, nums: List[int]) -> int:
        max_product = nums[0]
        curr_max = curr_min = nums[0]
        for num in nums[1:]:
            temp = curr_max
            curr_max = max(num, curr_max * num, curr_min * num)
            curr_min = min(num, temp * num, curr_min * num)
            max_product = max(max_product, curr_max)
        return max_product
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(1)

  <summary><span style="font-size:19px;"><strong>Burst Balloons - </strong><a href="https://leetcode.com/problems/burst-balloons/" target="_blank">#312</a></span></summary>

### 🧾 Input

* `nums`: List of balloon values

### 📤 Output

* Maximum coins you can collect by bursting balloons optimally

### ✅ Example

```python
Input: nums = [3,1,5,8]
Output: 167
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def maxCoins(self, nums: List[int]) -> int:
        nums = [1] + nums + [1]
        n = len(nums)
        dp = [[0] * n for _ in range(n)]

        for length in range(2, n):
            for left in range(n - length):
                right = left + length
                for i in range(left + 1, right):
                    dp[left][right] = max(
                        dp[left][right],
                        nums[left] * nums[i] * nums[right] + dp[left][i] + dp[i][right]
                    )
        return dp[0][n - 1]
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n³)
**Space:** O(n²)

  <summary><span style="font-size:19px;"><strong>Regular Expression Matching - </strong><a href="https://leetcode.com/problems/regular-expression-matching/" target="_blank">#10</a></span></summary>

### 🧾 Input

* `s`: String
* `p`: Pattern (with `.` and `*`)

### 📤 Output

* Boolean indicating if `s` matches pattern `p`

### ✅ Example

```python
Input: s = "mississippi", p = "mis*is*p*."
Output: False
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def isMatch(self, s: str, p: str) -> bool:
        m, n = len(s), len(p)
        dp = [[False] * (n + 1) for _ in range(m + 1)]
        dp[0][0] = True

        for j in range(2, n + 1):
            if p[j - 1] == '*':
                dp[0][j] = dp[0][j - 2]

        for i in range(1, m + 1):
            for j in range(1, n + 1):
                if p[j - 1] == '.' or p[j - 1] == s[i - 1]:
                    dp[i][j] = dp[i - 1][j - 1]
                elif p[j - 1] == '*':
                    dp[i][j] = dp[i][j - 2]
                    if p[j - 2] == '.' or p[j - 2] == s[i - 1]:
                        dp[i][j] |= dp[i - 1][j]
        return dp[m][n]
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(m × n)
**Space:** O(m × n)

  <summary><span style="font-size:19px;"><strong>Palindrome Partitioning II - </strong><a href="https://leetcode.com/problems/palindrome-partitioning-ii/" target="_blank">#132</a></span></summary>

### 🧾 Input

* `s`: A string

### 📤 Output

* Minimum cuts needed to partition string into palindromes

### ✅ Example

```python
Input: s = "aab"
Output: 1
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def minCut(self, s: str) -> int:
        n = len(s)
        is_palindrome = [[False] * n for _ in range(n)]
        dp = [0] * n

        for i in range(n):
            min_cut = i
            for j in range(i + 1):
                if s[j] == s[i] and (i - j < 2 or is_palindrome[j + 1][i - 1]):
                    is_palindrome[j][i] = True
                    min_cut = 0 if j == 0 else min(min_cut, dp[j - 1] + 1)
            dp[i] = min_cut
        return dp[-1]
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n²)
**Space:** O(n²)

* True if one is a scrambled version of the other

### ✅ Example

```python
Input: s1 = "great", s2 = "rgeat"
Output: True
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from functools import lru_cache

class Solution:
    def isScramble(self, s1: str, s2: str) -> bool:
        @lru_cache(None)
        def dfs(a, b):
            if a == b:
                return True
            if sorted(a) != sorted(b):
                return False
            for i in range(1, len(a)):
                if (dfs(a[:i], b[:i]) and dfs(a[i:], b[i:])) or \
                   (dfs(a[:i], b[-i:]) and dfs(a[i:], b[:-i])):
                    return True
            return False
        return dfs(s1, s2)
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n⁴) (due to memoization and substring operations)
**Space:** O(n³)

  <summary><span style="font-size:19px;"><strong>Word Break II - </strong><a href="https://leetcode.com/problems/word-break-ii/" target="_blank">#140</a></span></summary>

### 🧾 Input

* `s`: Input string
* `wordDict`: List of allowed words

### 📤 Output

* All possible sentences breaking the string using `wordDict`

### ✅ Example

```python
Input: s = "catsanddog", wordDict = ["cat","cats","and","sand","dog"]
Output: ["cats and dog","cat sand dog"]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def wordBreak(self, s: str, wordDict: List[str]) -> List[str]:
        word_set = set(wordDict)
        memo = {}

        def dfs(start):
            if start in memo:
                return memo[start]
            if start == len(s):
                return [""]
            res = []
            for end in range(start + 1, len(s) + 1):
                word = s[start:end]
                if word in word_set:
                    for sub in dfs(end):
                        res.append(word + (" " + sub if sub else ""))
            memo[start] = res
            return res

        return dfs(0)
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** Exponential in worst case (due to branching)
**Space:** O(n × m), with memoization reducing redundant calls

</details>

  <summary><span style="font-size:19px;"><strong>Interleaving String - </strong><a href="https://leetcode.com/problems/interleaving-string/" target="_blank">#97</a></span></summary>

### 🧾 Input

* `s1`, `s2`, `s3`: Strings

### 📤 Output

* Boolean indicating whether `s3` is an interleaving of `s1` and `s2`

### ✅ Example

```python
Input: s1 = "aabcc", s2 = "dbbca", s3 = "aadbbcbcac"
Output: True
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def isInterleave(self, s1: str, s2: str, s3: str) -> bool:
        if len(s1) + len(s2) != len(s3):
            return False

        dp = [[False]*(len(s2)+1) for _ in range(len(s1)+1)]
        dp[0][0] = True

        for i in range(1, len(s1)+1):
            dp[i][0] = dp[i-1][0] and s1[i-1] == s3[i-1]
        for j in range(1, len(s2)+1):
            dp[0][j] = dp[0][j-1] and s2[j-1] == s3[j-1]

        for i in range(1, len(s1)+1):
            for j in range(1, len(s2)+1):
                dp[i][j] = (dp[i-1][j] and s1[i-1] == s3[i+j-1]) or \
                           (dp[i][j-1] and s2[j-1] == s3[i+j-1])

        return dp[-1][-1]
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(m × n)
**Space:** O(m × n)

</details>

<details>
  <summary><span style="font-size:19px;"><strong>Max Profit in Job Scheduling - </strong><a href="https://leetcode.com/problems/maximum-profit-in-job-scheduling/" target="_blank">#1235</a></span></summary>

### 🧾 Input

* `startTime`, `endTime`, `profit`: Lists of equal length

### 📤 Output

* Maximum profit with non-overlapping jobs

### ✅ Example

```python
Input: startTime = [1,2,3,3], endTime = [3,4,5,6], profit = [50,10,40,70]
Output: 120
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from bisect import bisect_right

class Solution:
    def jobScheduling(self, startTime, endTime, profit) -> int:
        jobs = sorted(zip(startTime, endTime, profit), key=lambda x: x[1])
        dp = [(0, 0)]  # (endTime, profit)

        for s, e, p in jobs:
            i = bisect_right(dp, (s, float('inf'))) - 1
            if dp[i][1] + p > dp[-1][1]:
                dp.append((e, dp[i][1] + p))
        return dp[-1][1]
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n log n)
**Space:** O(n)

</details>

<details>
  <summary><span style="font-size:19px;"><strong>Number of Ways to Stay in the Same Place After Some Steps - </strong><a href="https://leetcode.com/problems/number-of-ways-to-stay-in-the-same-place-after-some-steps/" target="_blank">#1269</a></span></summary>

### 🧾 Input

* `steps`: Total steps to take
* `arrLen`: Length of array

### 📤 Output

* Number of ways to stay at index 0 after `steps`

### ✅ Example

```python
Input: steps = 3, arrLen = 2
Output: 4
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def numWays(self, steps: int, arrLen: int) -> int:
        MOD = 10**9 + 7
        max_pos = min(steps, arrLen)
        dp = [0] * (max_pos + 1)
        dp[0] = 1

        for _ in range(steps):
            prev = dp[:]
            for j in range(max_pos + 1):
                dp[j] = prev[j]
                if j > 0:
                    dp[j] = (dp[j] + prev[j - 1]) % MOD
                if j < max_pos:
                    dp[j] = (dp[j] + prev[j + 1]) % MOD
        return dp[0]
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(steps × min(steps, arrLen))
**Space:** O(min(steps, arrLen))

</details>
