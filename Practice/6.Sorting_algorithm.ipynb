<details>
  <summary><span style="font-size:19px;"><strong>Bubble Sort</strong></span></summary>

### 🧾 Input

* `arr`: List of integers

### 📤 Output

* Sorted list in ascending order

### ✅ Example

```python
Input: arr = [5, 1, 4, 2, 8]
Output: [1, 2, 4, 5, 8]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def bubbleSort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n²)  
**Space:** O(1)

</details>
</details>

<details>
  <summary><span style="font-size:19px;"><strong>Insertion Sort</strong></span></summary>

### 🧾 Input

* `arr`: List of integers

### 📤 Output

* Sorted list in ascending order

### ✅ Example

```python
Input: arr = [12, 11, 13, 5, 6]
Output: [5, 6, 11, 12, 13]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def insertionSort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n²)  
**Space:** O(1)

</details>
</details>

<details>
  <summary><span style="font-size:19px;"><strong>Selection Sort</strong></span></summary>

### 🧾 Input

* `arr`: List of integers

### 📤 Output

* Sorted list in ascending order

### ✅ Example

```python
Input: arr = [64, 25, 12, 22, 11]
Output: [11, 12, 22, 25, 64]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def selectionSort(arr):
    for i in range(len(arr)):
        min_idx = i
        for j in range(i+1, len(arr)):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n²)

**Space:** O(1)

</details>
</details>


<details>
  <summary><span style="font-size:19px;"><strong>Merge Sort</strong></span></summary>

### 🧾 Input

* `arr`: List of integers

### 📤 Output

* Sorted list in ascending order

### ✅ Example

```python
Input: arr = [38, 27, 43, 3, 9, 82, 10]
Output: [3, 9, 10, 27, 38, 43, 82]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def mergeSort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2
        L = arr[:mid]
        R = arr[mid:]

        mergeSort(L)
        mergeSort(R)

        i = j = k = 0

        while i < len(L) and j < len(R):
            if L[i] < R[j]:
                arr[k] = L[i]
                i += 1
            else:
                arr[k] = R[j]
                j += 1
            k += 1

        while i < len(L):
            arr[k] = L[i]
            i += 1
            k += 1
        while j < len(R):
            arr[k] = R[j]
            j += 1
            k += 1
    return arr
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>
**Time:** O(n log n)  
**Space:** O(n)
</details>
</details>

<details>
  <summary><span style="font-size:19px;"><strong>Quick Sort</strong></span></summary>

### 🧾 Input

* `arr`: List of integers

### 📤 Output

* Sorted list in ascending order

### ✅ Example

```python
Input: arr = [10, 7, 8, 9, 1, 5]
Output: [1, 5, 7, 8, 9, 10]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def quickSort(arr):
    if len(arr) <= 1:
        return arr
    else:
        pivot = arr[0]
        less = [x for x in arr[1:] if x <= pivot]
        greater = [x for x in arr[1:] if x > pivot]
        return quickSort(less) + [pivot] + quickSort(greater)
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>
**Time:** O(n log n) on average, O(n²) worst case  
**Space:** O(log n) (recursive stack)
</details>
</details>


<details>
  <summary><span style="font-size:19px;"><strong>Heap Sort</strong></span></summary>

### 🧾 Input

* `arr`: List of integers

### 📤 Output

* Sorted list in ascending order

### ✅ Example

```python
Input: arr = [4, 10, 3, 5, 1]
Output: [1, 3, 4, 5, 10]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def heapify(arr, n, i):
    largest = i
    l = 2*i + 1
    r = 2*i + 2

    if l < n and arr[l] > arr[largest]:
        largest = l

    if r < n and arr[r] > arr[largest]:
        largest = r

    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)

def heapSort(arr):
    n = len(arr)
    for i in range(n//2 - 1, -1, -1):
        heapify(arr, n, i)
    for i in range(n-1, 0, -1):
        arr[i], arr[0] = arr[0], arr[i]
        heapify(arr, i, 0)
    return arr
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>
**Time:** O(n log n)  
**Space:** O(1)
</details>
</details>

<details>
  <summary><span style="font-size:19px;"><strong>Counting Sort</strong></span></summary>

### 🧾 Input

* `arr`: List of non-negative integers

### 📤 Output

* Sorted list in ascending order

### ✅ Example

```python
Input: arr = [4, 2, 2, 8, 3, 3, 1]
Output: [1, 2, 2, 3, 3, 4, 8]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def countingSort(arr):
    max_val = max(arr)
    count = [0] * (max_val + 1)

    for num in arr:
        count[num] += 1

    output = []
    for i in range(len(count)):
        output.extend([i] * count[i])
    return output
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>
**Time:** O(n + k)  
**Space:** O(k) — where k is the range of input
</details>
</details>


<details>
  <summary><span style="font-size:19px;"><strong>Radix Sort</strong></span></summary>

### 🧾 Input

* `arr`: List of non-negative integers

### 📤 Output

* Sorted list in ascending order

### ✅ Example

```python
Input: arr = [170, 45, 75, 90, 802, 24, 2, 66]
Output: [2, 24, 45, 66, 75, 90, 170, 802]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def countingSortExp(arr, exp):
    n = len(arr)
    output = [0] * n
    count = [0] * 10

    for i in range(n):
        index = (arr[i] // exp) % 10
        count[index] += 1

    for i in range(1, 10):
        count[i] += count[i - 1]

    for i in range(n - 1, -1, -1):
        index = (arr[i] // exp) % 10
        output[count[index] - 1] = arr[i]
        count[index] -= 1

    for i in range(n):
        arr[i] = output[i]

def radixSort(arr):
    max_num = max(arr)
    exp = 1
    while max_num // exp > 0:
        countingSortExp(arr, exp)
        exp *= 10
    return arr
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>
**Time:** O(nk) — where k is the number of digits  
**Space:** O(n + k)
</details>
</details>


<details>
  <summary><span style="font-size:19px;"><strong>Bucket Sort</strong></span></summary>

### 🧾 Input

* `arr`: List of floating point numbers in \[0, 1)

### 📤 Output

* Sorted list in ascending order

### ✅ Example

```python
Input: arr = [0.897, 0.565, 0.656, 0.123, 0.665, 0.343]
Output: [0.123, 0.343, 0.565, 0.656, 0.665, 0.897]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def bucketSort(arr):
    n = len(arr)
    buckets = [[] for _ in range(n)]

    for val in arr:
        index = int(val * n)
        buckets[index].append(val)

    for bucket in buckets:
        bucket.sort()

    result = []
    for bucket in buckets:
        result.extend(bucket)
    return result
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>
**Time:** O(n + k) — best case with uniform distribution  
**Space:** O(n)
</details>
</details>


<details>
  <summary><span style="font-size:19px;"><strong>Topological Sort (Graphs)</strong></span></summary>

### 🧾 Input

* `V`: Number of vertices
* `adj`: Adjacency list of a Directed Acyclic Graph (DAG)

### 📤 Output

* A list representing topological order of vertices

### ✅ Example

```python
Input: V = 6, adj = [[2,3], [3], [3], [], [1,2], [0]]
Output: [5, 4, 0, 2, 3, 1] (any valid topological order)
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def topoSort(V, adj):
    in_degree = [0] * V
    for u in range(V):
        for v in adj[u]:
            in_degree[v] += 1

    queue = [i for i in range(V) if in_degree[i] == 0]
    topo_order = []

    while queue:
        u = queue.pop(0)
        topo_order.append(u)
        for v in adj[u]:
            in_degree[v] -= 1
            if in_degree[v] == 0:
                queue.append(v)
    return topo_order
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>
**Time:** O(V + E)  
**Space:** O(V)
</details>
</details>
