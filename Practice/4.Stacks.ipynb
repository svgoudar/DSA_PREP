### 🧾 Input

* `s`: A string containing just `'(', ')', '{', '}', '['` and `']'`

### 📤 Output

* `True` if the input string is valid, otherwise `False`

### ✅ Example

```python
Input: s = "()[]{}"
Output: True
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def isValid(self, s: str) -> bool:
        stack = []
        mapping = {')': '(', '}': '{', ']': '['}
        for ch in s:
            if ch in mapping:
                top = stack.pop() if stack else '#'
                if mapping[ch] != top:
                    return False
            else:
                stack.append(ch)
        return not stack
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(n)

  <summary><span style="font-size:19px;"><strong>Min Stack - </strong><a href="https://leetcode.com/problems/min-stack/" target="_blank">#155</a></span></summary>

### 🧾 Input

Operations to support: `push`, `pop`, `top`, and `getMin`

### 📤 Output

Each operation executes in constant time

### ✅ Example

```python
Input: ["MinStack","push","push","push","getMin","pop","top","getMin"]
       [[],[-2],[0],[-3],[],[],[],[]]
Output: [null,null,null,null,-3,null,0,-2]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class MinStack:

    def __init__(self):
        self.stack = []
        self.min_stack = []

    def push(self, val: int) -> None:
        self.stack.append(val)
        if not self.min_stack or val <= self.min_stack[-1]:
            self.min_stack.append(val)

    def pop(self) -> None:
        if self.stack.pop() == self.min_stack[-1]:
            self.min_stack.pop()

    def top(self) -> int:
        return self.stack[-1]

    def getMin(self) -> int:
        return self.min_stack[-1]
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(1) per operation
**Space:** O(n)

</details>
</details>

  <summary><span style="font-size:19px;"><strong>Implement Queue using Stacks - </strong><a href="https://leetcode.com/problems/implement-queue-using-stacks/" target="_blank">#232</a></span></summary>

### 🧾 Input

* Push and pop elements from a queue using two stacks

### 📤 Output

* Implement `push`, `pop`, `peek`, and `empty` methods

### ✅ Example

```python
Input: ["MyQueue", "push", "push", "peek", "pop", "empty"]
       [[], [1], [2], [], [], []]
Output: [null, null, null, 1, 1, false]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class MyQueue:

    def __init__(self):
        self.input = []
        self.output = []

    def push(self, x: int) -> None:
        self.input.append(x)

    def pop(self) -> int:
        self.peek()
        return self.output.pop()

    def peek(self) -> int:
        if not self.output:
            while self.input:
                self.output.append(self.input.pop())
        return self.output[-1]

    def empty(self) -> bool:
        return not self.input and not self.output
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(1) amortized per operation
**Space:** O(n)

</details>

### ✅ Example

```python
Input: ops = ["5","2","C","D","+"]
Output: 30
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def calPoints(self, ops: List[str]) -> int:
        stack = []
        for op in ops:
            if op == "C":
                stack.pop()
            elif op == "D":
                stack.append(2 * stack[-1])
            elif op == "+":
                stack.append(stack[-1] + stack[-2])
            else:
                stack.append(int(op))
        return sum(stack)
```

```python
class Solution:
    def removeDuplicates(self, s: str) -> str:
        stack = []
        for ch in s:
            if stack and stack[-1] == ch:
                stack.pop()
            else:
                stack.append(ch)
        return ''.join(stack)
```

```python
Input: prices = [8,4,6,2,3]
Output: [4,2,4,2,3]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def finalPrices(self, prices: List[int]) -> List[int]:
        stack = []
        for i in range(len(prices)):
            while stack and prices[stack[-1]] >= prices[i]:
                idx = stack.pop()
                prices[idx] -= prices[i]
            stack.append(i)
        return prices
```

### 🧾 Input

* `s`, `t`: Strings with `#` representing backspaces

### 📤 Output

* Return `True` if both strings are equal after applying backspace

### ✅ Example

```python
Input: s = "ab#c", t = "ad#c"
Output: True
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def backspaceCompare(self, s: str, t: str) -> bool:
        def build(string):
            stack = []
            for ch in string:
                if ch == '#':
                    if stack:
                        stack.pop()
                else:
                    stack.append(ch)
            return ''.join(stack)
        return build(s) == build(t)
```

**Time:** O(n + m)

  <summary><span style="font-size:19px;"><strong>Implement Stack using Queues - </strong><a href="https://leetcode.com/problems/implement-stack-using-queues/" target="_blank">#225</a></span></summary>

### 🧾 Input

* Implement stack using only standard queue operations

### 📤 Output

* Support `push`, `pop`, `top`, and `empty`

### ✅ Example

```python
Input: ["MyStack","push","push","top","pop","empty"]
       [[],[1],[2],[],[],[]]
Output: [null,null,null,2,2,false]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import deque

class MyStack:

    def __init__(self):
        self.q = deque()

    def push(self, x: int) -> None:
        self.q.append(x)
        for _ in range(len(self.q) - 1):
            self.q.append(self.q.popleft())

    def pop(self) -> int:
        return self.q.popleft()

    def top(self) -> int:
        return self.q[0]

    def empty(self) -> bool:
        return not self.q
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n) for `push`, O(1) for others
**Space:** O(n)

</details>
</details>

  <summary><span style="font-size:19px;"><strong>Evaluate Reverse Polish Notation - </strong><a href="https://leetcode.com/problems/evaluate-reverse-polish-notation/" target="_blank">#150</a></span></summary>

### 🧾 Input

* `tokens`: List of strings representing RPN expression

### 📤 Output

* Evaluate and return result as an integer

### ✅ Example

```python
Input: tokens = ["2","1","+","3","*"]
Output: 9
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def evalRPN(self, tokens: List[str]) -> int:
        stack = []
        for token in tokens:
            if token in "+-*/":
                b = stack.pop()
                a = stack.pop()
                if token == '+':
                    stack.append(a + b)
                elif token == '-':
                    stack.append(a - b)
                elif token == '*':
                    stack.append(a * b)
                elif token == '/':
                    stack.append(int(a / b))
            else:
                stack.append(int(token))
        return stack[0]
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(n)

</details>

### 📤 Output

* Return a list where `answer[i]` is the number of days to wait for a warmer temperature

### ✅ Example

```python
Input: temperatures = [73,74,75,71,69,72,76,73]
Output: [1,1,4,2,1,1,0,0]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def dailyTemperatures(self, temperatures: List[int]) -> List[int]:
        res = [0] * len(temperatures)
        stack = []
        for i, temp in enumerate(temperatures):
            while stack and temperatures[stack[-1]] < temp:
                prev = stack.pop()
                res[prev] = i - prev
            stack.append(i)
        return res
```

### 🧾 Input

* `nums`: Circular list of integers

### 📤 Output

* Return an array with the next greater number for each element

### ✅ Example

```python
Input: nums = [1,2,1]
Output: [2,-1,2]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def nextGreaterElements(self, nums: List[int]) -> List[int]:
        n = len(nums)
        res = [-1] * n
        stack = []
        for i in range(2 * n):
            while stack and nums[stack[-1]] < nums[i % n]:
                res[stack.pop()] = nums[i % n]
            if i < n:
                stack.append(i)
        return res
```

```python
class Solution:
    def simplifyPath(self, path: str) -> str:
        stack = []
        parts = path.split('/')
        for part in parts:
            if part == '..':
                if stack:
                    stack.pop()
            elif part and part != '.':
                stack.append(part)
        return '/' + '/'.join(stack)
```

<details>

```python
class Solution:
    def scoreOfParentheses(self, s: str) -> int:
        stack = [0]
        for char in s:
            if char == '(':
                stack.append(0)
            else:
                val = stack.pop()
                stack[-1] += max(2 * val, 1)
        return stack.pop()
```

* `asteroids`: Array of integers representing moving asteroids

### 📤 Output

* The state of asteroids after collisions

### ✅ Example

```python
Input: [5,10,-5]
Output: [5,10]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def asteroidCollision(self, asteroids: List[int]) -> List[int]:
        stack = []
        for ast in asteroids:
            while stack and ast < 0 < stack[-1]:
                if stack[-1] < -ast:
                    stack.pop()
                    continue
                elif stack[-1] == -ast:
                    stack.pop()
                break
            else:
                stack.append(ast)
        return stack
```

* `num`: A string representing a number

* Smallest possible number after removing k digits

### ✅ Example

```python
Input: num = "1432219", k = 3
Output: "1219"
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def removeKdigits(self, num: str, k: int) -> str:
        stack = []
        for digit in num:
            while k and stack and stack[-1] > digit:
                stack.pop()
                k -= 1
            stack.append(digit)
        stack = stack[:-k] if k else stack
        return ''.join(stack).lstrip('0') or "0"
```

```python
Input: [2,1,5,6,2,3]
Output: 10
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def largestRectangleArea(self, heights: List[int]) -> int:
        stack = []
        max_area = 0
        heights.append(0)
        for i, h in enumerate(heights):
            while stack and heights[stack[-1]] > h:
                height = heights[stack.pop()]
                width = i if not stack else i - stack[-1] - 1
                max_area = max(max_area, height * width)
            stack.append(i)
        return max_area
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def find132pattern(self, nums: List[int]) -> bool:
        stack = []
        third = float('-inf')
        for num in reversed(nums):
            if num < third:
                return True
            while stack and stack[-1] < num:
                third = stack.pop()
            stack.append(num)
        return False
```

### ✅ Example

```python
Input: "lee(t(c)o)de)"
Output: "lee(t(c)o)de"
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def minRemoveToMakeValid(self, s: str) -> str:
        s = list(s)
        stack = []
        for i, ch in enumerate(s):
            if ch == '(':
                stack.append(i)
            elif ch == ')':
                if stack:
                    stack.pop()
                else:
                    s[i] = ''
        for i in stack:
            s[i] = ''
        return ''.join(s)
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def maxDepthAfterSplit(self, seq: str) -> List[int]:
        res = []
        depth = 0
        for ch in seq:
            if ch == '(':
                depth += 1
                res.append(depth % 2)
            else:
                res.append(depth % 2)
                depth -= 1
        return res
```

  <summary><span style="font-size:19px;"><strong>Basic Calculator - </strong><a href="https://leetcode.com/problems/basic-calculator/" target="_blank">#224</a></span></summary>

### 🧾 Input

* `s`: A string representing a valid arithmetic expression containing `+`, `-`, `(`, `)`, and digits.

### 📤 Output

* The result of evaluating the expression.

### ✅ Example

```python
Input: s = "(1+(4+5+2)-3)+(6+8)"
Output: 23
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def calculate(self, s: str) -> int:
        stack = []
        result, number, sign = 0, 0, 1

        for ch in s:
            if ch.isdigit():
                number = number * 10 + int(ch)
            elif ch in '+-':
                result += sign * number
                number = 0
                sign = 1 if ch == '+' else -1
            elif ch == '(':
                stack.append(result)
                stack.append(sign)
                result, sign = 0, 1
            elif ch == ')':
                result += sign * number
                number = 0
                result *= stack.pop()
                result += stack.pop()
        return result + sign * number
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def calculate(self, s: str) -> int:
        def helper(it):
            stack, num, sign = [], 0, '+'
            while True:
                try:
                    ch = next(it)
                except StopIteration:
                    break
                if ch.isdigit():
                    num = num * 10 + int(ch)
                elif ch == '(':
                    num = helper(it)
                if ch in '+-*/)' or ch is None:
                    if sign == '+': stack.append(num)
                    elif sign == '-': stack.append(-num)
                    elif sign == '*': stack[-1] *= num
                    elif sign == '/': stack[-1] = int(stack[-1] / num)
                    num = 0
                    sign = ch
                    if ch == ')':
                        break
            return sum(stack)
        
        return helper(iter(s.replace(' ', '')))
```

### ✅ Example

```python
Input: s = "bcabc"
Output: "abc"
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def removeDuplicateLetters(self, s: str) -> str:
        stack = []
        seen = set()
        last_index = {ch: i for i, ch in enumerate(s)}
        
        for i, ch in enumerate(s):
            if ch in seen:
                continue
            while stack and ch < stack[-1] and i < last_index[stack[-1]]:
                seen.remove(stack.pop())
            stack.append(ch)
            seen.add(ch)
        return ''.join(stack)
```

### 🧾 Input

Operations on a custom stack:

* `push(val)`: Pushes an integer `val`
* `pop()`: Removes and returns the most frequent element (most recent in case of tie)

### 📤 Output

Simulated results of operations

### ✅ Example

```python
Input: ["push", "push", "push", "push", "push", "push", "pop", "pop", "pop", "pop"]
       [[5], [7], [5], [7], [4], [5], [], [], [], []]
Output: [null, null, null, null, null, null, 5, 7, 5, 4]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import defaultdict

class FreqStack:
    def __init__(self):
        self.freq = defaultdict(int)
        self.group = defaultdict(list)
        self.maxfreq = 0

    def push(self, val: int) -> None:
        f = self.freq[val] + 1
        self.freq[val] = f
        self.group[f].append(val)
        self.maxfreq = max(self.maxfreq, f)

    def pop(self) -> int:
        val = self.group[self.maxfreq].pop()
        self.freq[val] -= 1
        if not self.group[self.maxfreq]:
            self.maxfreq -= 1
        return val
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(1) per operation
**Space:** O(n)

### 🧾 Input

* `height`: List of non-negative integers representing elevation

### 📤 Output

* Total amount of rainwater trapped

### ✅ Example

```python
Input: height = [0,1,0,2,1,0,1,3,2,1,2,1]
Output: 6
```

<details>
  <summary><strong>Solution Code (Stack Approach)</strong></summary>

```python
class Solution:
    def trap(self, height: List[int]) -> int:
        stack, res = [], 0
        for i, h in enumerate(height):
            while stack and height[stack[-1]] < h:
                top = stack.pop()
                if not stack:
                    break
                dist = i - stack[-1] - 1
                bounded_height = min(h, height[stack[-1]]) - height[top]
                res += dist * bounded_height
            stack.append(i)
        return res
```

### 🧾 Input

* `heights`: List of bar heights

### 📤 Output

* Area of the largest rectangle in the histogram

### ✅ Example

```python
Input: heights = [2,1,5,6,2,3]
Output: 10
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def largestRectangleArea(self, heights: List[int]) -> int:
        stack = [-1]
        max_area = 0
        for i in range(len(heights)):
            while stack[-1] != -1 and heights[stack[-1]] > heights[i]:
                h = heights[stack.pop()]
                w = i - stack[-1] - 1
                max_area = max(max_area, h * w)
            stack.append(i)
        while stack[-1] != -1:
            h = heights[stack.pop()]
            w = len(heights) - stack[-1] - 1
            max_area = max(max_area, h * w)
        return max_area
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def longestValidParentheses(self, s: str) -> int:
        stack = [-1]
        max_len = 0
        for i, ch in enumerate(s):
            if ch == '(':
                stack.append(i)
            else:
                stack.pop()
                if not stack:
                    stack.append(i)
                else:
                    max_len = max(max_len, i - stack[-1])
        return max_len
```

  <summary><span style="font-size:19px;"><strong>Decode String - </strong><a href="https://leetcode.com/problems/decode-string/" target="_blank">#394</a></span></summary>

### 🧾 Input

* `s`: Encoded string with format like `"3[a2[c]]"`

### 📤 Output

* Decoded string

### ✅ Example

```python
Input: s = "3[a2[c]]"
Output: "accaccacc"
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def decodeString(self, s: str) -> str:
        stack = []
        curr_num = 0
        curr_str = ''
        for ch in s:
            if ch.isdigit():
                curr_num = curr_num * 10 + int(ch)
            elif ch == '[':
                stack.append((curr_str, curr_num))
                curr_str, curr_num = '', 0
            elif ch == ']':
                last_str, num = stack.pop()
                curr_str = last_str + num * curr_str
            else:
                curr_str += ch
        return curr_str
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(n)
