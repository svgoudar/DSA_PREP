### 🧾 Input

- `nums`: Sorted list of integers

### 📤 Output

- Length of array after removing duplicates (in-place)

### ✅ Example

```py
Input: nums = [1,1,2]
Output: 2
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def removeDuplicates(self, nums: List[int]) -> int:
        if not nums:
            return 0
        slow = 0
        for fast in range(1, len(nums)):
            if nums[fast] != nums[slow]:
                slow += 1
                nums[slow] = nums[fast]
        return slow + 1
```

### 🧾 Input

- `nums1`: First sorted array with extra space
- `m`: Number of valid elements in `nums1`
- `nums2`: Second sorted array
- `n`: Number of elements in `nums2`

### 📤 Output

- `nums1` should contain the merged sorted array

### ✅ Example

```py
Input: nums1 = [1,2,3,0,0,0], m = 3, nums2 = [2,5,6], n = 3
Output: [1,2,2,3,5,6]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def merge(self, nums1: List[int], m: int, nums2: List[int], n: int) -> None:
        i, j, k = m - 1, n - 1, m + n - 1
        while i >= 0 and j >= 0:
            if nums1[i] > nums2[j]:
                nums1[k] = nums1[i]
                i -= 1
            else:
                nums1[k] = nums2[j]
                j -= 1
            k -= 1
        while j >= 0:
            nums1[k] = nums2[j]
            j -= 1
            k -= 1
```

</details>

**Time Complexity:** O(m + n)

### 🧾 Input

- `numbers`: List of integers (sorted in non-decreasing order)
- `target`: Integer target sum

### 📤 Output

- Indices (1-based) of the two numbers such that they add up to `target`

### ✅ Example

```py
Input: numbers = [2,7,11,15], target = 9
Output: [1,2]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def twoSum(self, numbers: List[int], target: int) -> List[int]:
        left, right = 0, len(numbers) - 1
        while left < right:
            curr_sum = numbers[left] + numbers[right]
            if curr_sum == target:
                return [left + 1, right + 1]
            elif curr_sum < target:
                left += 1
            else:
                right -= 1
```

- `nums1`, `nums2`: Lists of integers

### 📤 Output

- List of common elements (including duplicates)

### ✅ Example

```py
Input: nums1 = [1,2,2,1], nums2 = [2,2]
Output: [2,2]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```py

from collections import Counter
class Solution:
    def intersect(self, nums1: List[int], nums2: List[int]) -> List[int]:
        counts = Counter(nums1)
        result = []
        for num in nums2:
            if counts[num] > 0:
                result.append(num)
                counts[num] -= 1
        return result
```

### 🧾 Input

* `nums`: List of integers

### 📤 Output

* List of all unique triplets that sum up to 0

### ✅ Example

```py
Input: nums = [-1,0,1,2,-1,-4]
Output: [[-1,-1,2],[-1,0,1]]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def threeSum(self, nums: List[int]) -> List[List[int]]:
        nums.sort()
        res = []
        for i in range(len(nums)):
            if i > 0 and nums[i] == nums[i - 1]:
                continue
            l, r = i + 1, len(nums) - 1
            while l < r:
                total = nums[i] + nums[l] + nums[r]
                if total < 0:
                    l += 1
                elif total > 0:
                    r -= 1
                else:
                    res.append([nums[i], nums[l], nums[r]])
                    while l < r and nums[l] == nums[l + 1]: l += 1
                    while l < r and nums[r] == nums[r - 1]: r -= 1
                    l += 1
                    r -= 1
        return res
```

### 🧾 Input

* `height`: List of non-negative integers

### 📤 Output

* Maximum area of water a container can store

### ✅ Example

```py
Input: height = [1,8,6,2,5,4,8,3,7]
Output: 49
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def maxArea(self, height: List[int]) -> int:
        left, right = 0, len(height) - 1
        max_area = 0
        while left < right:
            max_area = max(max_area, (right - left) * min(height[left], height[right]))
            if height[left] < height[right]:
                left += 1
            else:
                right -= 1
        return max_area
```

* `target`: Integer
* `nums`: List of positive integers

### 📤 Output

* Smallest length of a subarray with sum ≥ target

### ✅ Example

```py
Input: target = 7, nums = [2,3,1,2,4,3]
Output: 2
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def minSubArrayLen(self, target: int, nums: List[int]) -> int:
        left = 0
        total = 0
        min_len = float('inf')
        for right in range(len(nums)):
            total += nums[right]
            while total >= target:
                min_len = min(min_len, right - left + 1)
                total -= nums[left]
                left += 1
        return 0 if min_len == float('inf') else min_len
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def judgeSquareSum(self, c: int) -> bool:
        left, right = 0, int(c**0.5)
        while left <= right:
            total = left * left + right * right
            if total == c:
                return True
            elif total < c:
                left += 1
            else:
                right -= 1
        return False
```

### 📤 Output

* Rearrange `nums` into the next lexicographically greater permutation

### ✅ Example

```py
Input: nums = [1,2,3]
Output: [1,3,2]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def nextPermutation(self, nums: List[int]) -> None:
        i = len(nums) - 2
        while i >= 0 and nums[i] >= nums[i + 1]:
            i -= 1
        if i >= 0:
            j = len(nums) - 1
            while nums[j] <= nums[i]:
                j -= 1
            nums[i], nums[j] = nums[j], nums[i]
        nums[i + 1:] = reversed(nums[i + 1:])
```

  <summary><span style="font-size:19px;"><strong>Trapping Rain Water II - </strong><a href="https://leetcode.com/problems/trapping-rain-water-ii/" target="_blank">#407</a></span></summary>

### 🧾 Input

* `heightMap`: 2D list of integers representing elevation heights

### 📤 Output

* Total water trapped after raining

### ✅ Example

```py
Input: heightMap = [[1,4,3,1,3,2],[3,2,1,3,2,4],[2,3,3,2,3,1]]
Output: 4
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
import heapq

class Solution:
    def trapRainWater(self, heightMap: List[List[int]]) -> int:
        if not heightMap or not heightMap[0]:
            return 0
        m, n = len(heightMap), len(heightMap[0])
        visited = [[False] * n for _ in range(m)]
        heap = []

        for i in range(m):
            for j in (0, n - 1):
                heapq.heappush(heap, (heightMap[i][j], i, j))
                visited[i][j] = True
        for j in range(n):
            for i in (0, m - 1):
                if not visited[i][j]:
                    heapq.heappush(heap, (heightMap[i][j], i, j))
                    visited[i][j] = True

        water = 0
        directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]
        while heap:
            height, x, y = heapq.heappop(heap)
            for dx, dy in directions:
                nx, ny = x + dx, y + dy
                if 0 <= nx < m and 0 <= ny < n and not visited[nx][ny]:
                    water += max(0, height - heightMap[nx][ny])
                    heapq.heappush(heap, (max(height, heightMap[nx][ny]), nx, ny))
                    visited[nx][ny] = True
        return water
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(mn log(mn))
**Space Complexity:** O(mn)

</details>

```py
Input: arr = [9,4,2,10,7,8,8,1,9]
Output: 5
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def maxTurbulenceSize(self, arr: List[int]) -> int:
        up = down = res = 1
        for i in range(1, len(arr)):
            if arr[i] > arr[i - 1]:
                up, down = down + 1, 1
            elif arr[i] < arr[i - 1]:
                down, up = up + 1, 1
            else:
                up = down = 1
            res = max(res, max(up, down))
        return res
```

<details>

```python
class Solution:
    def reorderLogFiles(self, logs: List[str]) -> List[str]:
        def get_key(log):
            id_, rest = log.split(" ", 1)
            return (0, rest, id_) if rest[0].isalpha() else (1,)

        return sorted(logs, key=get_key)
```

  <summary><span style="font-size:19px;"><strong>Trapping Rain Water - </strong><a href="https://leetcode.com/problems/trapping-rain-water/" target="_blank">#42</a></span></summary>

### 🧾 Input

* `height`: List of non-negative integers representing elevation map

### 📤 Output

* Amount of rain water that can be trapped

### ✅ Example

```py
Input: height = [0,1,0,2,1,0,1,3,2,1,2,1]
Output: 6
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def trap(self, height: List[int]) -> int:
        left, right = 0, len(height) - 1
        left_max = right_max = water = 0
        while left < right:
            if height[left] < height[right]:
                left_max = max(left_max, height[left])
                water += left_max - height[left]
                left += 1
            else:
                right_max = max(right_max, height[right])
                water += right_max - height[right]
                right -= 1
        return water
```

</details>

  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(n)
**Space Complexity:** O(1)

```py
Input: nums = [1,1,0,1]
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def longestSubarray(self, nums: List[int]) -> int:
        left = 0
        zero_count = 0
        max_len = 0
        for right in range(len(nums)):
            if nums[right] == 0:
                zero_count += 1
            while zero_count > 1:
                if nums[left] == 0:
                    zero_count -= 1
                left += 1
            max_len = max(max_len, right - left)
        return max_len
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(n)
**Space Complexity:** O(1)

  <summary><span style="font-size:19px;"><strong>Palindrome Partitioning II - </strong><a href="https://leetcode.com/problems/palindrome-partitioning-ii/" target="_blank">#132</a></span></summary>

### 🧾 Input

* `s`: A string

### 📤 Output

* Minimum number of cuts needed to partition the string such that every substring is a palindrome

### ✅ Example

```py
Input: s = "aab"
Output: 1
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def minCut(self, s: str) -> int:
        n = len(s)
        dp = [0] * n
        is_palindrome = [[False] * n for _ in range(n)]

        for i in range(n):
            min_cut = i
            for j in range(i + 1):
                if s[j] == s[i] and (i - j < 2 or is_palindrome[j + 1][i - 1]):
                    is_palindrome[j][i] = True
                    min_cut = 0 if j == 0 else min(min_cut, dp[j - 1] + 1)
            dp[i] = min_cut
        return dp[-1]
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(n²)
**Space Complexity:** O(n²)

  <summary><span style="font-size:19px;"><strong>Substring with Concatenation of All Words - </strong><a href="https://leetcode.com/problems/substring-with-concatenation-of-all-words/" target="_blank">#30</a></span></summary>

### 🧾 Input

* `s`: A string
* `words`: List of words (all the same length)

### 📤 Output

* Starting indices of substring(s) that is the concatenation of all the words

### ✅ Example

```py
Input: s = "barfoothefoobarman", words = ["foo","bar"]
Output: [0,9]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def findSubstring(self, s: str, words: List[str]) -> List[int]:
        from collections import Counter
        if not s or not words:
            return []
        word_len = len(words[0])
        total_len = word_len * len(words)
        word_count = Counter(words)
        result = []

        for i in range(word_len):
            left = i
            right = i
            cur_count = Counter()
            while right + word_len <= len(s):
                word = s[right:right + word_len]
                right += word_len
                if word in word_count:
                    cur_count[word] += 1
                    while cur_count[word] > word_count[word]:
                        cur_count[s[left:left + word_len]] -= 1
                        left += word_len
                    if right - left == total_len:
                        result.append(left)
                else:
                    cur_count.clear()
                    left = right
        return result
```

</details>

  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(N \* M) where N is len(s), M is len(words)
**Space Complexity:** O(M)

  <summary><span style="font-size:19px;"><strong>Minimum Window Subsequence - </strong><a href="https://leetcode.com/problems/minimum-window-subsequence/" target="_blank">#727</a></span></summary>

### 🧾 Input

* `s1`: Source string
* `s2`: Target subsequence string

### 📤 Output

* Minimum window in `s1` which contains all characters of `s2` in order

### ✅ Example

```py
Input: s1 = "abcdebdde", s2 = "bde"
Output: "bcde"
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def minWindow(self, s1: str, s2: str) -> str:
        m, n = len(s1), len(s2)
        res = ""
        i = 0
        while i < m:
            if s1[i] == s2[0]:
                j = 0
                k = i
                while k < m and j < n:
                    if s1[k] == s2[j]:
                        j += 1
                    k += 1
                if j == n:
                    end = k
                    j -= 1
                    k -= 1
                    while j >= 0:
                        if s1[k] == s2[j]:
                            j -= 1
                        k -= 1
                    start = k + 1
                    if not res or end - start < len(res):
                        res = s1[start:end]
                    i = start
            i += 1
        return res
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(m \* n)
**Space Complexity:** O(1)

### 🧾 Input

* `dist`: List of distances
* `hour`: Time available to arrive

### 📤 Output

* Minimum integer speed required to reach on time, or -1 if impossible

### ✅ Example

```py
Input: dist = [1,3,2], hour = 6
Output: 1
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def minSpeedOnTime(self, dist: List[int], hour: float) -> int:
        import math
        if len(dist) - 1 >= hour:
            return -1

        def time_needed(speed):
            total = 0
            for d in dist[:-1]:
                total += math.ceil(d / speed)
            total += dist[-1] / speed
            return total

        left, right = 1, 10**7
        res = -1
        while left <= right:
            mid = (left + right) // 2
            if time_needed(mid) <= hour:
                res = mid
                right = mid - 1
            else:
                left = mid + 1
        return res
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(n log M), where M = max speed
**Space Complexity:** O(1)

  <summary><span style="font-size:19px;"><strong>Shortest Subarray with Sum at Least K - </strong><a href="https://leetcode.com/problems/shortest-subarray-with-sum-at-least-k/" target="_blank">#862</a></span></summary>

### 🧾 Input

* `nums`: List of integers
* `k`: Integer target sum

### 📤 Output

* Length of the shortest non-empty subarray with sum at least `k`, or -1 if none

### ✅ Example

```py
Input: nums = [1,2,3,4,5], k = 11
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import deque

class Solution:
    def shortestSubarray(self, nums: List[int], k: int) -> int:
        n = len(nums)
        prefix = [0] * (n + 1)
        for i in range(n):
            prefix[i + 1] = prefix[i] + nums[i]

        res = n + 1
        dq = deque()

        for i in range(n + 1):
            while dq and prefix[i] - prefix[dq[0]] >= k:
                res = min(res, i - dq.popleft())
            while dq and prefix[i] <= prefix[dq[-1]]:
                dq.pop()
            dq.append(i)

        return res if res <= n else -1
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(n)
**Space Complexity:** O(n)

  <summary><span style="font-size:19px;"><strong>Word Subsets - </strong><a href="https://leetcode.com/problems/word-subsets/" target="_blank">#916</a></span></summary>

### 🧾 Input

* `words1`, `words2`: Lists of strings

### 📤 Output

* List of universal words from `words1` that are supersets of all words in `words2`

### ✅ Example

```py
Input: words1 = ["amazon","apple","facebook","google","leetcode"], 
       words2 = ["e","o"]
Output: ["facebook","google","leetcode"]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import Counter

class Solution:
    def wordSubsets(self, words1: List[str], words2: List[str]) -> List[str]:
        max_count = Counter()
        for word in words2:
            count = Counter(word)
            for char in count:
                max_count[char] = max(max_count[char], count[char])

        def is_universal(word):
            count = Counter(word)
            for char in max_count:
                if count[char] < max_count[char]:
                    return False
            return True

        return [word for word in words1 if is_universal(word)]
```

</details>

  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(n \* k), where n = len(words1), k = average word length
**Space Complexity:** O(1) (26 letters)

  <summary><span style="font-size:19px;"><strong>Split Array with Equal Sum - </strong><a href="https://leetcode.com/problems/split-array-with-equal-sum/" target="_blank">#548</a></span></summary>

### 🧾 Input

* `nums`: List of integers

### 📤 Output

* Boolean: can you split array into 4 parts with equal sums

### ✅ Example

```py
Input: nums = [1,2,1,2,1,2,1]
Output: True
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def splitArray(self, nums: List[int]) -> bool:
        n = len(nums)
        if n < 7:
            return False
        for j in range(3, n - 3):
            s = set()
            left_sum = 0
            for i in range(1, j - 1):
                left_sum += nums[i - 1]
                if left_sum == sum(nums[i + 1:j]) - nums[j - 1]:
                    s.add(left_sum)
            right_sum = 0
            for k in range(n - 2, j + 1, -1):
                right_sum += nums[k + 1] if k + 1 < n else 0
                if right_sum == sum(nums[j + 1:k]) - nums[k]:
                    if right_sum in s:
                        return True
        return False
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(n²)
**Space Complexity:** O(n)

  <summary><span style="font-size:19px;"><strong>Longest Increasing Path in a Matrix - </strong><a href="https://leetcode.com/problems/longest-increasing-path-in-a-matrix/" target="_blank">#329</a></span></summary>

### 🧾 Input

* `matrix`: 2D list of integers

### 📤 Output

* Length of the longest increasing path in the matrix

### ✅ Example

```py
Input: matrix = [[9,9,4],[6,6,8],[2,1,1]]
Output: 4
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def longestIncreasingPath(self, matrix: List[List[int]]) -> int:
        if not matrix:
            return 0
        m, n = len(matrix), len(matrix[0])
        memo = [[0] * n for _ in range(m)]

        def dfs(x, y):
            if memo[x][y]:
                return memo[x][y]
            val = matrix[x][y]
            best = 1
            for dx, dy in [(-1,0), (1,0), (0,-1), (0,1)]:
                nx, ny = x + dx, y + dy
                if 0 <= nx < m and 0 <= ny < n and matrix[nx][ny] > val:
                    best = max(best, 1 + dfs(nx, ny))
            memo[x][y] = best
            return best

        return max(dfs(i, j) for i in range(m) for j in range(n))
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(m \* n)
**Space Complexity:** O(m \* n)

</details>
