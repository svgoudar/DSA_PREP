<details>
  <summary><span style="font-size:19px;"><strong>Linear Search</strong></span></summary>

### 🧾 Input

* `arr`: List of elements
* `target`: Element to search

### 📤 Output

* Index of target element if found, else -1

### ✅ Example

```python
Input: arr = [10, 20, 80, 30, 60], target = 30  
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def linearSearch(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)  
**Space:** O(1)

</details>
</details>


<details>
  <summary><span style="font-size:19px;"><strong>Binary Search</strong></span></summary>

### 🧾 Input

* `arr`: Sorted list of elements
* `target`: Element to search

### 📤 Output

* Index of target element if found, else -1

### ✅ Example

```python
Input: arr = [1, 3, 5, 7, 9, 11], target = 7  
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def binarySearch(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>
**Time:** O(log n)  
**Space:** O(1)
</details>
</details>

<details>
  <summary><span style="font-size:19px;"><strong>Ternary Search</strong></span></summary>

### 🧾 Input

* `arr`: Sorted list of elements (unimodal)
* `target`: Element to search

### 📤 Output

* Index of target element if found, else -1

### ✅ Example

```python
Input: arr = [1, 3, 5, 7, 9, 11, 13], target = 11  
Output: 5
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def ternarySearch(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid1 = low + (high - low) // 3
        mid2 = high - (high - low) // 3

        if arr[mid1] == target:
            return mid1
        if arr[mid2] == target:
            return mid2

        if target < arr[mid1]:
            high = mid1 - 1
        elif target > arr[mid2]:
            low = mid2 + 1
        else:
            low = mid1 + 1
            high = mid2 - 1
    return -1
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(log₃ n)  
**Space:** O(1)

</details>
</details>

<details>
  <summary><span style="font-size:19px;"><strong>Exponential Search</strong></span></summary>

### 🧾 Input

* `arr`: Sorted list of elements
* `target`: Element to search

### 📤 Output

* Index of target element if found, else -1

### ✅ Example

```python
Input: arr = [2, 3, 4, 10, 40, 55, 70, 85], target = 10  
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def binarySearch(arr, low, high, target):
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

def exponentialSearch(arr, target):
    if arr[0] == target:
        return 0

    i = 1
    while i < len(arr) and arr[i] <= target:
        i *= 2

    return binarySearch(arr, i // 2, min(i, len(arr)-1), target)
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(log i), where i is the index of the first element greater than or equal to target  
**Space:** O(1)

</details>

<details>
  <summary><span style="font-size:19px;"><strong>Interpolation Search</strong></span></summary>

### 🧾 Input

* `arr`: Sorted list of uniformly distributed integers
* `target`: Element to search

### 📤 Output

* Index of target element if found, else -1

### ✅ Example

```python
Input: arr = [10, 20, 30, 40, 50], target = 30  
Output: 2
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def interpolationSearch(arr, target):
    low, high = 0, len(arr) - 1

    while low <= high and target >= arr[low] and target <= arr[high]:
        if low == high:
            if arr[low] == target:
                return low
            return -1

        pos = low + ((target - arr[low]) * (high - low)) // (arr[high] - arr[low])

        if arr[pos] == target:
            return pos
        elif arr[pos] < target:
            low = pos + 1
        else:
            high = pos - 1
    return -1
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(log log n) — best case (uniform distribution)  
**Worst Case:** O(n)  
**Space:** O(1)

</details>
</details>
