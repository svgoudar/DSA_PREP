  <summary><span style="font-size:19px;"><strong>1. Reverse a Singly Linked List</strong></span></summary>

### 🧾 Input

* `head`: Head node of the singly linked list

### 📤 Output

* Head node of the reversed list

### ✅ Example

```python
Input: 1 → 2 → 3 → 4 → 5  
Output: 5 → 4 → 3 → 2 → 1
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def reverseList(head):
    prev = None
    curr = head
    while curr:
        nxt = curr.next
        curr.next = prev
        prev = curr
        curr = nxt
    return prev
```

</details>

**Time:** O(n)  
**Space:** O(1)

<details>
  <summary><span style="font-size:19px;"><strong>2. Merge Two Sorted Linked Lists</strong></span></summary>

### 🧾 Input

* Two sorted linked lists `l1` and `l2`

### 📤 Output

* A merged sorted linked list

### ✅ Example

```python
Input: l1 = 1→2→4, l2 = 1→3→4  
Output: 1→1→2→3→4→4
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def mergeTwoLists(l1, l2):
    dummy = ListNode(0)
    curr = dummy
    while l1 and l2:
        if l1.val < l2.val:
            curr.next = l1
            l1 = l1.next
        else:
            curr.next = l2
            l2 = l2.next
        curr = curr.next
    curr.next = l1 or l2
    return dummy.next
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n + m)  
**Space:** O(1)

</details>

  <summary><span style="font-size:19px;"><strong>3. Remove N-th Node From End of List</strong></span></summary>

### 🧾 Input

* `head`: Head of the linked list
* `n`: Position from the end

### 📤 Output

* Head of the updated linked list

### ✅ Example

```python
Input: 1→2→3→4→5, n = 2  
Output: 1→2→3→5
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def removeNthFromEnd(head, n):
    dummy = ListNode(0, head)
    slow = fast = dummy
    for _ in range(n):
        fast = fast.next
    while fast.next:
        slow = slow.next
        fast = fast.next
    slow.next = slow.next.next
    return dummy.next
```

</details>

**Time:** O(n)

<details>
  <summary><span style="font-size:19px;"><strong>5. Palindrome Linked List</strong></span></summary>

### 🧾 Input

* `head`: Head of the linked list

### 📤 Output

* Boolean: True if it's a palindrome, else False

### ✅ Example

```python
Input: 1 → 2 → 2 → 1  
Output: True
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def isPalindrome(head):
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    
    prev = None
    while slow:
        nxt = slow.next
        slow.next = prev
        prev = slow
        slow = nxt

    left, right = head, prev
    while right:
        if left.val != right.val:
            return False
        left = left.next
        right = right.next
    return True
```

</details>

**Time:** O(n)  
**Space:** O(1)

  <summary><span style="font-size:19px;"><strong>6. Remove Duplicates from Sorted Linked List</strong></span></summary>

* `head`: Head of a **sorted** linked list

### ✅ Example

```python
Input: 1 → 1 → 2 → 3 → 3  
Output: 1 → 2 → 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def deleteDuplicates(head):
    curr = head
    while curr and curr.next:
        if curr.val == curr.next.val:
            curr.next = curr.next.next
        else:
            curr = curr.next
    return head
```

  <summary><span style="font-size:19px;"><strong>7. Intersection of Two Linked Lists</strong></span></summary>

### 🧾 Input

* `headA`, `headB`: Two singly linked lists

### 📤 Output

* The intersecting node (by reference), or None if they don’t intersect

### ✅ Example

```python
Input: A = 4→1→8→4→5, B = 5→6→1→8→4→5  
Output: Reference to node with value 8
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def getIntersectionNode(headA, headB):
    if not headA or not headB:
        return None
    a, b = headA, headB
    while a != b:
        a = a.next if a else headB
        b = b.next if b else headA
    return a
```

</details>

**Time:** O(n + m)

  <summary><span style="font-size:19px;"><strong>8. Linked List Cycle Detection (Floyd’s Algorithm)</strong></span></summary>

### 🧾 Input

* `head`: Head of the linked list

### 📤 Output

* `True` if there's a cycle, else `False`

### ✅ Example

```python
Input: 3 → 2 → 0 → -4 (tail connects to node 1)  
Output: True
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def hasCycle(head):
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return True
    return False
```

  <summary><span style="font-size:19px;"><strong>9. Delete All Nodes That Have Duplicate Numbers</strong></span></summary>

### 🧾 Input

* `head`: Head of a **sorted** linked list

### 📤 Output

* Linked list with all duplicates completely removed

### ✅ Example

```python
Input: 1 → 2 → 3 → 3 → 4 → 4 → 5  
Output: 1 → 2 → 5
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def deleteDuplicates(head):
    dummy = ListNode(0, head)
    prev = dummy
    while head:
        if head.next and head.val == head.next.val:
            while head.next and head.val == head.next.val:
                head = head.next
            prev.next = head.next
        else:
            prev = prev.next
        head = head.next
    return dummy.next
```

  <summary><span style="font-size:19px;"><strong>10. Odd-Even Linked List</strong></span></summary>

### 🧾 Input

* `head`: Head of the linked list

### 📤 Output

* Rearranged list grouping odd-indexed nodes followed by even-indexed nodes

### ✅ Example

```python
Input: 1 → 2 → 3 → 4 → 5  
Output: 1 → 3 → 5 → 2 → 4
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def oddEvenList(head):
    if not head or not head.next:
        return head
    odd = head
    even = head.next
    even_head = even

    while even and even.next:
        odd.next = even.next
        odd = odd.next
        even.next = odd.next
        even = even.next

    odd.next = even_head
    return head
```

</details>

### 🧾 Input

* `l1`, `l2`: Two non-empty linked lists representing reversed integers

### 📤 Output

* Sum as a linked list in reversed order

### ✅ Example

```python
Input: l1 = [2,4,3], l2 = [5,6,4]
Output: [7,0,8]   # 342 + 465 = 807
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def addTwoNumbers(self, l1, l2):
        dummy = ListNode(0)
        curr, carry = dummy, 0
        while l1 or l2 or carry:
            v1 = l1.val if l1 else 0
            v2 = l2.val if l2 else 0
            carry, digit = divmod(v1 + v2 + carry, 10)
            curr.next = ListNode(digit)
            curr = curr.next
            l1 = l1.next if l1 else None
            l2 = l2.next if l2 else None
        return dummy.next
```

  <summary><span style="font-size:19px;"><strong>Partition List - </strong><a href="https://leetcode.com/problems/partition-list/" target="_blank">#86</a></span></summary>

### 🧾 Input

* `head`: Linked list head
* `x`: Partition value

### 📤 Output

* Reordered list such that nodes < x come before nodes ≥ x

### ✅ Example

```python
Input: head = [1,4,3,2,5,2], x = 3  
Output: [1,2,2,4,3,5]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def partition(self, head, x):
        before = before_tail = ListNode(0)
        after = after_tail = ListNode(0)
        while head:
            if head.val < x:
                before_tail.next = head
                before_tail = head
            else:
                after_tail.next = head
                after_tail = head
            head = head.next
        after_tail.next = None
        before_tail.next = after.next
        return before.next
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>
**Time:** O(n)  
**Space:** O(1)

### 🧾 Input

* `head`: Linked list head
* `k`: Number of positions to rotate right

### 📤 Output

* New head of the rotated list

### ✅ Example

```python
Input: head = [1,2,3,4,5], k = 2  
Output: [4,5,1,2,3]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def rotateRight(self, head, k):
        if not head or not head.next or k == 0:
            return head
        old_tail = head
        length = 1
        while old_tail.next:
            old_tail = old_tail.next
            length += 1
        old_tail.next = head
        k %= length
        new_tail = head
        for _ in range(length - k - 1):
            new_tail = new_tail.next
        new_head = new_tail.next
        new_tail.next = None
        return new_head
```

</details>

### 🧾 Input

* `head`: Head of a linked list where each node has a `random` pointer

### 📤 Output

* Deep copy of the list

### ✅ Example

```python
Input: head = [[7,null],[13,0],[11,4],[10,2],[1,0]]
Output: new deep-copied list
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def copyRandomList(self, head):
        if not head:
            return None
        curr = head
        while curr:
            nxt = curr.next
            curr.next = Node(curr.val)
            curr.next.next = nxt
            curr = nxt
        curr = head
        while curr:
            if curr.random:
                curr.next.random = curr.random.next
            curr = curr.next.next
        curr = head.next
        copy_head = curr
        prev = head
        while prev:
            prev.next = prev.next.next
            curr.next = curr.next.next if curr.next else None
            prev = prev.next
            curr = curr.next
        return copy_head
```

  <summary><span style="font-size:19px;"><strong>Sort List (Merge Sort) - </strong><a href="https://leetcode.com/problems/sort-list/" target="_blank">#148</a></span></summary>

### 🧾 Input

* `head`: Head of an unsorted linked list

### 📤 Output

* Sorted linked list in ascending order

### ✅ Example

```python
Input: head = [4,2,1,3]
Output: [1,2,3,4]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def sortList(self, head):
        if not head or not head.next:
            return head
        prev, slow, fast = None, head, head
        while fast and fast.next:
            prev, slow, fast = slow, slow.next, fast.next.next
        prev.next = None
        l1 = self.sortList(head)
        l2 = self.sortList(slow)
        return self.merge(l1, l2)

    def merge(self, l1, l2):
        dummy = curr = ListNode(0)
        while l1 and l2:
            if l1.val < l2.val:
                curr.next, l1 = l1, l1.next
            else:
                curr.next, l2 = l2, l2.next
            curr = curr.next
        curr.next = l1 or l2
        return dummy.next
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n log n)  
**Space:** O(log n) recursion for merges

</details>

  <summary><span style="font-size:19px;"><strong>Flatten Multilevel Doubly Linked List - </strong><a href="https://leetcode.com/problems/flatten-a-multilevel-doubly-linked-list/" target="_blank">#430</a></span></summary>

### 🧾 Input

* `head`: Head of multilevel doubly linked list with `child` pointers

### 📤 Output

* Flattened single-level doubly linked list

### ✅ Example

```python
Input: head = [1,2,3,4,5,6,null,null,null,7,8,9,10,null,null,11,12]
Output: [1,2,3,7,8,11,12,9,10,4,5,6]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def flatten(self, head):
        if not head:
            return head
        pseudo = prev = Node(0, None, head, None)
        stack = [head]
        while stack:
            curr = stack.pop()
            prev.next = curr
            curr.prev = prev
            if curr.next:
                stack.append(curr.next)
            if curr.child:
                stack.append(curr.child)
                curr.child = None
            prev = curr
        pseudo.next.prev = None
        return pseudo.next
```

</details>

  <summary><strong>Complexity Analysis</strong></summary>
**Time:** O(n)  
**Space:** O(n) for stack

  <summary><span style="font-size:19px;"><strong>Reverse Nodes in k-Group - </strong><a href="https://leetcode.com/problems/reverse-nodes-in-k-group/" target="_blank">#25</a></span></summary>

### 🧾 Input

* `head`: Linked list head
* `k`: Group size to reverse

### 📤 Output

* Reversed list in groups of k nodes

### ✅ Example

```python
Input: head = [1,2,3,4,5], k = 2  
Output: [2,1,4,3,5]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def reverseKGroup(self, head, k):
        def get_kth(curr, k):
            while curr and k:
                curr, k = curr.next, k - 1
            return curr

        dummy = ListNode(0)
        dummy.next = head
        group_prev = dummy

        while True:
            kth = get_kth(group_prev, k)
            if not kth:
                break
            group_next = kth.next

            prev, curr = kth.next, group_prev.next
            while curr != group_next:
                tmp = curr.next
                curr.next = prev
                prev = curr
                curr = tmp

            tmp = group_prev.next
            group_prev.next = kth
            group_prev = tmp

        return dummy.next
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>
**Time:** O(n)  
**Space:** O(1)
</details>

### 🧾 Input

* `head`: Linked list

### 📤 Output

* Remove nodes where a greater node appears later in list

### ✅ Example

```python
Input: head = [5,2,13,3,8]  
Output: [13,8]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def removeNodes(self, head):
        def reverse(head):
            prev = None
            while head:
                head.next, prev, head = prev, head, head.next
            return prev

        head = reverse(head)
        max_val = head.val
        curr = head
        while curr and curr.next:
            if curr.next.val < max_val:
                curr.next = curr.next.next
            else:
                curr = curr.next
                max_val = curr.val
        return reverse(head)
```

### 🧾 Input

* `head`: Linked list

### 📤 Output

* Reorder to: `L0 → Ln → L1 → Ln-1 → ...`

### ✅ Example

```python
Input: head = [1,2,3,4]  
Output: [1,4,2,3]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def reorderList(self, head):
        if not head:
            return None

        # Step 1: Find middle
        slow, fast = head, head
        while fast and fast.next:
            slow, fast = slow.next, fast.next.next

        # Step 2: Reverse second half
        prev, curr = None, slow.next
        slow.next = None
        while curr:
            curr.next, prev, curr = prev, curr, curr.next

        # Step 3: Merge two halves
        first, second = head, prev
        while second:
            tmp1, tmp2 = first.next, second.next
            first.next = second
            second.next = tmp1
            first, second = tmp1, tmp2
```

</details>

**Time:** O(n)  

**Space:** O(1)

  <summary><span style="font-size:19px;"><strong>Split Linked List in Parts - </strong><a href="https://leetcode.com/problems/split-linked-list-in-parts/" target="_blank">#725</a></span></summary>

### 🧾 Input

* `head`: Linked list
* `k`: Number of parts to split into

### 📤 Output

* `k` parts with nearly equal length

### ✅ Example

```python
Input: head = [1,2,3], k = 5  
Output: [[1],[2],[3],[],[]]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def splitListToParts(self, head, k):
        length = 0
        curr = head
        while curr:
            length += 1
            curr = curr.next

        part_size, extra = divmod(length, k)
        res = []
        curr = head

        for _ in range(k):
            dummy = ListNode(0)
            part_head = dummy
            for _ in range(part_size + (extra > 0)):
                if curr:
                    part_head.next = curr
                    curr = curr.next
                    part_head = part_head.next
            if part_head:
                part_head.next = None
            res.append(dummy.next)
            extra -= 1
        return res
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)  
**Space:** O(k)

</details>

  <summary><span style="font-size:19px;"><strong>Merge k Sorted Lists - </strong><a href="https://leetcode.com/problems/merge-k-sorted-lists/" target="_blank">#23</a></span></summary>

### 🧾 Input

* `lists`: List of linked list heads

### 📤 Output

* One merged sorted linked list

### ✅ Example

```python
Input: lists = [[1,4,5],[1,3,4],[2,6]]  
Output: [1,1,2,3,4,4,5,6]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
import heapq

class Solution:
    def mergeKLists(self, lists):
        heap = []
        for i, node in enumerate(lists):
            if node:
                heapq.heappush(heap, (node.val, i, node))

        dummy = ListNode(0)
        curr = dummy

        while heap:
            val, i, node = heapq.heappop(heap)
            curr.next = node
            curr = curr.next
            if node.next:
                heapq.heappush(heap, (node.next.val, i, node.next))

        return dummy.next
```

</details>

**Time:** O(N log k), where N = total nodes, k = number of lists  
**Space:** O(k)

### 🧾 Input

* `head`: Linked list
* `m`, `n`: Positions to reverse between (1-based)

### 📤 Output

* Modified list with nodes between m and n reversed

### ✅ Example

```python
Input: head = [1,2,3,4,5], m = 2, n = 4  
Output: [1,4,3,2,5]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def reverseBetween(self, head, m, n):
        dummy = ListNode(0)
        dummy.next = head
        prev = dummy

        for _ in range(m - 1):
            prev = prev.next

        reverse = None
        curr = prev.next

        for _ in range(n - m + 1):
            next_temp = curr.next
            curr.next = reverse
            reverse = curr
            curr = next_temp

        prev.next.next = curr
        prev.next = reverse

        return dummy.next
```

  <summary><span style="font-size:19px;"><strong>Find Maximum Twin Sum of Linked List - </strong><a href="https://leetcode.com/problems/maximum-twin-sum-of-a-linked-list/" target="_blank">#2130</a></span></summary>

### 🧾 Input

* `head`: Even-length linked list

### 📤 Output

* Maximum twin sum (i-th + n-1-i-th)

### ✅ Example

```python
Input: head = [5,4,2,1]  
Output: 6
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def pairSum(self, head):
        slow = fast = head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

        prev = None
        while slow:
            slow.next, prev, slow = prev, slow, slow.next

        max_sum = 0
        while prev:
            max_sum = max(max_sum, head.val + prev.val)
            head, prev = head.next, prev.next
        return max_sum
```

### 🧾 Input

* `head`: Head of the singly linked list

### 📤 Output

* Return `True` if the linked list is a palindrome

### ✅ Example

```python
Input: head = [1,2,2,1]
Output: True
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def isPalindrome(self, head):
        slow = fast = head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

        prev = None
        while slow:
            slow.next, prev, slow = prev, slow, slow.next

        while prev:
            if prev.val != head.val:
                return False
            prev = prev.next
            head = head.next
        return True
```

**Time:** O(n)  
**Space:** O(1)

  <summary><span style="font-size:19px;"><strong>Add Two Numbers II - </strong><a href="https://leetcode.com/problems/add-two-numbers-ii/" target="_blank">#445</a></span></summary>

### 🧾 Input

* `l1`, `l2`: Two non-reversed linked lists representing two numbers

### 📤 Output

* Sum as a linked list (most significant digit first)

### ✅ Example

```python
Input: l1 = [7,2,4,3], l2 = [5,6,4]
Output: [7,8,0,7]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def addTwoNumbers(self, l1, l2):
        stack1, stack2 = [], []
        while l1:
            stack1.append(l1.val)
            l1 = l1.next
        while l2:
            stack2.append(l2.val)
            l2 = l2.next

        carry = 0
        head = None

        while stack1 or stack2 or carry:
            x = stack1.pop() if stack1 else 0
            y = stack2.pop() if stack2 else 0
            total = x + y + carry
            carry = total // 10
            node = ListNode(total % 10)
            node.next = head
            head = node

        return head
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n + m)  

**Space:** O(n + m)

  <summary><span style="font-size:19px;"><strong>Copy Linked List with Random Pointer (O(1) space) - </strong><a href="https://leetcode.com/problems/copy-list-with-random-pointer/" target="_blank">#138</a></span></summary>

### 🧾 Input

* `head`: A linked list where each node has an additional random pointer

### 📤 Output

* A deep copy of the list

### ✅ Example

```python
Input: head = [[7,null],[13,0],[11,4],[10,2],[1,0]]
Output: [[7,null],[13,0],[11,4],[10,2],[1,0]]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def copyRandomList(self, head):
        if not head:
            return None

        curr = head
        while curr:
            node = Node(curr.val, curr.next, None)
            curr.next = node
            curr = node.next

        curr = head
        while curr:
            if curr.random:
                curr.next.random = curr.random.next
            curr = curr.next.next

        pseudo_head = Node(0)
        copy = pseudo_head
        curr = head

        while curr:
            copy.next = curr.next
            curr.next = curr.next.next
            curr = curr.next
            copy = copy.next

        return pseudo_head.next
```

**Time:** O(n)

  <summary><span style="font-size:19px;"><strong>Sort List in O(n log n) Time and O(1) Space - </strong><a href="https://leetcode.com/problems/sort-list/" target="_blank">#148</a></span></summary>

### 🧾 Input

* `head`: Head of the unsorted linked list

### 📤 Output

* Sorted linked list in ascending order

### ✅ Example

```python
Input: head = [4,2,1,3]
Output: [1,2,3,4]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def sortList(self, head):
        if not head or not head.next:
            return head

        slow, fast = head, head.next
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

        mid = slow.next
        slow.next = None

        left = self.sortList(head)
        right = self.sortList(mid)

        return self.merge(left, right)

    def merge(self, l1, l2):
        dummy = tail = ListNode(0)
        while l1 and l2:
            if l1.val < l2.val:
                tail.next, l1 = l1, l1.next
            else:
                tail.next, l2 = l2, l2.next
            tail = tail.next

        tail.next = l1 or l2
        return dummy.next
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>
**Time:** O(n log n)  
**Space:** O(log n) recursion (but no extra structures)
</details>

  <summary><span style="font-size:19px;"><strong>Split Linked List into Descending Sequences - </strong><a href="https://leetcode.com/problems/split-linked-list-in-parts/" target="_blank">Custom</a></span></summary>

### 🧾 Input

* `head`: Head of the linked list

### 📤 Output

* Return parts where each sublist is strictly descending from left to right

### ✅ Example

```python
Input: head = [9,8,7,10,5,3,1,6]
Output: [[9,8,7],[10,5,3,1],[6]]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def splitDescending(self, head):
        res = []
        while head:
            dummy = ListNode(0)
            tail = dummy
            while head and (tail == dummy or tail.val > head.val):
                tail.next = head
                tail = tail.next
                head = head.next
            tail.next = None
            res.append(dummy.next)
        return res
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)  

**Space:** O(1) (excluding result array)

  <summary><span style="font-size:19px;"><strong>Reconstruct Binary Tree from Linked List (with child pointers) - </strong><a href="https://leetcode.com/problems/clone-binary-tree-with-random-pointer/" target="_blank">Custom Variant</a></span></summary>

### 🧾 Input

* A linked list where each node contains:

  * `val`: value of the node
  * `left`: pointer to left child (as another node)
  * `right`: pointer to right child

This is more like a **binary tree** represented via linked structure (like N-ary trees in leetcode with pointers). The reconstruction means creating a full tree by cloning structure and values.

### 📤 Output

* A root of the cloned binary tree

### ✅ Example

```python
Input: head (root of binary tree with left/right pointers)
Output: deep clone of the entire binary tree
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Node:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def cloneBinaryTree(node, visited=None):
    if not node:
        return None
    if visited is None:
        visited = {}
    if node in visited:
        return visited[node]
    
    new_node = Node(node.val)
    visited[node] = new_node
    new_node.left = cloneBinaryTree(node.left, visited)
    new_node.right = cloneBinaryTree(node.right, visited)
    return new_node
```

  <summary><strong>Complexity Analysis</strong></summary>
**Time:** O(n) (every node cloned once)  
**Space:** O(n) (due to recursion + visited map)
</details>

<details>
  <summary><span style="font-size:19px;"><strong>LRU Cache (Doubly Linked List + Hash Map) - </strong><a href="https://leetcode.com/problems/lru-cache/" target="_blank">#146</a></span></summary>

### 🧾 Input

* Capacity `k`, then perform `get(key)` and `put(key, value)` operations

### 📤 Output

* `get(key)` returns value if present, else -1
* Cache evicts least-recently-used entry if capacity exceeded

### ✅ Example

```python
Input: ["LRUCache","put","put","get","put","get","get"]
       [[2],[1,1],[2,2],[1],[3,3],[2],[3]]
Output: [null,null,null,1,null,-1,3]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class DLinkedNode:
    def __init__(self, key=0, val=0):
        self.key = key
        self.val = val
        self.prev = self.next = None

class LRUCache:
    def __init__(self, capacity: int):
        self.cache = {}  # key: node
        self.capacity = capacity
        self.size = 0
        self.head, self.tail = DLinkedNode(), DLinkedNode()
        self.head.next = self.tail
        self.tail.prev = self.head

    def _add_node(self, node):
        node.prev = self.head
        node.next = self.head.next
        self.head.next.prev = node
        self.head.next = node

    def _remove_node(self, node):
        prev, nxt = node.prev, node.next
        prev.next = nxt
        nxt.prev = prev

    def _move_to_front(self, node):
        self._remove_node(node)
        self._add_node(node)

    def _pop_tail(self):
        node = self.tail.prev
        self._remove_node(node)
        return node

    def get(self, key: int) -> int:
        node = self.cache.get(key)
        if not node:
            return -1
        self._move_to_front(node)
        return node.val

    def put(self, key: int, value: int):
        node = self.cache.get(key)
        if node:
            node.val = value
            self._move_to_front(node)
        else:
            newNode = DLinkedNode(key, value)
            self.cache[key] = newNode
            self._add_node(newNode)
            self.size += 1

            if self.size > self.capacity:
                tail = self._pop_tail()
                del self.cache[tail.key]
                self.size -= 1
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(1) for both `get` and `put` operations  
**Space:** O(capacity)

</details>
