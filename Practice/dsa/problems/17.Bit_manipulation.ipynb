  <summary><span style="font-size:19px;"><strong>Single Number - </strong><a href="https://leetcode.com/problems/single-number/" target="_blank">#136</a></span></summary>

### 🧾 Input
* List of integers where every element appears twice except one

### 📤 Output
* The single number that appears once

### ✅ Example
```py
Input: [4,1,2,1,2]
Output: 4
````

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def singleNumber(self, nums: List[int]) -> int:
        res = 0
        for num in nums:
            res ^= num
        return res
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(1)

</details>
</details>

<details>
  <summary><span style="font-size:19px;"><strong>Number of 1 Bits - </strong><a href="https://leetcode.com/problems/number-of-1-bits/" target="_blank">#191</a></span></summary>

### 🧾 Input
* An unsigned integer

### 📤 Output
* The number of `1` bits (Hamming weight)

### ✅ Example
```py
Input: 00000000000000000000000000001011
Output: 3
````

```python
        count = 0
        while n:

            n >>= 1
```

### 🧾 Input
* Array of integers `arr`

### 📤 Output
* `True` if there exists `N` and `M` such that `N = 2 * M`

### ✅ Example
```py
Input: arr = [10,2,5,3]
Output: True
````

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def checkIfExist(self, arr: List[int]) -> bool:
        seen = set()
        for num in arr:
            if 2 * num in seen or (num % 2 == 0 and num // 2 in seen):
                return True
            seen.add(num)
        return False
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(n)

</details>
</details>

<details>
  <summary><span style="font-size:19px;"><strong>Find the Difference - </strong><a href="https://leetcode.com/problems/find-the-difference/" target="_blank">#389</a></span></summary>

### 🧾 Input
* Two strings `s` and `t` (where `t` is `s` with one extra letter)

### 📤 Output
* The extra character in `t`

### ✅ Example
```py
Input: s = "abcd", t = "abcde"
Output: "e"
````

  <summary><span style="font-size:19px;"><strong>Sum of Two Integers - </strong><a href="https://leetcode.com/problems/sum-of-two-integers/" target="_blank">#371</a></span></summary>

### 🧾 Input
* Two integers `a` and `b`

### 📤 Output
* Sum of `a` and `b` without using `+` or `-`

### ✅ Example
```py
Input: a = 1, b = 2
Output: 3
````

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def getSum(self, a: int, b: int) -> int:
        MAX = 0x7FFFFFFF
        mask = 0xFFFFFFFF
        while b != 0:
            a, b = (a ^ b) & mask, ((a & b) << 1) & mask
        return a if a <= MAX else ~(a ^ mask)
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(1)
**Space:** O(1)

</details>
</details>

<details>
  <summary><span style="font-size:19px;"><strong>XOR Operation in an Array - </strong><a href="https://leetcode.com/problems/xor-operation-in-an-array/" target="_blank">#1486</a></span></summary>

### 🧾 Input
* Two integers: `n` and `start`

### 📤 Output
* XOR of all elements: `start + 2*i` for `i` from `0` to `n-1`

### ✅ Example
```py
Input: n = 5, start = 0
Output: 8
````

### ✅ Example

```py
Output: 3
````

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def singleNumber(self, nums: List[int]) -> int:
        ones = twos = 0
        for num in nums:
            ones = (ones ^ num) & ~twos
            twos = (twos ^ num) & ~ones
        return ones
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(1)

</details>
</details>

<details>
  <summary><span style="font-size:19px;"><strong>Bitwise AND of Numbers Range - </strong><a href="https://leetcode.com/problems/bitwise-and-of-numbers-range/" target="_blank">#201</a></span></summary>

### 🧾 Input
* Two integers `left` and `right`

### 📤 Output
* Bitwise AND of all numbers between `left` and `right` inclusive

### ✅ Example
```py
Input: left = 5, right = 7
Output: 4
````

### 🧾 Input
* Integer `n`

### 📤 Output
* Array where each element `i` is the count of 1s in binary of `i`

### ✅ Example
```py
Input: 5
Output: [0,1,1,2,1,2]
````

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def countBits(self, n: int) -> List[int]:
        dp = [0] * (n + 1)
        for i in range(1, n + 1):
            dp[i] = dp[i >> 1] + (i & 1)
        return dp
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(n)

</details>
</details>

<details>
  <summary><span style="font-size:19px;"><strong>Maximum Product of Word Lengths - </strong><a href="https://leetcode.com/problems/maximum-product-of-word-lengths/" target="_blank">#318</a></span></summary>

### 🧾 Input
* List of strings

### 📤 Output
* Max product of lengths of two words that share no common letters

### ✅ Example
```py
Input: ["abcw","baz","foo","bar","xtfn","abcdef"]
Output: 16
````

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def maxProduct(self, words: List[str]) -> int:
        n = len(words)
        masks = [0] * n
        for i in range(n):
            for c in words[i]:
                masks[i] |= 1 << (ord(c) - ord('a'))

        max_prod = 0
        for i in range(n):
            for j in range(i+1, n):
                if masks[i] & masks[j] == 0:
                    max_prod = max(max_prod, len(words[i]) * len(words[j]))
        return max_prod
```

```python
        flips = 0

            abit = (a >> i) & 1
            bbit = (b >> i) & 1
```

  <summary><span style="font-size:19px;"><strong>Maximum XOR of Two Numbers in an Array - </strong><a href="https://leetcode.com/problems/maximum-xor-of-two-numbers-in-an-array/" target="_blank">#421</a></span></summary>

### 🧾 Input
* List of integers `nums`

### 📤 Output
* Maximum XOR of any two elements

### ✅ Example
```py
Input: [3,10,5,25,2,8]
Output: 28
````

<details>
  <summary><strong>Solution Code (Trie-based)</strong></summary>

```python
class TrieNode:
    def __init__(self):
        self.children = {}

class Solution:
    def findMaximumXOR(self, nums: List[int]) -> int:
        root = TrieNode()
        for num in nums:
            node = root
            for i in range(31, -1, -1):
                bit = (num >> i) & 1
                if bit not in node.children:
                    node.children[bit] = TrieNode()
                node = node.children[bit]

        max_xor = 0
        for num in nums:
            node = root
            curr_xor = 0
            for i in range(31, -1, -1):
                bit = (num >> i) & 1
                opp = 1 - bit
                if opp in node.children:
                    curr_xor |= (1 << i)
                    node = node.children[opp]
                else:
                    node = node.children[bit]
            max_xor = max(max_xor, curr_xor)
        return max_xor
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n \* 32)
**Space:** O(n \* 32)

</details>
</details>

<details>
  <summary><span style="font-size:19px;"><strong>XOR Queries of a Subarray - </strong><a href="https://leetcode.com/problems/xor-queries-of-a-subarray/" target="_blank">#1310</a></span></summary>

### 🧾 Input
* Array `arr` and list of `queries`

### 📤 Output
* XOR of subarrays specified by each query

### ✅ Example
```py
Input: arr = [1,3,4,8], queries = [[0,1],[1,2],[0,3],[3,3]]
Output: [2,7,14,8]
````

### 🧾 Input

* Count of triplets (i, j, k) such that `arr[i]^...^arr[j-1] == arr[j]^...^arr[k]`

### ✅ Example
```py
Input: arr = [2,3,1,6,7]
Output: 4
````

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def countTriplets(self, arr: List[int]) -> int:
        count = 0
        for i in range(len(arr)):
            xor = 0
            for j in range(i, len(arr)):
                xor ^= arr[j]
                if xor == 0 and j > i:
                    count += j - i
        return count
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n^2)
**Space:** O(1)

</details>
</details>

<details>
  <summary><span style="font-size:19px;"><strong>Binary Trees With Factors - </strong><a href="https://leetcode.com/problems/binary-trees-with-factors/" target="_blank">#823</a></span></summary>

### 🧾 Input
* Array of unique integers

### 📤 Output
* Count of binary trees where each node = product of its children (mod 1e9+7)

### ✅ Example
```py
Input: [2,4]
Output: 3
````

```python
class Solution:
    def numFactoredBinaryTrees(self, arr: List[int]) -> int:
        MOD = 10**9 + 7
        arr.sort()
        dp = {}
        for i, x in enumerate(arr):
            dp[x] = 1
            for j in range(i):
                if x % arr[j] == 0 and x // arr[j] in dp:
                    dp[x] += dp[arr[j]] * dp[x // arr[j]]
```

  <summary><span style="font-size:19px;"><strong>Smallest Sufficient Team - </strong><a href="https://leetcode.com/problems/smallest-sufficient-team/" target="_blank">#1125</a></span></summary>

### 🧾 Input
* Required skills and list of people with skills

### 📤 Output
* Indices of people forming smallest sufficient team

### ✅ Example
```py
Input: req_skills = ["java","nodejs","reactjs"]
       people = [["java"],["nodejs"],["nodejs","reactjs"]]
Output: [0,2]
````

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def smallestSufficientTeam(self, req_skills, people):
        from functools import lru_cache

        skill_id = {v: i for i, v in enumerate(req_skills)}
        people_skills = []
        for person in people:
            mask = 0
            for skill in person:
                mask |= 1 << skill_id[skill]
            people_skills.append(mask)

        n = len(req_skills)

        @lru_cache(None)
        def dp(i, mask):
            if mask == (1 << n) - 1:
                return []
            if i == len(people):
                return None

            # Option 1: skip
            option1 = dp(i+1, mask)
            # Option 2: include
            option2 = dp(i+1, mask | people_skills[i])
            if option2 is not None:
                option2 = [i] + option2

            if option1 is None:
                return option2
            if option2 is None:
                return option1
            return option1 if len(option1) <= len(option2) else option2

        return dp(0, 0)
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(2^n × m)
**Space:** O(2^n)

</details>
</details>


<details>
  <summary><span style="font-size:19px;"><strong>Find the Shortest Superstring - </strong><a href="https://leetcode.com/problems/find-the-shortest-superstring/" target="_blank">#943</a></span></summary>

### 🧾 Input
* List of strings `words`

### 📤 Output
* The shortest string that contains all words as substrings

### ✅ Example
```py
Input: ["alex","loves","leetcode"]
Output: "alexlovesleetcode"
````

```python
    def shortestSuperstring(self, words: List[str]) -> str:
        n = len(words)
        overlap = [[0]*n for _ in range(n)]
        for i, x in enumerate(words):
            for j, y in enumerate(words):
                if i != j:
                    for k in range(min(len(x), len(y)), 0, -1):
                        if x.endswith(y[:k]):
                            overlap[i][j] = k
                            break

        dp = [['']*n for _ in range(1 << n)]
        for i in range(n):
            dp[1 << i][i] = words[i]

        for mask in range(1 << n):
            for j in range(n):
                if not (mask & (1 << j)):
                    continue
                for k in range(n):
                    if j == k or not (mask & (1 << k)):
                        continue
                    cand = dp[mask ^ (1 << j)][k] + words[j][overlap[k][j]:]
                    if dp[mask][j] == '' or len(cand) < len(dp[mask][j]):
                        dp[mask][j] = cand

        full_mask = (1 << n) - 1
        return min((dp[full_mask][i] for i in range(n)), key=len)
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n²·2ⁿ)
**Space:** O(n·2ⁿ)

  <summary><span style="font-size:19px;"><strong>Find the Maximum Number of Marked Indices - </strong><a href="https://leetcode.com/problems/find-the-maximum-number-of-marked-indices/" target="_blank">#2761</a></span></summary>

### 🧾 Input
* List of integers `nums`

### 📤 Output
* Max number of indices such that `2*nums[i] <= nums[j]` for i < j and i ≠ j

### ✅ Example
```py
Input: [3,5,2,4]
Output: 2
````

<details>
  <summary><strong>Solution Code (Greedy + Two Pointers)</strong></summary>

```python
class Solution:
    def maxNumOfMarkedIndices(self, nums: List[int]) -> int:
        nums.sort()
        n = len(nums)
        i, j = 0, n // 2
        count = 0
        while i < n // 2 and j < n:
            if 2 * nums[i] <= nums[j]:
                count += 2
                i += 1
            j += 1
        return count
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n log n)
**Space:** O(1)

</details>
</details>

<details>
  <summary><span style="font-size:19px;"><strong>Minimum XOR Sum of Two Arrays - </strong><a href="https://leetcode.com/problems/minimum-xor-sum-of-two-arrays/" target="_blank">#1879</a></span></summary>

### 🧾 Input
* Two arrays `nums1` and `nums2` of same length

### 📤 Output
* Minimum possible XOR sum by pairing each element uniquely

### ✅ Example
```py
Input: nums1 = [1,2], nums2 = [2,3]
Output: 2
````

<details>
  <summary><strong>Solution Code (DP + Bitmask)</strong></summary>

```python
class Solution:
    def minimumXORSum(self, nums1: List[int], nums2: List[int]) -> int:
        n = len(nums1)
        dp = [float('inf')] * (1 << n)
        dp[0] = 0
        for mask in range(1 << n):
            count = bin(mask).count('1')
            for j in range(n):
                if not mask & (1 << j):
                    new_mask = mask | (1 << j)
                    dp[new_mask] = min(dp[new_mask], dp[mask] + (nums1[count] ^ nums2[j]))
        return dp[-1]
```

  <summary><span style="font-size:19px;"><strong>Maximum AND Sum of Array - </strong><a href="https://leetcode.com/problems/maximum-and-sum-of-array/" target="_blank">#2009</a></span></summary>

### 🧾 Input
* List of integers `nums` and `numSlots`

### 📤 Output
* Max sum of `nums[i] & slot_number` where each slot has ≤ 2 numbers

### ✅ Example
```py
Input: nums = [1,3,10,4,7,1], numSlots = 3
Output: 24
````

<details>
  <summary><strong>Solution Code (DP + Bitmask with Ternary Encoding)</strong></summary>

```python
class Solution:
    def maximumANDSum(self, nums: List[int], numSlots: int) -> int:
        from functools import lru_cache

        @lru_cache(None)
        def dp(i, mask):
            if i == len(nums):
                return 0
            res = 0
            for slot in range(numSlots):
                if (mask >> (slot * 2)) & 3 < 2:
                    new_mask = mask + (1 << (slot * 2))
                    res = max(res, (nums[i] & (slot + 1)) + dp(i + 1, new_mask))
            return res

        return dp(0, 0)
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(3ⁿ)
**Space:** O(3ⁿ)

</details>
</details>

<details>
  <summary><span style="font-size:19px;"><strong>Shortest Path to Get All Keys - </strong><a href="https://leetcode.com/problems/shortest-path-to-get-all-keys/" target="_blank">#864</a></span></summary>

### 🧾 Input
* 2D grid representing walls (`#`), keys (`a-f`), locks (`A-F`), start (`@`)

### 📤 Output
* Minimum steps to collect all keys

### ✅ Example
```py
Input: ["@.a..","###.#","b.A.B"]
Output: 8
````

<details>
  <summary><strong>Solution Code (BFS + Bitmask for Keys)</strong></summary>

```python
from collections import deque

class Solution:
    def shortestPathAllKeys(self, grid: List[str]) -> int:
        m, n = len(grid), len(grid[0])
        keys = 0
        for i in range(m):
            for j in range(n):
                if grid[i][j] == '@':
                    start = (i, j)
                elif grid[i][j].islower():
                    keys |= (1 << (ord(grid[i][j]) - ord('a')))

        queue = deque([(start[0], start[1], 0, 0)])  # x, y, keys, steps
        visited = set()
        while queue:
            x, y, mask, steps = queue.popleft()
            if (x, y, mask) in visited:
                continue
            visited.add((x, y, mask))
            if mask == keys:
                return steps
            for dx, dy in [(0,1),(1,0),(-1,0),(0,-1)]:
                nx, ny = x+dx, y+dy
                if 0 <= nx < m and 0 <= ny < n:
                    cell = grid[nx][ny]
                    if cell == '#':
                        continue
                    if cell.isupper() and not (mask & (1 << (ord(cell) - ord('A')))):
                        continue
                    new_mask = mask
                    if cell.islower():
                        new_mask |= (1 << (ord(cell) - ord('a')))
                    queue.append((nx, ny, new_mask, steps + 1))
        return -1
```

**Time:** O(m·n·2ᵏ)
**Space:** O(m·n·2ᵏ)
