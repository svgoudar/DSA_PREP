### 🧾 Input

* `numbers`: List of integers sorted in non-decreasing order
* `target`: Integer target sum

### 📤 Output

* Indices of the two numbers such that they add up to `target`

### ✅ Example

```py
Input: numbers = [2,7,11,15], target = 9
Output: [1,2]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def twoSum(self, numbers: List[int], target: int) -> List[int]:
        left, right = 0, len(numbers) - 1
        while left < right:
            total = numbers[left] + numbers[right]
            if total == target:
                return [left + 1, right + 1]
            elif total < target:
                left += 1
            else:
                right -= 1
```

</details>

**Time:** O(n)
**Space:** O(1)

### 🧾 Input

* `nums1`: List of integers
* `nums2`: List of integers

### 📤 Output

* Intersection of two arrays including duplicates

### ✅ Example

```py
Input: nums1 = [1,2,2,1], nums2 = [2,2]
Output: [2,2]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import Counter

class Solution:
    def intersect(self, nums1: List[int], nums2: List[int]) -> List[int]:
        counts = Counter(nums1)
        result = []
        for num in nums2:
            if counts[num] > 0:
                result.append(num)
                counts[num] -= 1
        return result
```

</details>

**Time:** O(n + m)
**Space:** O(min(n, m))

### 🧾 Input

* `nums`: List of integers

### 📤 Output

* The index that minimizes the absolute difference between prefix and suffix sums.

### ✅ Example

```py
Input: nums = [2,3,1,5,4]
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def minimumAverageDifference(self, nums: List[int]) -> int:
        total = sum(nums)
        min_diff = float('inf')
        res = 0
        prefix = 0

        for i in range(len(nums)):
            prefix += nums[i]
            left_avg = prefix // (i + 1)
            right_avg = 0 if i == len(nums) - 1 else (total - prefix) // (len(nums) - i - 1)
            diff = abs(left_avg - right_avg)
            if diff < min_diff:
                min_diff = diff
                res = i
        return res
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(1)

### 🧾 Input

* `nums`: List of positive integers
* `k`: Target product

### 📤 Output

* Number of contiguous subarrays where the product is less than `k`

### ✅ Example

```py
Input: nums = [10, 5, 2, 6], k = 100
Output: 8
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def numSubarrayProductLessThanK(self, nums: List[int], k: int) -> int:
        if k <= 1:
            return 0
        prod = 1
        left = 0
        count = 0
        for right, val in enumerate(nums):
            prod *= val
            while prod >= k:
                prod //= nums[left]
                left += 1
            count += right - left + 1
        return count
```

### 🧾 Input

* `nums`: List of positive integers
* `k`: Integer, the target minimum product

### 📤 Output

* Length of the smallest subarray such that the product is greater than or equal to `k`. Return `0` if no such subarray exists.

### ✅ Example

```py
Input: nums = [10, 5, 2, 6], k = 100
Output: 3
Explanation: The subarray [10, 5, 2] has a product of 100.
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def minSubArrayLenWithProductAtLeastK(self, nums: List[int], k: int) -> int:
        if k <= 1:
            return 1

        product = 1
        left = 0
        min_len = float('inf')

        for right in range(len(nums)):
            product *= nums[right]
            while product >= k:
                min_len = min(min_len, right - left + 1)
                product //= nums[left]
                left += 1

        return min_len if min_len != float('inf') else 0
```

### 🧾 Input

* `target`: Integer target sum
* `nums`: List of positive integers

### 📤 Output

* Minimal length of a contiguous subarray of which the sum ≥ `target`

### ✅ Example

```py
Input: target = 7, nums = [2,3,1,2,4,3]
Output: 2
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def minSubArrayLen(self, target: int, nums: List[int]) -> int:
        left = 0
        curr_sum = 0
        min_len = float('inf')

        for right in range(len(nums)):
            curr_sum += nums[right]
            while curr_sum >= target:
                min_len = min(min_len, right - left + 1)
                curr_sum -= nums[left]
                left += 1

        return 0 if min_len == float('inf') else min_len
```

  <summary><span style="font-size:19px;"><strong>Subarrays with K Different Integers - </strong><a href="https://leetcode.com/problems/subarrays-with-k-different-integers/" target="_blank">#992</a></span></summary>

### 🧾 Input

* `nums`: List of integers
* `k`: Target number of unique integers

### 📤 Output

* Number of subarrays with exactly `k` distinct integers

### ✅ Example

```py
Input: nums = [1,2,1,2,3], k = 2
Output: 7
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import defaultdict

class Solution:
    def subarraysWithKDistinct(self, nums: List[int], k: int) -> int:
        def atMostK(k):
            count = defaultdict(int)
            res = left = 0
            for right in range(len(nums)):
                if count[nums[right]] == 0:
                    k -= 1
                count[nums[right]] += 1

                while k < 0:
                    count[nums[left]] -= 1
                    if count[nums[left]] == 0:
                        k += 1
                    left += 1
                res += right - left + 1
            return res

        return atMostK(k) - atMostK(k - 1)
```

</details>

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(n)

  <summary><span style="font-size:19px;"><strong>3Sum Smaller - </strong><a href="https://leetcode.com/problems/3sum-smaller/" target="_blank">#259</a></span></summary>

### 🧾 Input

* `nums`: List of integers
* `target`: Integer target sum

### 📤 Output

* Number of index triplets such that the sum is less than `target`

### ✅ Example

```py
Input: nums = [-2,0,1,3], target = 2
Output: 2
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def threeSumSmaller(self, nums: List[int], target: int) -> int:
        nums.sort()
        count = 0
        n = len(nums)

        for i in range(n - 2):
            left, right = i + 1, n - 1
            while left < right:
                if nums[i] + nums[left] + nums[right] < target:
                    count += right - left
                    left += 1
                else:
                    right -= 1
        return count
```

### ✅ Example

```py
Input: fruits = [1,2,1,2,3]
Output: 4
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import defaultdict

class Solution:
    def totalFruit(self, fruits: List[int]) -> int:
        count = defaultdict(int)
        left = max_len = 0

        for right in range(len(fruits)):
            count[fruits[right]] += 1

            while len(count) > 2:
                count[fruits[left]] -= 1
                if count[fruits[left]] == 0:
                    del count[fruits[left]]
                left += 1

            max_len = max(max_len, right - left + 1)

        return max_len
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def subarraysDivByK(self, nums: List[int], k: int) -> int:
        mod_map = {0: -1}
        prefix = 0
        max_len = 0

        for i, num in enumerate(nums):
            prefix += num
            mod = prefix % k
            if mod < 0:
                mod += k
            if mod in mod_map:
                max_len = max(max_len, i - mod_map[mod])
            else:
                mod_map[mod] = i

        return max_len
```

### 🧾 Input

* `nums`: List of integers
* `k`: Target median value

### 📤 Output

* Number of subarrays where the median is exactly `k`

### ✅ Example

```py
Input: nums = [3,2,1,4,5], k = 4
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import Counter

class Solution:
    def countSubarrays(self, nums: List[int], k: int) -> int:
        prefix = Counter()
        i = nums.index(k)
        balance = 0
        prefix[0] = 1
        res = 0

        for j in range(i - 1, -1, -1):
            balance += 1 if nums[j] < k else -1
            prefix[balance] += 1

        balance = 0
        for j in range(i, len(nums)):
            balance += 1 if nums[j] > k else -1 if nums[j] < k else 0
            res += prefix[-balance] + prefix[-balance + 1]

        return res
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import defaultdict

class Solution:
    def subarraysWithKDistinct(self, nums: List[int], k: int) -> int:
        def atMostK(k):
            count = defaultdict(int)
            left = res = 0
            for right in range(len(nums)):
                if count[nums[right]] == 0:
                    k -= 1
                count[nums[right]] += 1

                while k < 0:
                    count[nums[left]] -= 1
                    if count[nums[left]] == 0:
                        k += 1
                    left += 1
                res += right - left + 1
            return res

        return atMostK(k) - atMostK(k - 1)
```

  <summary><span style="font-size:19px;"><strong>Shortest Subarray with Sum at Least K - </strong><a href="https://leetcode.com/problems/shortest-subarray-with-sum-at-least-k/" target="_blank">#862</a></span></summary>

### 🧾 Input

* `nums`: List of integers (can be negative)
* `k`: Target sum

### 📤 Output

* Length of the shortest non-empty subarray with sum at least `k`

### ✅ Example

```py
Input: nums = [2, -1, 2], k = 3
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import deque

class Solution:
    def shortestSubarray(self, nums: List[int], k: int) -> int:
        n = len(nums)
        prefix = [0] * (n + 1)
        for i in range(n):
            prefix[i + 1] = prefix[i] + nums[i]
        
        dq = deque()
        res = n + 1

        for i in range(n + 1):
            while dq and prefix[i] - prefix[dq[0]] >= k:
                res = min(res, i - dq.popleft())
            while dq and prefix[i] <= prefix[dq[-1]]:
                dq.pop()
            dq.append(i)

        return res if res <= n else -1
```

  <summary><span style="font-size:19px;"><strong>Max Sum of Rectangle No Larger Than K - </strong><a href="https://leetcode.com/problems/max-sum-of-rectangle-no-larger-than-k/" target="_blank">#363</a></span></summary>

### 🧾 Input

* `matrix`: 2D list of integers
* `k`: Integer target

### 📤 Output

* Maximum sum of a rectangle in the matrix such that the sum is no larger than `k`

### ✅ Example

```py
Input: matrix = [[1,0,1],[0,-2,3]], k = 2
Output: 2
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
import bisect

class Solution:
    def maxSumSubmatrix(self, matrix: List[List[int]], k: int) -> int:
        if not matrix:
            return 0
        max_sum = float('-inf')
        rows, cols = len(matrix), len(matrix[0])

        for left in range(cols):
            row_sums = [0] * rows
            for right in range(left, cols):
                for r in range(rows):
                    row_sums[r] += matrix[r][right]
                
                prefix = [0]
                curr_sum = 0
                for sum_val in row_sums:
                    curr_sum += sum_val
                    idx = bisect.bisect_left(prefix, curr_sum - k)
                    if idx < len(prefix):
                        max_sum = max(max_sum, curr_sum - prefix[idx])
                    bisect.insort(prefix, curr_sum)
        return max_sum
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(cols² × rows × log(rows))

  <summary><span style="font-size:19px;"><strong>Minimum Window Subsequence - </strong><a href="https://leetcode.com/problems/minimum-window-subsequence/" target="_blank">#727</a></span></summary>

### 🧾 Input

* `s1`: String to search in
* `s2`: Subsequence to find

### 📤 Output

* Minimum window in `s1` which contains `s2` as a subsequence

### ✅ Example

```py
Input: s1 = "abcdebdde", s2 = "bde"
Output: "bcde"
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def minWindow(self, s1: str, s2: str) -> str:
        m, n = len(s1), len(s2)
        min_len = float('inf')
        start = -1

        i = 0
        while i < m:
            if s1[i] == s2[0]:
                j = 0
                k = i
                while k < m:
                    if s1[k] == s2[j]:
                        j += 1
                        if j == n:
                            end = k + 1
                            j -= 1
                            while k >= i:
                                if s1[k] == s2[j]:
                                    j -= 1
                                    if j < 0:
                                        break
                                k -= 1
                            if end - k - 1 < min_len:
                                min_len = end - k - 1
                                start = k + 1
                            break
                    k += 1
            i += 1

        return "" if start == -1 else s1[start:start + min_len]
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(m × n)
**Space:** O(1)

</details>

  <summary><span style="font-size:19px;"><strong>Subarray Product Less Than K (at scale) - </strong><a href="https://leetcode.com/problems/subarray-product-less-than-k/" target="_blank">#713</a></span></summary>

### 🧾 Input

* `nums`: List of positive integers
* `k`: Integer

### 📤 Output

* Count of contiguous subarrays where product of all elements is less than `k`

### ✅ Example

```py
Input: nums = [10,5,2,6], k = 100
Output: 8
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def numSubarrayProductLessThanK(self, nums: List[int], k: int) -> int:
        if k <= 1:
            return 0
        prod = 1
        left = 0
        count = 0
        for right, val in enumerate(nums):
            prod *= val
            while prod >= k:
                prod //= nums[left]
                left += 1
            count += right - left + 1
        return count
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(1)

### 🧾 Input

* `nums`: List of integers
* `k`: Target sum

### 📤 Output

* Length of the longest subarray with sum equal to `k`

### ✅ Example

```py
Input: nums = [1, -1, 5, -2, 3], k = 3
Output: 4
Explanation: The subarray [1, -1, 5, -2] sums to 3 and has the longest length.
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def maxSubArrayLen(self, nums: List[int], k: int) -> int:
        prefix_sum = 0
        max_len = 0
        seen = {0: -1}
        for i, num in enumerate(nums):
            prefix_sum += num
            if prefix_sum - k in seen:
                max_len = max(max_len, i - seen[prefix_sum - k])
            if prefix_sum not in seen:
                seen[prefix_sum] = i
        return max_len
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(n)

  <summary><span style="font-size:19px;"><strong>Split Array With Equal Sum - </strong><a href="https://leetcode.com/problems/split-array-with-equal-sum/" target="_blank">#548</a></span></summary>

### 🧾 Input

* `nums`: List of integers

### 📤 Output

* Boolean indicating if it's possible to split array into 4 parts with equal sums

### ✅ Example

```py
Input: nums = [1,2,1,2,1,2,1]
Output: True
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def splitArray(self, nums: List[int]) -> bool:
        n = len(nums)
        for j in range(3, n - 3):
            seen = set()
            left_sum = sum(nums[:j])
            temp_sum = 0
            for i in range(1, j - 1):
                temp_sum += nums[i - 1]
                if temp_sum == left_sum - temp_sum - nums[i]:
                    seen.add(temp_sum)
            temp_sum = 0
            for k in range(j + 2, n - 1):
                temp_sum += nums[k - 1]
                if temp_sum == left_sum - temp_sum - nums[k] and temp_sum in seen:
                    return True
        return False
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n²)
**Space:** O(n)

### 🧾 Input

* `nums`: List of integers

### 📤 Output

* Count of beautiful subarrays (XOR of elements is 0)

### ✅ Example

```py
Input: nums = [4,3,1,2,4]
Output: 2
Explanation: There are two beautiful subarrays: [4,3,1,2] and [1,2,4,3]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import defaultdict

class Solution:
    def beautifulSubarrays(self, nums: List[int]) -> int:
        xor_map = defaultdict(int)
        xor_map[0] = 1
        xor_val = 0
        count = 0

        for num in nums:
            xor_val ^= num
            count += xor_map[xor_val]
            xor_map[xor_val] += 1

        return count
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(n)

  <summary><span style="font-size:19px;"><strong>Count of Range Sum - </strong><a href="https://leetcode.com/problems/count-of-range-sum/" target="_blank">#327</a></span></summary>

### 🧾 Input

* `nums`: List of integers
* `lower`, `upper`: Integer range

### 📤 Output

* Count of subarrays whose sum lies in `[lower, upper]`

### ✅ Example

```py
Input: nums = [-2,5,-1], lower = -2, upper = 2
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def countRangeSum(self, nums: List[int], lower: int, upper: int) -> int:
        from bisect import bisect_left, bisect_right, insort
        prefix = [0]
        total = 0
        count = 0
        for num in nums:
            total += num
            left = total - upper
            right = total - lower
            count += bisect_right(prefix, right) - bisect_left(prefix, left)
            insort(prefix, total)
        return count
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n log n)
**Space:** O(n)

</details>

  <summary><span style="font-size:19px;"><strong>Range Product Except Self - </strong><a href="https://leetcode.com/problems/range-product-except-self/" target="_blank">#238</a></span></summary>

### 🧾 Input

* `nums`: List of integers

### 📤 Output

* Return an array where `res[i]` is the product of all elements of `nums` except `nums[i]`

### ✅ Example

```py
Input: nums = [1,2,3,4]
Output: [24,12,8,6]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def productExceptSelf(self, nums: List[int]) -> List[int]:
        n = len(nums)
        res = [1] * n

        # Left product
        left = 1
        for i in range(n):
            res[i] = left
            left *= nums[i]

        # Right product
        right = 1
        for i in range(n - 1, -1, -1):
            res[i] *= right
            right *= nums[i]

        return res
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(1) (output array not counted)

</details>
