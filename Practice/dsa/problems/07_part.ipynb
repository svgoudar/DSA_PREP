#### Problem 1: [Two Sum](https://leetcode.com/problems/two-sum/)

**Input**

* `nums`: List of integers
* `target`: Integer target sum

**Output**

* Indices of two numbers such that they add up to target

**Example**

```py
Input: nums = [2,7,11,15], target = 9
Output: [0,1]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        seen = {}
        for i, num in enumerate(nums):
            complement = target - num
            if complement in seen:
                return [seen[complement], i]
            seen[num] = i
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(n)
**Space Complexity:** O(n)

#### Problem 49: [Group Anagrams](https://leetcode.com/problems/group-anagrams/)

**Input**

* `strs`: List of strings

**Output**

* Grouped anagrams as a list of lists

**Example**

```py
Input: strs = ["eat","tea","tan","ate","nat","bat"]
Output: [["eat","tea","ate"],["tan","nat"],["bat"]]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        anagrams = defaultdict(list)
        for word in strs:
            key = tuple(sorted(word))
            anagrams[key].append(word)
        return list(anagrams.values())
```

#### Problem 560: [Subarray Sum Equals K](https://leetcode.com/problems/subarray-sum-equals-k/)

**Input**

* `nums`: List of integers
* `k`: Integer

**Output**

* Number of continuous subarrays that sum to `k`

**Example**

```py
Input: nums = [1,1,1], k = 2
Output: 2
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def subarraySum(self, nums: List[int], k: int) -> int:
        count = 0
        prefix_sum = 0
        prefix_map = defaultdict(int)
        prefix_map[0] = 1
        for num in nums:
            prefix_sum += num
            count += prefix_map.get(prefix_sum - k, 0)
            prefix_map[prefix_sum] += 1
        return count
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(n)

#### Problem 128: [Longest Consecutive Sequence](https://leetcode.com/problems/longest-consecutive-sequence/)

**Input**

* `nums`: List of integers

**Output**

* Length of the longest consecutive elements sequence

**Example**

```py
Input: nums = [100, 4, 200, 1, 3, 2]
Output: 4
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        num_set = set(nums)
        longest = 0
        for num in num_set:
            if num - 1 not in num_set:
                current = num
                length = 1
                while current + 1 in num_set:
                    current += 1
                    length += 1
                longest = max(longest, length)
        return longest
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(n)
**Space Complexity:** O(n)

</details>

#### Problem 138: [Copy List with Random Pointer](https://leetcode.com/problems/copy-list-with-random-pointer/)

**Input**

* Head of a linked list with `next` and `random` pointers

**Output**

* Deep copy of the list

**Example**

```py
Input: head = [[7,null],[13,0],[11,4],[10,2],[1,0]]
Output: Copied list
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def copyRandomList(self, head: 'Optional[Node]') -> 'Optional[Node]':
        old_to_new = {None: None}
        curr = head
        while curr:
            old_to_new[curr] = Node(curr.val)
            curr = curr.next
        curr = head
        while curr:
            old_to_new[curr].next = old_to_new[curr.next]
            old_to_new[curr].random = old_to_new[curr.random]
            curr = curr.next
        return old_to_new[head]
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(n)

#### Problem 438: [Find All Anagrams in a String](https://leetcode.com/problems/find-all-anagrams-in-a-string/)

**Input**

* `s`, `p`: Strings

**Output**

* Starting indices of anagrams of `p` in `s`

**Example**

```py
Input: s = "cbaebabacd", p = "abc"
Output: [0,6]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def findAnagrams(self, s: str, p: str) -> List[int]:
        res = []
        p_count = Counter(p)
        s_count = Counter()
        for i in range(len(s)):
            s_count[s[i]] += 1
            if i >= len(p):
                if s_count[s[i - len(p)]] == 1:
                    del s_count[s[i - len(p)]]
                else:
                    s_count[s[i - len(p)]] -= 1
            if s_count == p_count:
                res.append(i - len(p) + 1)
        return res
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(n)
**Space Complexity:** O(1)

#### Problem 811: [Subdomain Visit Count](https://leetcode.com/problems/subdomain-visit-count/)

* `cpdomains`: List of domain strings with visit counts

**Output**

* List of strings with counts for each subdomain

**Example**

```py
Input: ["9001 discuss.leetcode.com"]
Output: ["9001 leetcode.com", "9001 discuss.leetcode.com", "9001 com"]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def subdomainVisits(self, cpdomains: List[str]) -> List[str]:
        count = defaultdict(int)
        for entry in cpdomains:
            visits, domain = entry.split()
            visits = int(visits)
            fragments = domain.split('.')
            for i in range(len(fragments)):
                subdomain = ".".join(fragments[i:])
                count[subdomain] += visits
        return [f\"{v} {d}\" for d, v in count.items()]
```

#### Problem 76: [Minimum Window Substring](https://leetcode.com/problems/minimum-window-substring/)

**Input**

- `s`: Source string
- `t`: Target string

**Output**

- Smallest substring in `s` that contains all characters in `t`

**Example**

```py
Input: s = "ADOBECODEBANC", t = "ABC"
Output: "BANC"
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import Counter

class Solution:
    def minWindow(self, s: str, t: str) -> str:
        if not t or not s:
            return ""

        dict_t = Counter(t)
        required = len(dict_t)

        l, r = 0, 0
        formed = 0
        window_counts = {}
        ans = float("inf"), None, None

        while r < len(s):
            character = s[r]
            window_counts[character] = window_counts.get(character, 0) + 1

            if character in dict_t and window_counts[character] == dict_t[character]:
                formed += 1

            while l <= r and formed == required:
                character = s[l]

                if r - l + 1 < ans[0]:
                    ans = (r - l + 1, l, r)

                window_counts[character] -= 1
                if character in dict_t and window_counts[character] < dict_t[character]:
                    formed -= 1

                l += 1
            r += 1

        return "" if ans[0] == float("inf") else s[ans[1]: ans[2] + 1]
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(|s| + |t|)  
**Space Complexity:** O(|s| + |t|)

</details>

#### Problem 30: [Substring with Concatenation of All Words](https://leetcode.com/problems/substring-with-concatenation-of-all-words/)

**Input**

- `s`: String
- `words`: List of words (all same length)

**Output**

- List of starting indices where a concatenation of each word in `words` occurs in `s`

**Example**

```py
Input: s = "barfoothefoobarman", words = ["foo","bar"]
Output: [0,9]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import Counter

class Solution:
    def findSubstring(self, s: str, words: List[str]) -> List[int]:
        if not s or not words:
            return []

        word_len = len(words[0])
        num_words = len(words)
        total_len = word_len * num_words
        word_count = Counter(words)
        result = []

        for i in range(word_len):
            left = i
            right = i
            curr_count = Counter()
            count = 0

            while right + word_len <= len(s):
                word = s[right:right + word_len]
                right += word_len

                if word in word_count:
                    curr_count[word] += 1
                    count += 1

                    while curr_count[word] > word_count[word]:
                        curr_count[s[left:left + word_len]] -= 1
                        left += word_len
                        count -= 1

                    if count == num_words:
                        result.append(left)
                else:
                    curr_count.clear()
                    count = 0
                    left = right

        return result
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(n * word_len)  
**Space Complexity:** O(n)

</details>

#### Problem 126: [Word Ladder II](https://leetcode.com/problems/word-ladder-ii/)

**Input**

- `beginWord`: Starting word (string)
- `endWord`: Target word (string)
- `wordList`: List of allowed words (List[str])

**Output**

- All shortest transformation sequences from beginWord to endWord

**Example**

```py
Input: beginWord = "hit", endWord = "cog", wordList = ["hot","dot","dog","lot","log","cog"]
Output: [["hit","hot","dot","dog","cog"],["hit","hot","lot","log","cog"]]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import defaultdict, deque

class Solution:
    def findLadders(self, beginWord, endWord, wordList):
        wordSet = set(wordList)
        if endWord not in wordSet:
            return []

        # Step 1: Use BFS to build graph and find shortest path length
        graph = defaultdict(list)
        level = {beginWord: 0}
        queue = deque([beginWord])

        while queue:
            word = queue.popleft()
            for i in range(len(word)):
                for c in 'abcdefghijklmnopqrstuvwxyz':
                    nextWord = word[:i] + c + word[i+1:]
                    if nextWord in wordSet and nextWord not in level:
                        level[nextWord] = level[word] + 1
                        queue.append(nextWord)
                        graph[word].append(nextWord)
                    elif nextWord in level and level[nextWord] == level[word] + 1:
                        graph[word].append(nextWord)

        # Step 2: Use DFS to find all shortest paths
        res = []
        def dfs(path):
            word = path[-1]
            if word == endWord:
                res.append(list(path))
                return
            for nei in graph[word]:
                path.append(nei)
                dfs(path)
                path.pop()

        dfs([beginWord])
        return res
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(N * L^2) (where N = number of words, L = length of each word)  
**Space Complexity:** O(N * L)

</details>

#### Problem 269: [Alien Dictionary](https://leetcode.com/problems/alien-dictionary/)

**Input**

- `words`: A list of words sorted lexicographically by the rules of the alien language

**Output**

- A string representing the characters in the correct order of the alien language

**Example**

```py
Input: words = ["wrt","wrf","er","ett","rftt"]
Output: "wertf"
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import defaultdict, deque

class Solution:
    def alienOrder(self, words: List[str]) -> str:
        graph = defaultdict(set)
        indegree = {c: 0 for word in words for c in word}

        for i in range(len(words) - 1):
            w1, w2 = words[i], words[i + 1]
            min_len = min(len(w1), len(w2))
            if w1[:min_len] == w2[:min_len] and len(w1) > len(w2):
                return ""
            for c1, c2 in zip(w1, w2):
                if c1 != c2:
                    if c2 not in graph[c1]:
                        graph[c1].add(c2)
                        indegree[c2] += 1
                    break

        queue = deque([c for c in indegree if indegree[c] == 0])
        result = []
        while queue:
            c = queue.popleft()
            result.append(c)
            for nei in graph[c]:
                indegree[nei] -= 1
                if indegree[nei] == 0:
                    queue.append(nei)

        return "".join(result) if len(result) == len(indegree) else ""
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(C) where C is the total number of characters in all words  
**Space Complexity:** O(1) (since number of distinct characters is bounded)

</details>

#### Problem 32: [Longest Valid Parentheses](https://leetcode.com/problems/longest-valid-parentheses/)

**Input**

- `s`: A string containing just '(' and ')'

**Output**

- Length of the longest valid (well-formed) parentheses substring

**Example**

```py
Input: s = ")()())"
Output: 4
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def longestValidParentheses(self, s: str) -> int:
        max_len = 0
        stack = [-1]
        for i, char in enumerate(s):
            if char == '(':
                stack.append(i)
            else:
                stack.pop()
                if not stack:
                    stack.append(i)
                else:
                    max_len = max(max_len, i - stack[-1])
        return max_len
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(n)  
**Space Complexity:** O(n)

</details>

#### Problem 336: [Palindrome Pairs](https://leetcode.com/problems/palindrome-pairs/)

**Input**

* `words`: List of unique strings

**Output**

* List of index pairs such that the concatenation of the two words is a palindrome

**Example**

```py
Input: words = ["bat","tab","cat"]
Output: [[0,1],[1,0]]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def palindromePairs(self, words: List[str]) -> List[List[int]]:
        word_dict = {w: i for i, w in enumerate(words)}
        res = []
        
        for i, word in enumerate(words):
            for j in range(len(word) + 1):
                prefix, suffix = word[:j], word[j:]
                if prefix == prefix[::-1]:
                    back = suffix[::-1]
                    if back != word and back in word_dict:
                        res.append([word_dict[back], i])
                if j != len(word) and suffix == suffix[::-1]:
                    front = prefix[::-1]
                    if front != word and front in word_dict:
                        res.append([i, word_dict[front]])
        return res
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(k \* n²), where `k` is number of words, and `n` is max length of a word
**Space Complexity:** O(k)

</details>

#### Problem 355: [Design Twitter](https://leetcode.com/problems/design-twitter/)

**Input**

```py
Input: ["Twitter","postTweet","getNewsFeed","follow","postTweet","getNewsFeed","unfollow","getNewsFeed"]
       [[],[1,5],[1],[1,2],[2,6],[1],[1,2],[1]]
Output: [null,null,[5],null,null,[6,5],null,[5]]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
import heapq

class Twitter:

    def __init__(self):
        self.time = 0
        self.user_tweets = defaultdict(list)
        self.following = defaultdict(set)

    def postTweet(self, userId: int, tweetId: int) -> None:
        self.user_tweets[userId].append((self.time, tweetId))
        self.time -= 1

    def getNewsFeed(self, userId: int) -> List[int]:
        heap = []
        self.following[userId].add(userId)
        for uid in self.following[userId]:
            heap.extend(self.user_tweets[uid][-10:])
        return [tweetId for _, tweetId in heapq.nsmallest(10, heap)]

    def follow(self, followerId: int, followeeId: int) -> None:
        self.following[followerId].add(followeeId)

    def unfollow(self, followerId: int, followeeId: int) -> None:
        self.following[followerId].discard(followeeId)
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:**

* postTweet: O(1)
* getNewsFeed: O(n log n)
* follow/unfollow: O(1)

**Space Complexity:** O(n)

</details>

#### Problem 146: [LRU Cache](https://leetcode.com/problems/lru-cache/)

**Input**

* `get(key)`, `put(key, value)` operations with limited capacity

**Output**

* Implements LRU cache behavior

**Example**

```py
Input: ["LRUCache","put","put","get","put","get","get"]
       [[2],[1,1],[2,2],[1],[3,3],[2],[3]]
Output: [null,null,null,1,null,-1,3]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import OrderedDict

class LRUCache:

    def __init__(self, capacity: int):
        self.cache = OrderedDict()
        self.capacity = capacity

    def get(self, key: int) -> int:
        if key not in self.cache:
            return -1
        self.cache.move_to_end(key)
        return self.cache[key]

    def put(self, key: int, value: int) -> None:
        if key in self.cache:
            self.cache.move_to_end(key)
        self.cache[key] = value
        if len(self.cache) > self.capacity:
            self.cache.popitem(last=False)
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(1) for get and put
**Space Complexity:** O(capacity)

</details>

#### Problem 408: [Valid Word Abbreviation](https://leetcode.com/problems/valid-word-abbreviation/)

**Input**

* `word`: Original word
* `abbr`: Abbreviation string

**Output**

* Boolean indicating if abbreviation is valid for the word

**Example**

```py
Input: word = "internationalization", abbr = "i12iz4n"
Output: True
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def validWordAbbreviation(self, word: str, abbr: str) -> bool:
        i = j = 0
        while i < len(word) and j < len(abbr):
            if abbr[j].isdigit():
                if abbr[j] == '0':
                    return False
                num = 0
                while j < len(abbr) and abbr[j].isdigit():
                    num = num * 10 + int(abbr[j])
                    j += 1
                i += num
            else:
                if i >= len(word) or word[i] != abbr[j]:
                    return False
                i += 1
                j += 1
        return i == len(word) and j == len(abbr)
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(n), where n = len(abbr)
**Space Complexity:** O(1)

</details>

#### Problem 115: [Distinct Subsequences](https://leetcode.com/problems/distinct-subsequences/)

**Input**

* `s`: Source string
* `t`: Target string

**Output**

* Count of distinct subsequences of `s` which equals `t`

**Example**

```py
Input: s = "rabbbit", t = "rabbit"
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def numDistinct(self, s: str, t: str) -> int:
        m, n = len(s), len(t)
        dp = [[0]*(n+1) for _ in range(m+1)]

        for i in range(m+1):
            dp[i][0] = 1

        for i in range(1, m+1):
            for j in range(1, n+1):
                if s[i-1] == t[j-1]:
                    dp[i][j] = dp[i-1][j-1] + dp[i-1][j]
                else:
                    dp[i][j] = dp[i-1][j]
        return dp[m][n]
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time Complexity:** O(m \* n)
**Space Complexity:** O(m \* n)

</details>
