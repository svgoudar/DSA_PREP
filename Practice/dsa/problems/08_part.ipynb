### Sorting Algorithm

#### Bubble Sort

**Input**

* `arr`: List of integers

**Output**

* Sorted list in ascending order

**Example**

```python
Input: arr = [5, 1, 4, 2, 8]
Output: [1, 2, 4, 5, 8]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def bubbleSort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n²)  
**Space:** O(1)

</details>

#### Insertion Sort

**Input**

* `arr`: List of integers

**Output**

* Sorted list in ascending order

**Example**

```python
Input: arr = [12, 11, 13, 5, 6]
Output: [5, 6, 11, 12, 13]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def insertionSort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n²)  
**Space:** O(1)

</details>

#### Selection Sort

**Input**

* `arr`: List of integers

**Output**

* Sorted list in ascending order

**Example**

```python
Input: arr = [64, 25, 12, 22, 11]
Output: [11, 12, 22, 25, 64]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def selectionSort(arr):
    for i in range(len(arr)):
        min_idx = i
        for j in range(i+1, len(arr)):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n²)

**Space:** O(1)

</details>

#### Merge Sort

**Input**

* `arr`: List of integers

**Output**

* Sorted list in ascending order

**Example**

```python
Input: arr = [38, 27, 43, 3, 9, 82, 10]
Output: [3, 9, 10, 27, 38, 43, 82]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def mergeSort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = mergeSort(arr[:mid])
    right = mergeSort(arr[mid:])
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n log n)  
**Space:** O(n)

</details>

#### Quick Sort

**Input**

* `arr`: List of integers

**Output**

* Sorted list in ascending order

**Example**

```python
Input: arr = [10, 7, 8, 9, 1, 5]
Output: [1, 5, 7, 8, 9, 10]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def quickSort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quickSort(left) + middle + quickSort(right)
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n log n) average, O(n²) worst  
**Space:** O(log n)

</details>

#### Heap Sort

**Input**

* `arr`: List of integers

**Output**

* Sorted list in ascending order

**Example**

```python
Input: arr = [12, 11, 13, 5, 6, 7]
Output: [5, 6, 7, 11, 12, 13]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def heapify(arr, n, i):
    largest = i
    left = 2 * i + 1
    right = 2 * i + 2
    if left < n and arr[left] > arr[largest]:
        largest = left
    if right < n and arr[right] > arr[largest]:
        largest = right
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)

def heapSort(arr):
    n = len(arr)
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)
    for i in range(n - 1, 0, -1):
        arr[0], arr[i] = arr[i], arr[0]
        heapify(arr, i, 0)
    return arr
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n log n)  
**Space:** O(1)

</details>

#### Counting Sort

**Input**

* `arr`: List of non-negative integers

**Output**

* Sorted list in ascending order

**Example**

```python
Input: arr = [4, 2, 2, 8, 3, 3, 1]
Output: [1, 2, 2, 3, 3, 4, 8]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def countingSort(arr):
    max_val = max(arr)
    count = [0] * (max_val + 1)
    for num in arr:
        count[num] += 1
    result = []
    for i, c in enumerate(count):
        result.extend([i] * c)
    return result
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n + k) where k is range  
**Space:** O(k)

</details>

#### Radix Sort

**Input**

* `arr`: List of non-negative integers

**Output**

* Sorted list in ascending order

**Example**

```python
Input: arr = [170, 45, 75, 90, 802, 24, 2, 66]
Output: [2, 24, 45, 66, 75, 90, 170, 802]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def countingSortForRadix(arr, exp):
    n = len(arr)
    output = [0] * n
    count = [0] * 10
    for i in range(n):
        index = arr[i] // exp
        count[index % 10] += 1
    for i in range(1, 10):
        count[i] += count[i - 1]
    i = n - 1
    while i >= 0:
        index = arr[i] // exp
        output[count[index % 10] - 1] = arr[i]
        count[index % 10] -= 1
        i -= 1
    for i in range(n):
        arr[i] = output[i]

def radixSort(arr):
    max_val = max(arr)
    exp = 1
    while max_val // exp > 0:
        countingSortForRadix(arr, exp)
        exp *= 10
    return arr
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(d × (n + k)) where d is digits  
**Space:** O(n + k)

</details>

#### Bucket Sort

**Input**

* `arr`: List of floats in range [0, 1)

**Output**

* Sorted list in ascending order

**Example**

```python
Input: arr = [0.897, 0.565, 0.656, 0.1234, 0.665, 0.3434]
Output: [0.1234, 0.3434, 0.565, 0.656, 0.665, 0.897]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def bucketSort(arr):
    n = len(arr)
    buckets = [[] for _ in range(n)]
    for num in arr:
        index = int(num * n)
        buckets[index].append(num)
    for bucket in buckets:
        bucket.sort()
    result = []
    for bucket in buckets:
        result.extend(bucket)
    return result
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n + k) average  
**Space:** O(n + k)

</details>

#### Topological Sort

**Input**

* `graph`: Dictionary representing directed acyclic graph (DAG)

**Output**

* List representing topological ordering

**Example**

```python
Input: graph = {5: [2, 0], 4: [0, 1], 2: [3], 3: [1]}
Output: [5, 4, 2, 3, 1, 0] (one valid ordering)
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def topologicalSortUtil(v, visited, stack, graph):
    visited[v] = True
    if v in graph:
        for neighbor in graph[v]:
            if not visited.get(neighbor, False):
                topologicalSortUtil(neighbor, visited, stack, graph)
    stack.append(v)

def topologicalSort(graph):
    visited = {}
    stack = []
    for vertex in graph:
        if not visited.get(vertex, False):
            topologicalSortUtil(vertex, visited, stack, graph)
    return stack[::-1]
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(V + E)  
**Space:** O(V)

</details>
