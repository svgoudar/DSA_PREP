### 🧾 Problem

Given two strings `s` and `t`, return `true` if `t` is an anagram of `s`, and `false` otherwise.

An Anagram is a word or phrase formed by rearranging the letters of a different word or phrase, typically using all the original letters exactly once.

---

### 🔢 Input

* `s`: A string (e.g., `"anagram"`)
* `t`: Another string (e.g., `"nagaram"`)

---

### 📤 Output

* Boolean value `True` if `t` is an anagram of `s`, else `False`

---

### ✅ Example

```python
Input: s = "anagram", t = "nagaram"
Output: True

Input: s = "rat", t = "car"
Output: False
```

---

<details>
  <summary><strong>🧠 Solution Code</strong></summary>

```py
class Solution(object):
    def isAnagram(self, s, t):
        """
        :type s: str
        :type t: str
        :rtype: bool
        """
        if len(s) != len(t):
            return False
        
        for i in set(s):
            if i not in t:
                return False
            if s.count(i) != t.count(i):
                return False
        return True
```

</details>

<details>
  <summary><strong>🧮 Complexity Analysis</strong></summary>

* **Time Complexity:** O(n) – due to sorting both strings
* **Space Complexity:** O(1) – constant space (excluding input and output)

</details>

### 🧾 Problem

Given two strings `haystack` and `needle`, return the index of the first occurrence of `needle` in `haystack`, or `-1` if `needle` is not part of `haystack`.

If `needle` is an empty string, return `0`.

---

### 🔢 Input

* `haystack`: A string (e.g., `"hello"`)
* `needle`: A string to find within `haystack` (e.g., `"ll"`)

---

### 📤 Output

* An integer representing the index of the first occurrence of `needle` in `haystack`, or `-1` if not found.

---

### ✅ Example

```python
Input: haystack = "hello", needle = "ll"
Output: 2

Input: haystack = "aaaaa", needle = "bba"
Output: -1

Input: haystack = "abc", needle = ""
Output: 0
```

---

<details>
  <summary><strong>🧠 Solution Code</strong></summary>

```py
class Solution:
    def strStr(self, haystack, needle):
        for i in range(len(haystack) - len(needle) + 1):
            if haystack[i:i+len(needle)] == needle:
                return i
        return -1
```

</details>

<details>
  <summary><strong>🧮 Complexity Analysis</strong></summary>

* **Time Complexity:** O((n - m + 1) \* m) in the worst case, where `n` is the length of `haystack` and `m` is the length of `needle`.
* **Space Complexity:** O(1) – no extra space used.

> If using built-in `find()`, it is optimized and runs in linear time on average.

</details>

### 🧾 Problem

Write a function that reverses a string. The input string is given as an array of characters `s`.

---

### 🔢 Input

* `s`: List of characters (e.g., `["h","e","l","l","o"]`)

---

### 📤 Output

* The same list `s` should be modified in-place to reverse the characters (e.g., `["o","l","l","e","h"]`).

---

### ✅ Example

```python
Input: s = ["h","e","l","l","o"]
Output: ["o","l","l","e","h"]
```

---

<details>
  <summary><strong>🧠 Solution Code</strong></summary>

```py
class Solution(object):
    def reverseString(self, s):
        """
        Do not return anything, modify s in-place instead.
        """
        left, right = 0, len(s) - 1
        while left < right:
            s[left], s[right] = s[right], s[left]
            left += 1
            right -= 1
```

</details>

<details>
  <summary><strong>🧮 Complexity Analysis</strong></summary>

* **Time Complexity:** O(n)
* **Space Complexity:** O(1) (in-place)

</details>

### 🧾 Problem

Convert a sentence to "Goat Latin" by these rules:

1. If a word begins with a vowel, append "ma".
2. If it begins with a consonant, move the first letter to the end, then append "ma".
3. Add one letter `'a'` to the end of each word per its word index, starting with 1.

---

### 🔢 Input

* `sentence`: A string (e.g., `"I speak Goat Latin"`)

---

### 📤 Output

* Transformed Goat Latin sentence (e.g., `"Imaa peaksmaaa oatGmaaaa atinLmaaaaa"`)

---

### ✅ Example

```python
Input: sentence = "I speak Goat Latin"
Output: "Imaa peaksmaaa oatGmaaaa atinLmaaaaa"
```

---

<details>
  <summary><strong>🧠 Solution Code</strong></summary>

```py
class Solution(object):
    def toGoatLatin(self, S):
        """
        :type S: str
        :rtype: str
        """
        vowels = set('aeiouAEIOU')
        words = S.split()
        res = []
        for i, w in enumerate(words, 1):
            if w[0] in vowels:
                new = w + "ma"
            else:
                new = w[1:] + w[0] + "ma"
            new += "a" * i
            res.append(new)
        return " ".join(res)
```

</details>

* **Time Complexity:** O(n + k²) where `k` is average word length

### 🧾 Problem

Implement the function `toLowerCase()` which converts a string to lowercase.

---

### 🔢 Input

* `s`: A string (e.g., `"Hello"`)

---

### 📤 Output

* Lowercase version of the string (e.g., `"hello"`)

---

### ✅ Example

```python
Input: s = "Hello"
Output: "hello"
```

---

<details>
  <summary><strong>🧠 Solution Code</strong></summary>

```py
class Solution(object):
    def toLowerCase(self, s):
        """
        :type s: str
        :rtype: str
        """
        result = []
        for ch in s:
            if 'A' <= ch <= 'Z':
                result.append(chr(ord(ch) - ord('A') + ord('a')))
            else:
                result.append(ch)
        return "".join(result)
```

Given a string `s`, determine if it is a palindrome considering only alphanumeric characters and ignoring cases.

---

### 🔢 Input

* `s`: A string (e.g., `"A man, a plan, a canal: Panama"`)

---

### 📤 Output

* Boolean value — `True` if `s` is a palindrome per the rules, else `False`.

---

### ✅ Example

```python
Input: s = "A man, a plan, a canal: Panama"
Output: True

Input: s = "race a car"
Output: False
```

---

<details>
  <summary><strong>🧠 Solution Code</strong></summary>

```py
class Solution(object):
    def isPalindrome(self, s):
        """
        :type s: str
        :rtype: bool
        """
        filtered = [ch.lower() for ch in s if ch.isalnum()]
        return filtered == filtered[::-1]
```

Your friend types their name `name`, but sometimes when typing a key gets long pressed, and the character may be repeated one or more times. Check if the typed string `typed` could be a result of such long presses.

---

### 🔢 Input

* `name`: The intended name string (e.g., `"alex"`)
* `typed`: The actual typed string (e.g., `"aaleex"`)

---

### 📤 Output

* Boolean value — `True` if `typed` matches a long-pressed version of `name`, else `False`.

---

### ✅ Example

```python
Input: name = "alex", typed = "aaleex"
Output: True

Input: name = "saeed", typed = "ssaaedd"
Output: False
```

---

<details>
  <summary><strong>🧠 Solution Code</strong></summary>

```py
class Solution:
    def isLongPressedName(self, name: str, typed: str) -> bool:
        j = 0
        i = 0
        n = len(name)
        t = len(typed)
        
        while j < t:
            if i < n and name[i] == typed[j]:
                i +=1
            elif j == 0 or typed[j] != typed[j-1]:
                return False
            j +=1
        return i == n
```

### 🧾 Problem

The count-and-say sequence is a sequence of digit strings defined recursively:

* `countAndSay(1) = "1"`
* For `n > 1`, `countAndSay(n)` is the verbal description of `countAndSay(n-1)`.

Generate the `n`th term of this sequence.

---

### 🔢 Input

* `n`: Integer (e.g., `4`)

---

### 📤 Output

* The `n`th string in the count‑and‑say sequence

---

### ✅ Example

```python
Input: n = 4
Output: "1211"
# Explanation:
# term1 = "1"
# term2 = "11"       ("one 1")
# term3 = "21"       ("two 1s")
# term4 = "1211"     ("one 2, one 1")
```

---

<details>
  <summary><strong>🧠 Solution Code</strong></summary>

```py
class Solution(object):
    def countAndSay(self, n):
        """
        :type n: int
        :rtype: str
        """
        res = "1"
        for _ in range(1, n):
            prev = res
            res = ""
            count = 1
            for i in range(1, len(prev)):
                if prev[i] == prev[i - 1]:
                    count += 1
                else:
                    res += str(count) + prev[i - 1]
                    count = 1
            res += str(count) + prev[-1]
        return res
```

</details>

  <summary><strong>🧮 Complexity Analysis</strong></summary>

* **Time Complexity:** approximately O(2ⁿ) in worst case due to growth of string length
* **Space Complexity:** O(2ⁿ) for output string

### 🧾 Problem

Given a string `s`, find the length of the longest substring without repeating characters.

---

### 🔢 Input

* `s`: A string (e.g., `"abcabcbb"`)

---

### 📤 Output

* Integer length (e.g., `3` for `"abc"`)

---

### ✅ Example

```python
Input: s = "abcabcbb"
Output: 3
```

---

<details>
  <summary><strong>🧠 Solution Code</strong></summary>

```py
class Solution(object):
    def lengthOfLongestSubstring(self, s):
        """
        :type s: str
        :rtype: int
        """
        seen = {}
        left = res = 0
        for right, ch in enumerate(s):
            if ch in seen and seen[ch] >= left:
                left = seen[ch] + 1
            seen[ch] = right
            res = max(res, right - left + 1)
        return res
```

</details>

  <summary><strong>🧮 Complexity Analysis</strong></summary>

* **Time Complexity:** O(n)
* **Space Complexity:** O(min(n, m)) where m = charset size

### 🧾 Problem

Given string `s` and integer `k`, repeatedly remove adjacent duplicate characters of length `k` until no more.

---

### 🔢 Input

* `s`: string
* `k`: int

---

### 📤 Output

* The final string after all removals.

---

### ✅ Example

```python
Input: s = "deeedbbcccbdaa", k = 3
Output: "aa"
```

---

<details>
  <summary><strong>🧠 Solution Code</strong></summary>

```py
class Solution(object):
    def removeDuplicates(self, s, k):
        """
        :type s: str
        :type k: int
        :rtype: str
        """
        stack = []  # Each element is [char, count]

        for char in s:
            if stack and stack[-1][0] == char:
                stack[-1][1] += 1  # increment count
                if stack[-1][1] == k:
                    stack.pop()  # remove the group
            else:
                stack.append([char, 1])  # new char, count = 1

        # Reconstruct string from stack
        return ''.join(char * count for char, count in stack)

```

  <summary><strong>🧮 Complexity Analysis</strong></summary>

### ✅ Example

```python
Input: num1 = "123", num2 = "456"
Output: "56088"
```

<details>
  <summary><strong>🧠 Solution Code</strong></summary>

```py
class Solution(object):
    def multiply(self, num1, num2):
        m, n = len(num1), len(num2)
        res = [0]*(m+n)
        for i in reversed(range(m)):
            for j in reversed(range(n)):
                mul = int(num1[i])*int(num2[j])
                sum_ = mul + res[i+j+1]
                res[i+j+1] = sum_ %10
                res[i+j] += sum_//10
        # skip leading zeros
        idx = 0
        while idx<len(res)-1 and res[idx]==0:
            idx+=1
        return ''.join(map(str,res[idx:]))

# simplified

class Solution(object):
    def multiply(self, num1, num2):
        """
        :type num1: str
        :type num2: str
        :rtype: str
        """
        return str(int(num1) * int(num2))
        
```

<details>
  <summary><strong>🧠 Solution Code</strong></summary>

```py
class Solution(object):
    def wordPattern(self, pattern, s):
        """
        :type pattern: str
        :type s: str
        :rtype: bool
        """
        words = s.split()
        if len(pattern) != len(words):
            return False
        mp, used = {}, set()
        for c, w in zip(pattern, words):
            if c in mp:
                if mp[c] != w:
                    return False
            else:
                if w in used:
                    return False
                mp[c] = w
                used.add(w)
        return True
```

### 🔢 Input

* `s`: string

### 📤 Output

### ✅ Example

```python
Input: order = "cba", s = "abcd"
Output: "cbad"
```

---

<details>
  <summary><strong>🧠 Solution Code</strong></summary>

```py
class Solution(object):
    def customSortString(self, order, s):
        """
        :type order: str
        :type s: str
        :rtype: str
        """
        count = {}
        for ch in s:
            count[ch] = count.get(ch,0) +1
        res = []
        for ch in order:
            res.append(ch * count.pop(ch,0))
        for ch, cnt in count.items():
            res.append(ch * cnt)
        return "".join(res)
```

### 🧾 Problem

Given strings `s` and pattern `p` where `p` includes '?' and '\*' wildcard characters, implement wildcard matching.

* '?' matches any single character.
* '\*' matches any sequence of characters (including empty).

Return `true` if `s` matches the pattern.

---

### 🔢 Input

* `s`: Target string
* `p`: Pattern string with wildcards

---

### 📤 Output

* Boolean value whether `s` matches `p`

---

### ✅ Example

```python
Input: s = "adceb", p = "*a*b"
Output: True

Input: s = "acdcb", p = "a*c?b"
Output: False
```

---

<details>
  <summary><strong>🧠 Solution Code (DP)</strong></summary>

```py
class Solution(object):
    def isMatch(self, s, p):
        m, n = len(s), len(p)
        dp = [[False]*(n+1) for _ in range(m+1)]
        dp[0][0] = True
        for j in range(1, n+1):
            if p[j-1] == '*':
                dp[0][j] = dp[0][j-1]
        for i in range(1, m+1):
            for j in range(1, n+1):
                if p[j-1] == s[i-1] or p[j-1] == '?':
                    dp[i][j] = dp[i-1][j-1]
                elif p[j-1] == '*':
                    dp[i][j] = dp[i][j-1] or dp[i-1][j]
        return dp[m][n]
```

  <summary><span style="font-size:19px;"><strong>Regular Expression Matching - </strong><a href="https://leetcode.com/problems/regular-expression-matching/" target="_blank">#10</a></span></summary>

### 🧾 Problem

Implement regex matching with support for '.' and '\*':

* '.' matches any single character.
* '\*' matches zero or more of the preceding element.

Return `true` if `text` matches `pattern`.

---

### 🔢 Input

* `text`: Target string
* `pattern`: Regex pattern

---

### 📤 Output

* Boolean value

---

### ✅ Example

```python
Input: text = "aab", pattern = "c*a*b"
Output: True

Input: text = "mississippi", pattern = "mis*is*p*."
Output: False
```

---

<details>
  <summary><strong>🧠 Solution Code (DP)</strong></summary>

```py

# Top Down aproach / Memoization

class Solution(object):
    def isMatch(self, s, p):
        # Create a memoization dictionary to store results of subproblems
        memo = {}
        # Define a helper function for recursion with indices i (for s) and j (for p)
        def dp(i, j):
            # If result is already computed, return from memo
            if (i, j) in memo:
                return memo[(i, j)]
            # If we've reached the end of the pattern
            if j == len(p):
                # Return True if s is also fully matched
                ans = i == len(s)
            else:
                # Check if the current character matches (if s is not exhausted)
                first_match = i < len(s) and (p[j] == s[i] or p[j] == '.')
                # If next character in p is '*', we have two choices:
                if (j + 1) < len(p) and p[j + 1] == '*':
                    # 1. Don't use the star (move p by 2)
                    #2. Use the star if first_match (move s by 1, keep p)
                    ans = dp(i, j + 2) or (first_match and dp(i + 1, j))
                else:
                    # If not '*', proceed only if first_match, move both by 1
                    ans = first_match and dp(i + 1, j + 1)
            # Store result in memo and return
            memo[(i, j)] = ans
            return ans
        # Start recursion from the beginning of s and p
        return dp(0, 0)
```

</details>

<details>
  <summary><strong>🧮 Complexity Analysis</strong></summary>

* **Time Complexity:** O(mn)
* **Space Complexity:** O(mn)

</details>

### 🧾 Problem

Given strings `s` and `t`, return the minimum window in `s` which contains all characters of `t`. Return empty string if no such window.

---

### 🔢 Input

* `s`: Target string
* `t`: Pattern string

---

### 📤 Output

* Smallest substring of `s` containing all chars of `t`

---

### ✅ Example

```python
Input: s = "ADOBECODEBANC", t = "ABC"
Output: "BANC"
```

---

<details>
  <summary><strong>🧠 Solution Code (Sliding Window)</strong></summary>

```py
class Solution(object):
    def minWindow(self, s, t):
        from collections import Counter
        need, window = Counter(t), {}
        have, needCount = 0, len(need)
        res = ""
        l = 0
        for r, ch in enumerate(s):
            window[ch] = window.get(ch, 0) + 1
            if ch in need and window[ch] == need[ch]:
                have += 1
            while have == needCount:
                substring = s[l:r+1]
                if not res or len(substring) < len(res):
                    res = substring
                window[s[l]] -= 1
                if s[l] in need and window[s[l]] < need[s[l]]:
                    have -= 1
                l += 1
        return res
```

### 🧾 Problem

Given a list of unique words, return all pairs `[i, j]` such that `words[i] + words[j]` forms a palindrome.

---

### 🔢 Input

* `words`: List of unique strings

---

### 📤 Output

* List of index pairs `[i, j]`

---

### ✅ Example

```python
Input: words = ["bat","tab","cat"]
Output: [[0,1],[1,0]]
```

---

<details>
  <summary><strong>🧠 Solution Code (Trie + Palindrome Checks)</strong></summary>

```py
class Solution(object):
    def palindromePairs(self, words):
        def is_pal(s):
            return s == s[::-1]
        trie = {}
        for i, w in enumerate(words):
            node = trie
            for j, ch in enumerate(reversed(w)):
                if is_pal(w[:len(w)-j]):
                    node.setdefault('pal', []).append(i)
                node = node.setdefault(ch, {})
            node['idx'] = i
        res = []
        for i, w in enumerate(words):
            node = trie
            for j, ch in enumerate(w):
                if 'idx' in node and node['idx'] != i and is_pal(w[j:]):
                    res.append([i, node['idx']])
                if ch not in node:
                    break
                node = node[ch]
            else:
                for j in node.get('pal', []):
                    if i != j:
                        res.append([i, j])
        return res
```

* **Time Complexity:** O(k²\*N + Σk) where k = avg length
* **Space Complexity:** O(Nk)

  <summary><span style="font-size:19px;"><strong>Word Ladder II - </strong><a href="https://leetcode.com/problems/word-ladder-ii/" target="_blank">#126</a></span></summary>

### 🧾 Problem

Given `beginWord`, `endWord`, and `wordList`, return all shortest transformation sequences from `beginWord` to `endWord`, where each step changes one letter and the word must be in the wordList.

---

### 🔢 Input

* `beginWord`, `endWord`: strings
* `wordList`: list of strings

---

### 📤 Output

* List of lists, each representing a shortest transformation chain

---

### ✅ Example

```python
Input: beginWord = "hit", endWord = "cog", wordList = ["hot","dot","dog","lot","log","cog"]
Output: [
  ["hit","hot","dot","dog","cog"],
  ["hit","hot","lot","log","cog"]
]
```

---

<details>
  <summary><strong>🧠 Solution Code (BFS + DFS)</strong></summary>

```py
class Solution:
    def findLadders(self, beginWord, endWord, wordList):
        prefix_d = defaultdict(list)
        for word in wordList:
            for i in range(0,len(word)):
                prefix_d[word[0:i]+"*"+word[i+1:]].append(word)
        
        order = {beginWord: []}
        queue = deque([beginWord])
        temp_q = deque()
        go_on = True
        end_list = []
        
        while queue and go_on:  # There is no node even added to temp_q
            temp_d = {}
            while queue:        # Pop every node on this level
                cur = queue.popleft()
                for i in range(0, len(cur)):
                    for j in prefix_d[cur[0:i]+"*"+cur[i+1:]]:
                        if j == endWord:
                            end_list.append(j)
                            go_on = False
                        if j not in order:
                            if j not in temp_d:
                                temp_d[j] = [cur]
                                temp_q.append(j)
                            else:
                                temp_d[j].append(cur)
            queue = temp_q
            temp_q = deque()
            order.update(temp_d)
        
        ret = []
        
        # DFS to restore the paths
        def dfs(path, node):
            path = path + [node]    # add the node(Deepcopy)
            if order[node] == []:
                ret.append(list(path[::-1]))
                return
            for i in order[node]:
                dfs(path, i)
        if endWord in order:
            dfs([], endWord)
        else:
            return []
        
        return ret
```

</details>

<details>
  <summary><strong>🧮 Complexity Analysis</strong></summary>

* **Time Complexity:** Roughly O(N \* 26^L)
* **Space Complexity:** Proportional to number of paths

</details>

<details>

```py
class Solution(object):
    def calculate(self, s):
        res, num, sign = 0, 0, 1
        stack = []
        for c in s:
            if c.isdigit():
                num = num*10 + int(c)
            elif c in '+-':
                res += sign*num
                num = 0
                sign = 1 if c=='+' else -1
            elif c == '(':
                stack.append(res)
                stack.append(sign)
                res, sign = 0, 1
            elif c == ')':
                res += sign*num
                num = 0
                res *= stack.pop()
                res += stack.pop()
        res += sign*num
        return res
```

<details>
  <summary><strong>🧠 Solution Code (BFS)</strong></summary>

```py
class Solution(object):
    def removeInvalidParentheses(self, s):
        def is_valid(st):
            bal = 0
            for c in st:
                if c=='(':
                    bal+=1
                elif c==')':
                    bal-=1
                    if bal<0:
                        return False
            return bal==0
        
        level = {s}
        while True:
            valid = [st for st in level if is_valid(st)]
            if valid:
                return valid
            next_level = set()
            for st in level:
                for i in range(len(st)):
                    if st[i] in '()':
                        next_level.add(st[:i] + st[i+1:])
            level = next_level
```

Given a string `s1`, determine if `s2` is a scramble of `s1`. Scrambling involves recursively partitioning and swapping.

---

### 🔢 Input

* `s1`, `s2`: Strings of equal length

---

### 📤 Output

* Boolean value

---

### ✅ Example

```python
Input: s1 = "great", s2 = "rgeat"
Output: True
```

---

<details>
  <summary><strong>🧠 Solution Code (Recursive + Memo)</strong></summary>

```py
class Solution(object):
    def isScramble(self, s1, s2, memo={}):
        if s1==s2:
            return True
        if (s1,s2) in memo:
            return memo[(s1,s2)]
        if sorted(s1)!=sorted(s2):
            return False
        n = len(s1)
        for i in range(1, n):
            if (self.isScramble(s1[:i], s2[:i], memo) and
                self.isScramble(s1[i:], s2[i:], memo)) or \
               (self.isScramble(s1[:i], s2[n-i:], memo) and
                self.isScramble(s1[i:], s2[:n-i], memo)):
                memo[(s1,s2)] = True
                return True
        memo[(s1,s2)] = False
        return False
```

<details>

```py
class Solution(object):
    def numDistinct(self, s, t):
        m, n = len(s), len(t)
        dp = [[0]*(n+1) for _ in range(m+1)]
        for i in range(m+1):
            dp[i][0] = 1
        for i in range(1, m+1):
            for j in range(1, n+1):
                dp[i][j] = dp[i-1][j]
                if s[i-1] == t[j-1]:
                    dp[i][j] += dp[i-1][j-1]
        return dp[m][n]
```

### 🧾 Problem

Maintain a data structure that supports:

* `addNum(val)`: add value from data stream,
* `getIntervals()`: return disjoint intervals of added values.

---

### 🔢 Input

* Sequence of `addNum` calls, then `getIntervals`

---

### 📤 Output

* List of disjoint intervals

---

### ✅ Example

```python
addNum(1)
addNum(3)
addNum(7)
getIntervals() -> [[1,1],[3,3],[7,7]]
addNum(2)
getIntervals() -> [[1,3],[7,7]]
```

---

<details>
  <summary><strong>🧠 Solution Code (SortedList)</strong></summary>

```py
from bisect import bisect_left
class SummaryRanges(object):
    def __init__(self):
        self.intervals = []
    def addNum(self, val):
        i = bisect_left(self.intervals, [val, val])
        if i != 0 and self.intervals[i-1][1] + 1 >= val:
            i -= 1
        start, end = val, val
        while i < len(self.intervals) and self.intervals[i][0] <= end + 1:
            start = min(start, self.intervals[i][0])
            end = max(end, self.intervals[i][1])
            self.intervals.pop(i)
        self.intervals.insert(i, [start, end])
    def getIntervals(self):
        return self.intervals
```

</details>

<details>
  <summary><strong>🧮 Complexity Analysis</strong></summary>

* **Time Complexity:** O(n) per insertion in worst case
* **Space Complexity:** O(n)
