### Check Prime Number

```python
# Write your solution here
def check_prime(num):
    l_num = num //2
    if num >1:
        for i in range(2,l_num+1):
            if num % l_num ==0 :
                return False
            else:
                return True
    else:
        return False
            

```

```python
check_prime(17)
```

### Check whether valid square number

```python
# Write your solution here
def check_valid_square(num):
    if num <1:
        return False
    for i in range(num//2):
        if i*i == num:
            return True

    return False
```

```python
check_valid_square(0)
```
