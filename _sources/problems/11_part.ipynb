**Input**

* `g`: Children's greed factors
* `s`: Cookie sizes

**Output**

* Maximum number of content children

**Example**

```py
Input: g = [1,2,3], s = [1,1]
Output: 1
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def findContentChildren(self, g: List[int], s: List[int]) -> int:
        g.sort()
        s.sort()
        i = j = 0
        while i < len(g) and j < len(s):
            if s[j] >= g[i]:
                i += 1
            j += 1
        return i
```

#### Problem 134: [Gas Station](https://leetcode.com/problems/gas-station/)

**Input**

* `gas`: Gas at each station
* `cost`: Cost to go to next station

**Output**

* Starting station index (or -1 if not possible)

**Example**

```py
Input: gas = [1,2,3,4,5], cost = [3,4,5,1,2]
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def canCompleteCircuit(self, gas: List[int], cost: List[int]) -> int:
        total, tank, start = 0, 0, 0
        for i in range(len(gas)):
            tank += gas[i] - cost[i]
            total += gas[i] - cost[i]
            if tank < 0:
                start = i + 1
                tank = 0
        return start if total >= 0 else -1
```

**Time:** O(n)

#### Problem 860: [Lemonade Change](https://leetcode.com/problems/lemonade-change/)

**Input**

* `bills`: List of 5, 10, or 20-dollar bills

* Can give change to all customers?

**Example**

```py
Input: bills = [5,5,5,10,20]
Output: True
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def lemonadeChange(self, bills: List[int]) -> bool:
        five = ten = 0
        for bill in bills:
            if bill == 5:
                five += 1
            elif bill == 10:
                if five == 0: return False
                five -= 1
                ten += 1
            else:
                if ten > 0 and five > 0:
                    ten -= 1
                    five -= 1
                elif five >= 3:
                    five -= 3
                else:
                    return False
        return True
```

#### Problem 874: [Walking Robot Simulation](https://leetcode.com/problems/walking-robot-simulation/)

**Input**

* `commands`: List of movement and turn instructions
* `obstacles`: Coordinates of obstacles

**Output**

* Maximum Euclidean distance squared from the origin

**Example**

```py
Input: commands = [4,-1,3], obstacles = []
Output: 25
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def robotSim(self, commands: List[int], obstacles: List[List[int]]) -> int:
        dx = [0, 1, 0, -1]
        dy = [1, 0, -1, 0]
        x = y = direction = 0
        obstacle_set = set(map(tuple, obstacles))
        max_dist = 0

        for cmd in commands:
            if cmd == -1:
                direction = (direction + 1) % 4
            elif cmd == -2:
                direction = (direction - 1) % 4
            else:
                for _ in range(cmd):
                    nx, ny = x + dx[direction], y + dy[direction]
                    if (nx, ny) not in obstacle_set:
                        x, y = nx, ny
                        max_dist = max(max_dist, x*x + y*y)
        return max_dist
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n + m)
**Space:** O(m), where m = number of obstacles

</details>

#### Problem 2299: [Strong Password Checker II](https://leetcode.com/problems/strong-password-checker-ii/)

**Input**

* `password`: A string

**Output**

* Is the password strong?

**Example**

```py
Input: password = "IloveLe3tcode!"
Output: True
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def strongPasswordCheckerII(self, password: str) -> bool:
        if len(password) < 8:
            return False

        has_lower = has_upper = has_digit = has_special = False
        special_chars = set("!@#$%^&*()-+")

        for i, ch in enumerate(password):
            if i > 0 and password[i] == password[i - 1]:
                return False
            if ch.islower(): has_lower = True
            elif ch.isupper(): has_upper = True
            elif ch.isdigit(): has_digit = True
            elif ch in special_chars: has_special = True

        return all([has_lower, has_upper, has_digit, has_special])
```

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(1)

#### Problem 763: [Partition Labels](https://leetcode.com/problems/partition-labels/)

**Input**

* `s`: A string

**Output**

* List of integers representing the size of each partition

**Example**

```python
Input: s = "ababcbacadefegdehijhklij"
Output: [9,7,8]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def partitionLabels(self, s: str) -> List[int]:
        last = {c: i for i, c in enumerate(s)}
        j = anchor = 0
        res = []

        for i, ch in enumerate(s):
            j = max(j, last[ch])
            if i == j:
                res.append(i - anchor + 1)
                anchor = i + 1
        return res
```

#### Problem 45: [Jump Game II](https://leetcode.com/problems/jump-game-ii/)

**Input**

* `nums`: List of non-negative integers

**Output**

* Minimum number of jumps to reach the last index

**Example**

```py
Input: nums = [2,3,1,1,4]
Output: 2
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def jump(self, nums: List[int]) -> int:
        jumps = cur_end = cur_farthest = 0
        for i in range(len(nums) - 1):
            cur_farthest = max(cur_farthest, i + nums[i])
            if i == cur_end:
                jumps += 1
                cur_end = cur_farthest
        return jumps
```

#### Problem 135: [Candy](https://leetcode.com/problems/candy/)

**Input**

* `ratings`: List of children's ratings

**Output**

* Minimum number of candies to distribute

**Example**

```python
Input: ratings = [1,0,2]
Output: 5
Explanation: You must give to the second child only 1 candy, then to the first child 2 candies and to the third child 2 candies.
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def candy(self, ratings: List[int]) -> int:
        n = len(ratings)
        candies = [1] * n

        for i in range(1, n):
            if ratings[i] > ratings[i - 1]:
                candies[i] = candies[i - 1] + 1

        for i in range(n - 2, -1, -1):
            if ratings[i] > ratings[i + 1]:
                candies[i] = max(candies[i], candies[i + 1] + 1)

        return sum(candies)
```

</details>

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(n)

#### Problem 354: [Russian Doll Envelopes](https://leetcode.com/problems/russian-doll-envelopes/)

**Input**

* `envelopes`: List of envelopes as `[width, height]`

**Output**

* Maximum number of envelopes you can Russian doll

**Example**

```python
Input: envelopes = [[5,4],[6,4],[6,7],[2,3]]
Output: 3
Explanation: The maximum number of envelopes you can Russian doll is 3 ([2,3] => [5,4] => [6,7]).
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
import bisect

class Solution:
    def maxEnvelopes(self, envelopes: List[List[int]]) -> int:
        envelopes.sort(key=lambda x: (x[0], -x[1]))
        dp = []

        for _, h in envelopes:
            idx = bisect.bisect_left(dp, h)
            if idx == len(dp):
                dp.append(h)
            else:
                dp[idx] = h

        return len(dp)
```

**Time:** O(n log n)
**Space:** O(n)

#### Problem 1696: [Jump Game VI](https://leetcode.com/problems/jump-game-vi/)

**Input**

* `nums`: List of integers
* `k`: Maximum number of steps you can jump forward

**Output**

* Maximum score to reach the last index

**Example**

```python
Input: nums = [1,-1,-2,4,-7,3], k = 2  
Output: 7
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import deque

class Solution:
    def maxResult(self, nums: List[int], k: int) -> int:
        dq = deque()
        dp = [0] * len(nums)
        dp[0] = nums[0]
        dq.append(0)

        for i in range(1, len(nums)):
            # Remove indices outside the window
            while dq and dq[0] < i - k:
                dq.popleft()
            dp[i] = nums[i] + dp[dq[0]]
            while dq and dp[i] >= dp[dq[-1]]:
                dq.pop()
            dq.append(i)

        return dp[-1]
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(n)

</details>

#### Problem 659: [Split Array Into Consecutive Subsequences](https://leetcode.com/problems/split-array-into-consecutive-subsequences/)

**Input**

* `nums`: Sorted list of integers

**Output**

* Whether you can split the array into consecutive subsequences of length ≥ 3

**Example**

```python
Input: nums = [1,2,3,3,4,5]  
Output: True
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
from collections import Counter, defaultdict

class Solution:
    def isPossible(self, nums: List[int]) -> bool:
        freq = Counter(nums)
        append_freq = defaultdict(int)

        for num in nums:
            if freq[num] == 0:
                continue
            if append_freq[num - 1] > 0:
                append_freq[num - 1] -= 1
                append_freq[num] += 1
            elif freq[num + 1] > 0 and freq[num + 2] > 0:
                freq[num + 1] -= 1
                freq[num + 2] -= 1
                append_freq[num + 2] += 1
            else:
                return False
            freq[num] -= 1

        return True
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)
**Space:** O(n)

* Minimum integer eating speed `k` such that all bananas are eaten in `h` hours

**Example**

```python
Input: piles = [3,6,7,11], h = 8  
Output: 4
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
import math

class Solution:
    def minEatingSpeed(self, piles: List[int], h: int) -> int:
        left, right = 1, max(piles)

        while left < right:
            mid = (left + right) // 2
            total = sum(math.ceil(p / mid) for p in piles)
            if total > h:
                left = mid + 1
            else:
                right = mid
        return left
```

#### Problem 1478: [Allocate Mailboxes](https://leetcode.com/problems/allocate-mailboxes/)

**Input**

* `houses`: List of house positions
* `k`: Number of mailboxes to place

**Output**

* Minimum total distance between each house and its nearest mailbox

**Example**

```python
Input: houses = [1,4,8,10,20], k = 3
Output: 5
Explanation: Allocate mailboxes at positions 1, 10 and 20.
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def minDistance(self, houses: List[int], k: int) -> int:
        houses.sort()
        n = len(houses)
        # cost[i][j] = cost to place 1 mailbox for houses[i..j]
        cost = [[0] * n for _ in range(n)]
        for i in range(n):
            for j in range(i, n):
                median = houses[(i + j) // 2]
                cost[i][j] = sum(abs(houses[x] - median) for x in range(i, j + 1))
        
        # dp[i][k] = min cost to serve houses[0..i] with k mailboxes
        dp = [[float('inf')] * (k + 1) for _ in range(n)]
        for i in range(n):
            dp[i][1] = cost[0][i]
        
        for mailboxes in range(2, k + 1):
            for i in range(n):
                for j in range(i):
                    dp[i][mailboxes] = min(dp[i][mailboxes], dp[j][mailboxes - 1] + cost[j + 1][i])
        
        return dp[n - 1][k]
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n³) (can be optimized to O(n²k))
**Space:** O(n² + nk)

</details>

#### Problem 871: [Minimum Number of Refueling Stops](https://leetcode.com/problems/minimum-number-of-refueling-stops/)

**Input**

* `target`: Distance to reach
* `startFuel`: Initial fuel
* `stations`: List of fuel stations (position, fuel)

**Output**

* Minimum number of refueling stops to reach the target, or -1 if impossible

**Example**

```python
Input: target = 100, startFuel = 10, stations = [[10,60],[20,30],[30,30],[60,40]]
Output: 2
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
import heapq

class Solution:
    def minRefuelStops(self, target: int, startFuel: int, stations: List[List[int]]) -> int:
        pq = []
        stations.append([target, 0])
        fuel, prev, refuels = startFuel, 0, 0
        
        for location, capacity in stations:
            fuel -= location - prev
            while pq and fuel < 0:
                fuel += -heapq.heappop(pq)
                refuels += 1
            if fuel < 0:
                return -1
            heapq.heappush(pq, -capacity)
            prev = location
        return refuels
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n log n)
**Space:** O(n)

#### Problem 410: [Split Array Largest Sum](https://leetcode.com/problems/split-array-largest-sum/)

**Input**

* `nums`: List of integers
* `k`: Number of subarrays to split into

**Output**

* The minimum largest sum among these `k` subarrays

**Example**

```python
Input: nums = [7,2,5,10,8], k = 2
Output: 18
Explanation: Split into [7,2,5] and [10,8], largest sum is 18.
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def splitArray(self, nums: List[int], k: int) -> int:
        def canSplit(mid):
            total, count = 0, 1
            for num in nums:
                if total + num > mid:
                    count += 1
                    total = num
                else:
                    total += num
            return count <= k
        
        left, right = max(nums), sum(nums)
        while left < right:
            mid = (left + right) // 2
            if canSplit(mid):
                right = mid
            else:
                left = mid + 1
        return left
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n log(sum(nums)))
**Space:** O(1)

</details>

#### Aggressive Cows – SPOJ (AGGRCOW)

**Input**

* `stalls`: Positions of stalls (unsorted)
* `k`: Number of cows to place

**Output**

* Maximum minimum distance between any two cows

**Example**

```python
Input: stalls = [1, 2, 8, 4, 9], k = 3
Output: 3
```

**Explanation**

Place cows at positions 1, 4, and 8 — the minimum distance is 3.

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def aggressiveCows(self, stalls: List[int], k: int) -> int:
        stalls.sort()
        
        def canPlaceCows(distance):
            count, last_pos = 1, stalls[0]
            for i in range(1, len(stalls)):
                if stalls[i] - last_pos >= distance:
                    count += 1
                    last_pos = stalls[i]
                if count == k:
                    return True
            return False
        
        left, right, ans = 1, stalls[-1] - stalls[0], 0
        while left <= right:
            mid = (left + right) // 2
            if canPlaceCows(mid):
                ans = mid
                left = mid + 1
            else:
                right = mid - 1
        return ans
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n log(max\_dist))
**Space:** O(1)

</details>

**Explanation**

You can shoot at x = 6 (covers \[2,8] and \[1,6]) and x = 12 (covers \[7,12] and \[10,16]).

#### Problem 842: [Split Array into Fibonacci Sequence](https://leetcode.com/problems/split-array-into-fibonacci-sequence/)

**Input**

* `S`: A string of digits

**Output**

* List of integers forming a valid Fibonacci-like sequence, or empty list

**Example**

```python
Input: "123456579"
Output: [123, 456, 579]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def splitIntoFibonacci(self, S: str) -> List[int]:
        def backtrack(index, path):
            if index == len(S) and len(path) >= 3:
                return path
            for i in range(index + 1, len(S) + 1):
                if S[index] == '0' and i > index + 1:
                    break
                num = int(S[index:i])
                if num > 2**31 - 1:
                    break
                if len(path) < 2 or path[-1] + path[-2] == num:
                    res = backtrack(i, path + [num])
                    if res:
                        return res
                elif len(path) >= 2 and path[-1] + path[-2] < num:
                    break
            return []
        
        return backtrack(0, [])
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(2ⁿ) worst case due to recursion
**Space:** O(n) call stack

</details>

#### Problem 2517: [Maximize Sweetness of Chocolate](https://leetcode.com/problems/maximum-tastiness-of-candy-boxes/)

**Input**

* `price`: List of prices of candies (treated as sweetness levels)
* `k`: Number of boxes (subsets) to pick

**Output**

* Maximum possible *minimum difference* of sweetness between picked boxes

**Example**

```python
Input: price = [13, 5, 1, 8, 21, 2], k = 3
Output: 8
```

**Explanation**

Choose chocolates such that difference between any two chosen is at least 8.

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def maximumTastiness(self, price: List[int], k: int) -> int:
        price.sort()
        
        def canSelect(min_diff):
            count = 1
            last = price[0]
            for i in range(1, len(price)):
                if price[i] - last >= min_diff:
                    count += 1
                    last = price[i]
                if count >= k:
                    return True
            return False
        
        left, right = 0, price[-1] - price[0]
        ans = 0
        while left <= right:
            mid = (left + right) // 2
            if canSelect(mid):
                ans = mid
                left = mid + 1
            else:
                right = mid - 1
        return ans
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n log(max\_diff))
**Space:** O(1)

</details>

#### Problem 1547: [Minimum Cost to Cut a Stick](https://leetcode.com/problems/minimum-cost-to-cut-a-stick/)

**Input**

* `n`: Total length of the stick
* `cuts`: List of positions to cut

**Output**

* Minimum total cost to perform all cuts

**Example**

```python
Input: n = 7, cuts = [1, 3, 4, 5]
Output: 16
```

**Explanation**

Use dynamic programming to find the best order of cuts.

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def minCost(self, n: int, cuts: List[int]) -> int:
        cuts = [0] + sorted(cuts) + [n]
        m = len(cuts)
        dp = [[0] * m for _ in range(m)]
        
        for l in range(2, m):
            for i in range(m - l):
                j = i + l
                dp[i][j] = min(dp[i][k] + dp[k][j] for k in range(i+1, j)) + cuts[j] - cuts[i]
        
        return dp[0][m - 1]
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n²)
**Space:** O(n²)

</details>

#### Problem 1283: [Find the Smallest Divisor Given a Threshold](https://leetcode.com/problems/find-the-smallest-divisor-given-a-threshold/)

**Input**

* `nums`: List of integers
* `threshold`: Maximum allowed total sum after division

**Output**

* The smallest possible divisor such that the **sum of all elements divided (rounded up)** is ≤ `threshold`.

---

**Example**

```python
Input: nums = [1, 2, 5, 9], threshold = 6
Output: 5
```

**Explanation**

We want the smallest divisor such that:

* `ceil(1/d) + ceil(2/d) + ceil(5/d) + ceil(9/d) ≤ 6`
* Try divisors from 1 up to max(nums), use binary search to minimize.

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
import math

class Solution:
    def smallestDivisor(self, nums: List[int], threshold: int) -> int:
        left, right = 1, max(nums)

        while left < right:
            mid = (left + right) // 2
            total = sum(math.ceil(num / mid) for num in nums)
            if total > threshold:
                left = mid + 1
            else:
                right = mid

        return left
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n × log(max(nums)))
**Space:** O(1)

</details>

#### Problem 422: [Valid Word Square](https://leetcode.com/problems/valid-word-square/) *(Greedy + Backtracking)*

**Input**

* `words`: List of strings

**Output**

* Boolean value indicating if words form a valid word square

**Example**

```python
Input: ["abcd", "bnrt", "crmy", "dtye"]
Output: True
```

**Explanation**

Each row and column must form the same word.

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def validWordSquare(self, words: List[str]) -> bool:
        for i in range(len(words)):
            for j in range(len(words[i])):
                if j >= len(words) or i >= len(words[j]) or words[i][j] != words[j][i]:
                    return False
        return True
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n²)
**Space:** O(1)
