## Linked Lists

### Easy

#### 1. Reverse a Singly Linked List

**Input**

* `head`: Head node of the singly linked list

---

**Output**

* Head node of the reversed list

---

**Example**

```python
Input: 1 → 2 → 3 → 4 → 5  
Output: 5 → 4 → 3 → 2 → 1
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def reverseList(head):
    prev = None
    curr = head
    while curr:
        nxt = curr.next
        curr.next = prev
        prev = curr
        curr = nxt
    return prev
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)  
**Space:** O(1)

</details>

#### Merge Two Sorted Linked Lists

**Input**

* `l1`: Head of the first sorted linked list
* `l2`: Head of the second sorted linked list

---

**Output**

* Head of the merged sorted linked list

---

**Example**

```python
Input: l1 = 1 → 3 → 5, l2 = 2 → 4 → 6  
Output: 1 → 2 → 3 → 4 → 5 → 6
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def mergeTwoLists(l1, l2):
    dummy = ListNode()
    curr = dummy
    while l1 and l2:
        if l1.val < l2.val:
            curr.next = l1
            l1 = l1.next
        else:
            curr.next = l2
            l2 = l2.next
        curr = curr.next
    curr.next = l1 or l2
    return dummy.next
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n + m)  
**Space:** O(1)

</details>

#### Remove Duplicates from a Sorted Linked List

**Input**

* `head`: Head of a sorted linked list

---

**Output**

* Head of the list with duplicates removed

---

**Example**

```python
Input: 1 → 1 → 2 → 3 → 3  
Output: 1 → 2 → 3
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def deleteDuplicates(head):
    curr = head
    while curr and curr.next:
        if curr.val == curr.next.val:
            curr.next = curr.next.next
        else:
            curr = curr.next
    return head
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)  
**Space:** O(1)

</details>

#### Detect a Cycle in a Linked List

**Input**

* `head`: Head of the linked list

---

**Output**

* `True` if a cycle exists, otherwise `False`

---

**Example**

```python
Input: 1 → 2 → 3 → 4 → 2 (cycle back to node 2)  
Output: True
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def hasCycle(head):
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return True
    return False
```

#### Palindrome Linked List

**Input**

* `head`: Head of the linked list

---

**Output**

* `True` if the list is a palindrome, otherwise `False`

---

**Example**

```python
Input: 1 → 2 → 2 → 1  
Output: True
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def isPalindrome(head):
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    prev = None
    while slow:
        nxt = slow.next
        slow.next = prev
        prev = slow
        slow = nxt
    while prev:
        if prev.val != head.val:
            return False
        prev = prev.next
        head = head.next
    return True
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)  
**Space:** O(1)

#### Intersection of Two Linked Lists

**Input**

* `headA`: Head of the first linked list
* `headB`: Head of the second linked list

---

**Output**

* Node where the two lists intersect, or `None`

---

**Example**

```python
Input: headA = 1 → 2 → 3 → 6 → 7, headB = 4 → 5 → 6 → 7  
Output: 6
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def getIntersectionNode(headA, headB):
    if not headA or not headB:
        return None
    a, b = headA, headB
    while a != b:
        a = a.next if a else headB
        b = b.next if b else headA
    return a
```

#### Add Two Numbers

**Input**

* `l1`: Head of the first linked list (number in reverse order)
* `l2`: Head of the second linked list (number in reverse order)

---

**Output**

* Head of the linked list representing the sum

---

**Example**

```python
Input: l1 = 2 → 4 → 3, l2 = 5 → 6 → 4  
Output: 7 → 0 → 8  (342 + 465 = 807)
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def addTwoNumbers(l1, l2):
    dummy = ListNode()
    curr = dummy
    carry = 0
    while l1 or l2 or carry:
        val1 = l1.val if l1 else 0
        val2 = l2.val if l2 else 0
        total = val1 + val2 + carry
        carry = total // 10
        curr.next = ListNode(total % 10)
        curr = curr.next
        l1 = l1.next if l1 else None
        l2 = l2.next if l2 else None
    return dummy.next
```

</details>

#### Rotate List

**Input**

* `head`: Head of the linked list
* `k`: Number of rotations

---

**Output**

* Head of the rotated list

---

**Example**

```python
Input: head = 1 → 2 → 3 → 4 → 5, k = 2  
Output: 4 → 5 → 1 → 2 → 3
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def rotateRight(head, k):
    if not head or not head.next or k == 0:
        return head
    length = 1
    tail = head
    while tail.next:
        tail = tail.next
        length += 1
    k = k % length
    if k == 0:
        return head
    slow = fast = head
    for _ in range(k):
        fast = fast.next
    while fast.next:
        slow = slow.next
        fast = fast.next
    new_head = slow.next
    slow.next = None
    fast.next = head
    return new_head
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)  
**Space:** O(1)

</details>

#### Partition List

**Input**

* `head`: Head of the linked list
* `x`: Partition value

---

**Output**

* Head of the partitioned list (nodes < x before nodes >= x)

---

**Example**

```python
Input: head = 1 → 4 → 3 → 2 → 5 → 2, x = 3  
Output: 1 → 2 → 2 → 4 → 3 → 5
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def partition(head, x):
    before = before_head = ListNode()
    after = after_head = ListNode()
    while head:
        if head.val < x:
            before.next = head
            before = before.next
        else:
            after.next = head
            after = after.next
        head = head.next
    after.next = None
    before.next = after_head.next
    return before_head.next
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)  
**Space:** O(1)

</details>

#### Reverse Linked List II

**Input**

* `head`: Head of the linked list
* `left`: Start position
* `right`: End position

---

**Output**

* Head of the list with nodes from position left to right reversed

---

**Example**

```python
Input: head = 1 → 2 → 3 → 4 → 5, left = 2, right = 4  
Output: 1 → 4 → 3 → 2 → 5
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def reverseBetween(head, left, right):
    dummy = ListNode(0, head)
    prev = dummy
    for _ in range(left - 1):
        prev = prev.next
    curr = prev.next
    for _ in range(right - left):
        nxt = curr.next
        curr.next = nxt.next
        nxt.next = prev.next
        prev.next = nxt
    return dummy.next
```

#### Reorder List

**Input**

* `head`: Head of the linked list

---

**Output**

* None (reorder the list in-place: L0 → Ln → L1 → Ln-1 → ...)

---

**Example**

```python
Input: 1 → 2 → 3 → 4  
Output: 1 → 4 → 2 → 3
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def reorderList(head):
    if not head or not head.next:
        return
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    prev, curr = None, slow
    while curr:
        nxt = curr.next
        curr.next = prev
        prev = curr
        curr = nxt
    first, second = head, prev
    while second.next:
        temp1, temp2 = first.next, second.next
        first.next = second
        second.next = temp1
        first, second = temp1, temp2
```

#### Linked List Cycle II

**Input**

* `head`: Head of the linked list

---

* Node where the cycle begins, or `None`

**Example**

```python
Input: 1 → 2 → 3 → 4 → 2 (cycle begins at node 2)  
Output: 2
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def detectCycle(head):
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            slow = head
            while slow != fast:
                slow = slow.next
                fast = fast.next
            return slow
    return None
```

#### Sort List

**Input**

* `head`: Head of the linked list

---

**Output**

* Head of the sorted list

---

**Example**

```python
Input: 4 → 2 → 1 → 3  
Output: 1 → 2 → 3 → 4
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def sortList(head):
    if not head or not head.next:
        return head
    slow, fast = head, head.next
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    mid = slow.next
    slow.next = None
    left = sortList(head)
    right = sortList(mid)
    dummy = ListNode()
    curr = dummy
    while left and right:
        if left.val < right.val:
            curr.next = left
            left = left.next
        else:
            curr.next = right
            right = right.next
        curr = curr.next
    curr.next = left or right
    return dummy.next
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n log n)  
**Space:** O(log n)

</details>

#### Merge k Sorted Lists

**Input**

* `lists`: Array of heads of k sorted linked lists

---

**Output**

* Head of the merged sorted linked list

---

**Example**

```python
Input: [[1 → 4 → 5], [1 → 3 → 4], [2 → 6]]  
Output: 1 → 1 → 2 → 3 → 4 → 4 → 5 → 6
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
import heapq

def mergeKLists(lists):
    heap = []
    for i, node in enumerate(lists):
        if node:
            heapq.heappush(heap, (node.val, i, node))
    dummy = ListNode()
    curr = dummy
    while heap:
        val, i, node = heapq.heappop(heap)
        curr.next = node
        curr = curr.next
        if node.next:
            heapq.heappush(heap, (node.next.val, i, node.next))
    return dummy.next
```

#### Reverse Nodes in k-Group

**Input**

* `head`: Head of the linked list
* `k`: Group size

---

**Output**

* Head of the list with nodes reversed in k-groups

---

**Example**

```python
Input: head = 1 → 2 → 3 → 4 → 5, k = 2  
Output: 2 → 1 → 4 → 3 → 5
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def reverseKGroup(head, k):
    def reverse(head, k):
        prev = None
        curr = head
        for _ in range(k):
            nxt = curr.next
            curr.next = prev
            prev = curr
            curr = nxt
        return prev
    
    count = 0
    node = head
    while node and count < k:
        node = node.next
        count += 1
    if count < k:
        return head
    new_head = reverse(head, k)
    head.next = reverseKGroup(node, k)
    return new_head
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)  
**Space:** O(n/k)

</details>

#### Copy List with Random Pointer

**Input**

* `head`: Head of the linked list with a random pointer

---

**Output**

* Deep copy of the list

---

**Example**

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def copyRandomList(head):
    if not head:
        return None
    curr = head
    while curr:
        new_node = Node(curr.val, curr.next, None)
        curr.next = new_node
        curr = new_node.next
    curr = head
    while curr:
        if curr.random:
            curr.next.random = curr.random.next
        curr = curr.next.next
    old_list = head
    new_list = head.next
    new_head = head.next
    while old_list:
        old_list.next = old_list.next.next
        new_list.next = new_list.next.next if new_list.next else None
        old_list = old_list.next
        new_list = new_list.next
    return new_head
```

#### LRU Cache

**Input**

* `capacity`: Maximum capacity of the cache

---

**Output**

* Implementation of LRU cache with `get` and `put` operations

---

**Example**

```python
cache = LRUCache(2)
cache.put(1, 1)
cache.put(2, 2)
cache.get(1)  # returns 1
cache.put(3, 3)  # evicts key 2
cache.get(2)  # returns -1 (not found)
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class LRUCache:
    def __init__(self, capacity):
        self.capacity = capacity
        self.cache = {}
        self.head = Node(0, 0)
        self.tail = Node(0, 0)
        self.head.next = self.tail
        self.tail.prev = self.head
    
    def get(self, key):
        if key in self.cache:
            node = self.cache[key]
            self._remove(node)
            self._add(node)
            return node.val
        return -1
    
    def put(self, key, value):
        if key in self.cache:
            self._remove(self.cache[key])
        node = Node(key, value)
        self._add(node)
        self.cache[key] = node
        if len(self.cache) > self.capacity:
            lru = self.head.next
            self._remove(lru)
            del self.cache[lru.key]
    
    def _remove(self, node):
        node.prev.next = node.next
        node.next.prev = node.prev
    
    def _add(self, node):
        node.prev = self.tail.prev
        node.next = self.tail
        self.tail.prev.next = node
        self.tail.prev = node
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(1) for get and put  
**Space:** O(capacity)

</details>

#### Flatten a Multilevel Doubly Linked List

**Input**

* `head`: Head of a multilevel doubly linked list

---

**Output**

* Head of the flattened list

---

**Example**

```python
Input: 1 → 2 → 3 → 4 → 5 → 6 (with child pointers)  
Output: 1 → 2 → 3 → 7 → 8 → 11 → 12 → 9 → 10 → 4 → 5 → 6
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def flatten(head):
    if not head:
        return None
    stack = [head]
    prev = None
    while stack:
        curr = stack.pop()
        if prev:
            prev.next = curr
            curr.prev = prev
        if curr.next:
            stack.append(curr.next)
        if curr.child:
            stack.append(curr.child)
            curr.child = None
        prev = curr
    return head
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)  
**Space:** O(n)

</details>

#### Design Skiplist

**Input**

* None (constructor for Skiplist)

---

**Output**

* Implementation of a skiplist with `search`, `add`, and `erase` operations

---

**Example**

```python
skiplist = Skiplist()
skiplist.add(1)
skiplist.add(2)
skiplist.search(1)  # returns True
skiplist.erase(1)
skiplist.search(1)  # returns False
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
import random

class Skiplist:
    def __init__(self):
        self.head = Node(-1)
        self.level = 0
    
    def search(self, target):
        curr = self.head
        for i in range(self.level, -1, -1):
            while curr.forward[i] and curr.forward[i].val < target:
                curr = curr.forward[i]
        curr = curr.forward[0]
        return curr and curr.val == target
    
    def add(self, num):
        update = [None] * 32
        curr = self.head
        for i in range(self.level, -1, -1):
            while curr.forward[i] and curr.forward[i].val < num:
                curr = curr.forward[i]
            update[i] = curr
        level = 0
        while random.random() < 0.5 and level < 31:
            level += 1
        if level > self.level:
            for i in range(self.level + 1, level + 1):
                update[i] = self.head
            self.level = level
        new_node = Node(num, [None] * (level + 1))
        for i in range(level + 1):
            new_node.forward[i] = update[i].forward[i]
            update[i].forward[i] = new_node
    
    def erase(self, num):
        update = [None] * 32
        curr = self.head
        for i in range(self.level, -1, -1):
            while curr.forward[i] and curr.forward[i].val < num:
                curr = curr.forward[i]
            update[i] = curr
        curr = curr.forward[0]
        if not curr or curr.val != num:
            return False
        for i in range(self.level + 1):
            if update[i].forward[i] != curr:
                break
            update[i].forward[i] = curr.forward[i]
        while self.level > 0 and not self.head.forward[self.level]:
            self.level -= 1
        return True
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(log n) on average  
**Space:** O(n)

</details>

**Output**

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class AllOne:
    def __init__(self):
        self.key_count = {}
        self.count_keys = defaultdict(set)
        self.min_count = 0
        self.max_count = 0
    
    def inc(self, key):
        count = self.key_count.get(key, 0)
        self.key_count[key] = count + 1
        if count > 0:
            self.count_keys[count].discard(key)
            if not self.count_keys[count] and count == self.min_count:
                self.min_count = count + 1
        else:
            self.min_count = 1
        self.count_keys[count + 1].add(key)
        self.max_count = max(self.max_count, count + 1)
    
    def dec(self, key):
        count = self.key_count[key]
        self.count_keys[count].discard(key)
        if count == 1:
            del self.key_count[key]
        else:
            self.key_count[key] = count - 1
            self.count_keys[count - 1].add(key)
        if not self.count_keys[count]:
            if count == self.max_count:
                self.max_count = count - 1
            if count == self.min_count:
                self.min_count = count - 1 if self.count_keys[count - 1] else 0
    
    def getMaxKey(self):
        return next(iter(self.count_keys[self.max_count])) if self.max_count else ""
    
    def getMinKey(self):
        return next(iter(self.count_keys[self.min_count])) if self.min_count else ""
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(1) for all operations  
**Space:** O(n)

#### Design HashMap with Linked List

**Input**

* None (constructor)

---

**Output**

* Implementation of HashMap using linked lists

---

**Example**

```python
hashMap = MyHashMap()
hashMap.put(1, 1)
hashMap.put(2, 2)
hashMap.get(1)  # returns 1
hashMap.remove(1)
hashMap.get(1)  # returns -1
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class MyHashMap:
    def __init__(self):
        self.size = 1000
        self.buckets = [[] for _ in range(self.size)]
    
    def _hash(self, key):
        return key % self.size
    
    def put(self, key, value):
        idx = self._hash(key)
        for i, (k, v) in enumerate(self.buckets[idx]):
            if k == key:
                self.buckets[idx][i] = (key, value)
                return
        self.buckets[idx].append((key, value))
    
    def get(self, key):
        idx = self._hash(key)
        for k, v in self.buckets[idx]:
            if k == key:
                return v
        return -1
    
    def remove(self, key):
        idx = self._hash(key)
        for i, (k, v) in enumerate(self.buckets[idx]):
            if k == key:
                del self.buckets[idx][i]
                return
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n/k) on average  
**Space:** O(k + m)

</details>

#### Convert Sorted List to Binary Search Tree

**Input**

* `head`: Head of a sorted linked list

---

**Output**

* Root of the balanced BST

---

**Example**

```python
Input: -10 → -3 → 0 → 5 → 9  
Output: BST with root 0
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def sortedListToBST(head):
    if not head:
        return None
    if not head.next:
        return TreeNode(head.val)
    slow = fast = head
    prev = None
    while fast and fast.next:
        prev = slow
        slow = slow.next
        fast = fast.next.next
    if prev:
        prev.next = None
    root = TreeNode(slow.val)
    root.left = sortedListToBST(head)
    root.right = sortedListToBST(slow.next)
    return root
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n log n)  
**Space:** O(log n)

</details>

#### Split Linked List in Parts

**Input**

* `head`: Head of the linked list
* `k`: Number of parts

---

**Output**

* Array of k parts (as balanced as possible)

---

**Example**

```python
Input: head = 1 → 2 → 3, k = 5  
Output: [[1], [2], [3], [], []]
```

---

<details>
  <summary><strong>Solution Code</strong></summary>

```python
def splitListToParts(head, k):
    length = 0
    curr = head
    while curr:
        length += 1
        curr = curr.next
    part_size, extra = divmod(length, k)
    result = []
    curr = head
    for i in range(k):
        result.append(curr)
        size = part_size + (1 if i < extra else 0)
        for j in range(size - 1):
            if curr:
                curr = curr.next
        if curr:
            nxt = curr.next
            curr.next = None
            curr = nxt
    return result
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n + k)  
**Space:** O(k)

</details>
