**Input**

* `nums`: Array of integers where every element appears twice except one

**Output**

* The single element that appears only once

**Example**

```python
Input: nums = [4,1,2,1,2]
Output: 4
```

<details>

#### Problem 191: [Number of 1 Bits](https://leetcode.com/problems/number-of-1-bits/)

**Input**

* `n`: An integer (unsigned 32-bit)

**Output**

* Number of 1 bits in binary representation

**Example**

```python
Input: n = 11
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def hammingWeight(self, n):
        count = 0
        while n:
            count += n & 1
            n >>= 1
        return count
```

```python
class Solution:
    def singleNumber(self, nums):
        ones, twos = 0, 0
        for num in nums:
            ones = (ones ^ num) & ~twos
            twos = (twos ^ num) & ~ones
        return ones
```

#### Problem 201: [Bitwise AND of Numbers Range](https://leetcode.com/problems/bitwise-and-of-numbers-range/)

* Bitwise AND of all numbers in range

**Example**

```python
Input: left = 5, right = 7
Output: 4
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def rangeBitwiseAnd(self, left, right):
        shift = 0
        while left != right:
            left >>= 1
            right >>= 1
            shift += 1
        return left << shift
```

#### Problem 260: [Single Number III](https://leetcode.com/problems/single-number-iii/)

**Input**

* `nums`: Array where all elements appear twice except two

**Output**

* The two elements that appear only once

**Example**

```python
Input: nums = [1,2,1,3,2,5]
Output: [3,5]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def singleNumber(self, nums):
        xor = 0
        for num in nums:
            xor ^= num

        diff_bit = xor & -xor
        result = [0, 0]

        for num in nums:
            if num & diff_bit:
                result[0] ^= num
            else:
                result[1] ^= num

        return result
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)  
**Space:** O(1)

</details>

#### Problem 318: [Maximum Product of Word Lengths](https://leetcode.com/problems/maximum-product-of-word-lengths/)

**Input**

* `words`: Array of strings

**Output**

* Maximum product of lengths of two words with no common letters

**Example**

```python
Input: words = ["abcw","baz","foo","bar","xtfn","abcdef"]
Output: 16
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def maxProduct(self, words):
        bits = [0] * len(words)
        for i, word in enumerate(words):
            for char in word:
                bits[i] |= 1 << (ord(char) - ord('a'))

        max_prod = 0
        for i in range(len(words)):
            for j in range(i+1, len(words)):
                if bits[i] & bits[j] == 0:
                    max_prod = max(max_prod, len(words[i]) * len(words[j]))

        return max_prod
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n²)  
**Space:** O(n)

</details>

#### Problem 371: [Sum of Two Integers](https://leetcode.com/problems/sum-of-two-integers/)

* Sum without using + or - operators

**Example**

```python
Input: a = 1, b = 2
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def getSum(self, a, b):
        mask = 0xFFFFFFFF
        while b & mask:
            carry = (a & b) << 1
            a = a ^ b
            b = carry
        return a & mask if b > 0 else a
```

#### Problem 393: [UTF-8 Validation](https://leetcode.com/problems/utf-8-validation/)

**Input**

* `data`: Array of integers representing bytes

**Output**

* True if valid UTF-8 encoding

**Example**

```python
Input: data = [197,130,1]
Output: True
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def validUtf8(self, data):
        n_bytes = 0

        for num in data:
            if n_bytes == 0:
                if num >> 5 == 0b110:
                    n_bytes = 1
                elif num >> 4 == 0b1110:
                    n_bytes = 2
                elif num >> 3 == 0b11110:
                    n_bytes = 3
                elif num >> 7:
                    return False
            else:
                if num >> 6 != 0b10:
                    return False
                n_bytes -= 1

        return n_bytes == 0
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n)  
**Space:** O(1)

</details>

#### Problem 421: [Maximum XOR of Two Numbers](https://leetcode.com/problems/maximum-xor-of-two-numbers-in-an-array/)

**Input**

* `nums`: Array of integers

**Output**

* Maximum XOR of two numbers

**Example**

```python
Input: nums = [3,10,5,25,2,8]
Output: 28
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def findMaximumXOR(self, nums):
        max_xor = 0
        mask = 0

        for i in range(31, -1, -1):
            mask |= (1 << i)
            prefixes = {num & mask for num in nums}
            temp = max_xor | (1 << i)

            for prefix in prefixes:
                if temp ^ prefix in prefixes:
                    max_xor = temp
                    break

        return max_xor
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(32 × n)  
**Space:** O(n)

</details>

#### Problem 477: [Total Hamming Distance](https://leetcode.com/problems/total-hamming-distance/)

**Input**

* `nums`: Array of integers

**Output**

* Sum of Hamming distances between all pairs

**Example**

```python
Input: nums = [4,14,2]
Output: 6
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def totalHammingDistance(self, nums):
        total = 0
        n = len(nums)

        for i in range(32):
            count_ones = sum((num >> i) & 1 for num in nums)
            total += count_ones * (n - count_ones)

        return total
```

</details>

  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(32 × n)  
**Space:** O(1)

```python
class Solution:
    def subarrayBitwiseORs(self, arr):
        result = set()
        current = set()

        for num in arr:
            current = {num | x for x in current} | {num}
            result |= current

        return len(result)
```

#### Problem 1318: [Minimum Flips to Make a OR b Equal to c](https://leetcode.com/problems/minimum-flips-to-make-a-or-b-equal-to-c/)

**Input**

* `a`, `b`, `c`: Three integers

**Output**

* Minimum flips to make a OR b == c

**Example**

```python
Input: a = 2, b = 6, c = 5
Output: 3
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def minFlips(self, a, b, c):
        flips = 0
        for i in range(32):
            bit_a = (a >> i) & 1
            bit_b = (b >> i) & 1
            bit_c = (c >> i) & 1

            if bit_c == 0:
                flips += bit_a + bit_b
            else:
                if bit_a == 0 and bit_b == 0:
                    flips += 1

        return flips
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(1)  
**Space:** O(1)

</details>

#### Problem 78: [Subsets](https://leetcode.com/problems/subsets/)

**Input**

* `nums`: Array of distinct integers

**Output**

* All possible subsets (power set)

**Example**

```python
Input: nums = [1,2,3]
Output: [[],[1],[2],[1,2],[3],[1,3],[2,3],[1,2,3]]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def subsets(self, nums):
        result = []
        n = len(nums)

        for i in range(1 << n):
            subset = []
            for j in range(n):
                if i & (1 << j):
                    subset.append(nums[j])
            result.append(subset)

        return result
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(2^n × n)  
**Space:** O(2^n × n)

</details>

#### Problem 211: [Design Add and Search Words Data Structure](https://leetcode.com/problems/design-add-and-search-words-data-structure/)

**Input**

* Operations: `addWord(word)` and `search(word)` with '.' wildcard

**Output**

* Boolean for search operation

**Example**

```python
Input: ["WordDictionary","addWord","addWord","search","search"]
       [[],["bad"],["dad"],["pad"],["bad"]]
Output: [null,null,null,false,true]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class WordDictionary:
    def __init__(self):
        self.root = {}

    def addWord(self, word):
        node = self.root
        for char in word:
            if char not in node:
                node[char] = {}
            node = node[char]
        node['$'] = True

    def search(self, word):
        def dfs(node, i):
            if i == len(word):
                return '$' in node
            if word[i] == '.':
                return any(dfs(node[c], i+1) for c in node if c != '$')
            if word[i] not in node:
                return False
            return dfs(node[word[i]], i+1)

        return dfs(self.root, 0)
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(m) for add, O(m × 26^k) for search (k = wildcards)  
**Space:** O(total characters)

</details>

#### Problem 1239: [Maximum Length of a Concatenated String with Unique Characters](https://leetcode.com/problems/maximum-length-of-a-concatenated-string-with-unique-characters/)

**Input**

* `arr`: Array of strings

**Output**

* Maximum length of concatenated string with unique characters

**Example**

```python
Input: arr = ["un","iq","ue"]
Output: 4
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def maxLength(self, arr):
        def is_unique(s):
            return len(s) == len(set(s))

        def backtrack(index, current):
            if not is_unique(current):
                return 0

            max_len = len(current)
            for i in range(index, len(arr)):
                max_len = max(max_len, backtrack(i + 1, current + arr[i]))

            return max_len

        return backtrack(0, "")
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(2^n)  
**Space:** O(n)

</details>

#### Problem 1442: [Count Triplets That Can Form Two Arrays of Equal XOR](https://leetcode.com/problems/count-triplets-that-can-form-two-arrays-of-equal-xor/)

**Input**

* `arr`: Array of integers

**Output**

* Count of valid triplets (i, j, k)

**Example**

```python
Input: arr = [2,3,1,6,7]
Output: 4
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def countTriplets(self, arr):
        count = 0
        n = len(arr)

        for i in range(n):
            xor = arr[i]
            for k in range(i + 1, n):
                xor ^= arr[k]
                if xor == 0:
                    count += k - i

        return count
```

#### Problem 1707: [Maximum XOR With an Element From Array](https://leetcode.com/problems/maximum-xor-with-an-element-from-array/)

**Input**

* `nums`: Array of integers
* `queries`: Array of [xi, mi] pairs

**Output**

* Maximum XOR for each query

**Example**

```python
Input: nums = [0,1,2,3,4], queries = [[3,1],[1,3],[5,6]]
Output: [3,3,7]
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class TrieNode:
    def __init__(self):
        self.children = {}

class Solution:
    def maximizeXor(self, nums, queries):
        nums.sort()
        sorted_queries = sorted(enumerate(queries), key=lambda x: x[1][1])

        root = TrieNode()
        result = [-1] * len(queries)
        num_idx = 0

        for query_idx, (xi, mi) in sorted_queries:
            while num_idx < len(nums) and nums[num_idx] <= mi:
                self.insert(root, nums[num_idx])
                num_idx += 1

            if num_idx > 0:
                result[query_idx] = self.query(root, xi)

        return result

    def insert(self, root, num):
        node = root
        for i in range(31, -1, -1):
            bit = (num >> i) & 1
            if bit not in node.children:
                node.children[bit] = TrieNode()
            node = node.children[bit]

    def query(self, root, num):
        node = root
        max_xor = 0
        for i in range(31, -1, -1):
            bit = (num >> i) & 1
            toggled = 1 - bit
            if toggled in node.children:
                max_xor |= (1 << i)
                node = node.children[toggled]
            else:
                node = node.children[bit]
        return max_xor
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(n log n + q log q + 32(n + q))  
**Space:** O(32n)

</details>

#### Problem 1863: [Sum of All Subset XOR Totals](https://leetcode.com/problems/sum-of-all-subset-xor-totals/)

**Input**

* `nums`: Array of integers

**Output**

* Sum of XOR totals for every subset

**Example**

```python
Input: nums = [1,3]
Output: 6
```

<details>
  <summary><strong>Solution Code</strong></summary>

```python
class Solution:
    def subsetXORSum(self, nums):
        def backtrack(index, current_xor):
            if index == len(nums):
                return current_xor

            include = backtrack(index + 1, current_xor ^ nums[index])
            exclude = backtrack(index + 1, current_xor)

            return include + exclude

        return backtrack(0, 0)
```

</details>

<details>
  <summary><strong>Complexity Analysis</strong></summary>

**Time:** O(2^n)  
**Space:** O(n)

</details>
