# Is balanced 

```python
def is_balanced(S):
    #Write down your code here
    dic = {')':'(',']':'[','}':'{'}
    if S == '':
        return True
    stack = []
    for x in S :
        if x in dic:
            
            top = stack.pop() if stack else '#'
            
            if dic[x] != top:
                return False
        else:
            stack.append(x)

    return not stack        
        
```

# Remove Consecutive Duplicates

```python
class Solution:
    def removeDuplicates(self, s: str) -> str:
        #Write your code here
        stk = []
        # Iterate through each character in the string
        for c in s:
            # Check if the stack is not empty and the top element is equal to the current character
            if stk and stk[-1] == c:
                # Remove the last element from the stack i.e., remove duplicate pair
                stk.pop()
            else:
                # Add the current character to the stack
                stk.append(c)
        # Convert the stack to a string and return
        return ''.join(stk)
```

# Next Smaller Element

```python
def next_smaller_elements(arr):
    """
    Function to find the next smaller element for each element in the array.
    :param arr: List[int] -> List of integers
    :return: List[int] -> List of next smaller elements for each element
    """
    # TODO: Implement this function
    stack = []
    
    reult = [-1] * len(arr)
    for x in range(len(arr)):
        
        while stack and arr[stack[-1]] > arr[x] :
            reult[stack.pop()] = arr[x]
        
        stack.append(x)
    
    return reult
```

# Evaluate Postfix Expression
https://leetcode.com/problems/evaluate-reverse-polish-notation/description/

```python

def evaluate_postfix(expression):
    """
    Function to evaluate a postfix expression.
    :param expression: List[str] -> A list of strings representing the postfix expression
    :return: int -> The result of the postfix evaluation
    """
    # TODO: Implement this function
    stack = []
    
    for x in expression:
        if x.isdigit() or (x[0] == '-' and x[1:].isdigit()):
            stack.append(int(x))
        else:
            right_operand = stack.pop()
            left_operand = stack.pop()
            if x == '+':
                stack.append(left_operand + right_operand)
            elif x == '-':
                stack.append(left_operand - right_operand)
            elif x == '*':
                stack.append(left_operand * right_operand)
            elif x == '/':
                # Integer division, truncating towards zero
                stack.append(int(left_operand / right_operand))

    
    return stack.pop()
    
        
```

# Winner of the Circular Game
https://leetcode.com/problems/find-the-winner-of-the-circular-game/

```python
from collections import deque

def find_winner(n, k):
    """
    Find the winner of the circle game.

    Parameters:
    n (int): The number of friends.
    k (int): The counting number.

    Returns:
    int: The number of the winning friend.
    """
    # Implement the function
    players = deque(range(1, n + 1))
    
    while len(players) > 1:
        players.rotate(-(k - 1))
        # Remove the k-th player
        players.popleft()
    
    # The remaining person is the winner
    return players[0]
```

# Largest Rectangle in Histogram
https://leetcode.com/problems/largest-rectangle-in-histogram/description/

```python
def largestRectangleArea(heights):
    # Implement your solution here
    stack = []
    n = len(heights)
    max_area = 0 
    index = 0 
    
    for i in range(n):
        # While the current bar is shorter than the bar at stack top
        while stack and heights[i] < heights[stack[-1]]:
            # Pop the top
            height = heights[stack.pop()]
            # Width is the distance between the current index and the index at the new top of the stack
            width = i if not stack else i - stack[-1] - 1
            # Calculate area
            max_area = max(max_area, height * width)
        
        # Push the current bar onto the stack
        stack.append(i)
    
    # Calculate area for remaining bars in stack
    while stack:
        height = heights[stack.pop()]
        width = n if not stack else n - stack[-1] - 1
        max_area = max(max_area, height * width)
    
    return max_area
```
