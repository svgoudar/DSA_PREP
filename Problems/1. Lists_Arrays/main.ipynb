### Reverse the lists

```python
def reverse_list(lst):
    # Your code goes here
    left,right = 0,len(lst) -1
    while left < right:
        lst[left], lst[right] = lst[right], lst[left]
        left += 1
        right -= 1
    return lst
```

### Maximum difference between two consecutive elements in a list.

```python
def max_consecutive_difference(lst):
    if len(lst) < 2:
        return 0  # No difference possible if < 2 elements

    max_diff = 0  # Initialize max difference

    for i in range(len(lst) - 1):
        diff = abs(lst[i] - lst[i + 1])
        if diff > max_diff:
            max_diff = diff  # Update max difference if larger found

    return max_diff
```

### Merge sorted Lists

```python
def merge_two_sorted_lists(list1, list2):
    # Initialize pointers for both lists
    i, j = 0, 0
    result = []  # Initialize an empty list to store the merged result

    # Traverse both lists and merge them in sorted order
    while i < len(list1) and j < len(list2):
        if list1[i] < list2[j]:
            result.append(list1[i])  # Add the smaller element to the result list
            i += 1
        else:
            result.append(list2[j])  # Add the smaller element to the result list
            j += 1

    # If there are remaining elements in list1, add them to the result
    while i < len(list1):
        result.append(list1[i])
        i += 1

    # If there are remaining elements in list2, add them to the result
    while j < len(list2):
        result.append(list2[j])
        j += 1

    return result  # Return the merged sorted list
```

### Rotate a List (Without Slicing)

```python
def rotate_list(lst, k):
    # Your code goes here
    if not lst:
        return lst
    v= len(lst)
    k = k%v
    for _ in range(k):
        last = lst.pop()
        lst.insert(0,last)

    return lst
```

### Merge 2 List into Dictionary

```python
def merge_lists_to_dictionary(keys, values):
    # Your code goes here
    if len(keys) != len(values):
        return False
    # return {keys[i]:values[i] for i in range(len(keys))}
    return dict(zip(keys,values))
```

### Words Frequency in a Sentence

```python
def count_word_frequency(sentence):
    # Your code goes here
    f = sentence.split(" ")
    d = {}
    if sentence == '':
        return d
    for i in f:
        if i not in d:
            d[i] = 1
        else:
            d[i] +=1
    return d
```

### Palindromic Tuple

```python
def is_palindromic_tuple(tup):
    # Initialize pointers for the start and end of the tuple
    start = 0
    end = len(tup) - 1

    # Loop until the start pointer is less than or equal to the end pointer
    while start <= end:
        # If elements at start and end pointers are not equal, it's not a palindrome
        if tup[start] != tup[end]:
            return False
        # Move the pointers towards the center
        start += 1
        end -= 1

    # If all elements matched, it is a palindrome
    return True
```

### Merge Dictionaries with Common Keys

```python
def merge_dicts_with_overlapping_keys(dicts):
    # Your code goes here
    result = {}

    for dictionary in dicts:
        for key, value in dictionary.items():
            # If the key is already in the result, add the current value
            if key in result:
                result[key] += value
            # Otherwise, set the key with the current value
            else:
                result[key] = value

    return result

```

### Check if List is Subset of another *List*

```python
def is_subset(lst1, lst2):
    # Your code goes here
    v= 0
    for x in lst1:
        if x in lst2:
            v+=1

    return len(lst1) == v
```

### Check Prime Number

```python
def is_prime(n):
    """
    Function to check if a number is prime.

    Parameters:
    n (int): The number to check.

    Returns:
    bool: True if n is prime, False otherwise.
    """
    # Your code here
    if n >1:
        for i in range(2,n//2+1):
            if n % i == 0:
                return False
        else:
            return True

    else:
        return False
```

### Check whether valid square number

```python
def is_perfect_square(num):
    """
    Function to check if a number is a perfect square.

    Parameters:
    num (int): The number to check.

    Returns:
    bool: True if num is a perfect square, False otherwise.
    """
    # Your code here
    if num < 1:
        return False
    if num == 1:
        return True
    for i in range(1,num//2):
        if i*i == num:
            return True
    else:
        return False
```
