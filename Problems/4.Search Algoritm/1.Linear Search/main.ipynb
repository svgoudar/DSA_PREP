### Linear Search 

```python
def linear_search(arr,target):
    size = len(arr)
    for index in range(0,size):
        if(arr[index]==target):
            return index

    return -1



my_list = [10,23,45,70,11]
target = 10

result = linear_search(my_list,target)

print(result)
```

### Linear Serach Problems

#### Linear Serach on 2D list

```python
def search_2d(matrix, target):
    for i in range(len(matrix)):
        for j in range(len(matrix[i])):
            if matrix[i][j] == target:
                return (i, j)
    return (-1, -1)

# Example
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print(search_2d(matrix, 5))  # Output: (1, 1)
```

#### Find First and Last Occurrence

```python
def first_last_occurrence(arr, target):
    first = last = -1
    for i in range(len(arr)):
        if arr[i] == target:
            if first == -1:
                first = i
            last = i
    return first, last

# Example
print(first_last_occurrence([1, 2, 3, 2, 4, 2], 2))  # Output: (1, 5)
```

#### Find All Indices of Target

```python
def find_all_indices(arr, target):
    return [i for i in range(len(arr)) if arr[i] == target]

# Example
print(find_all_indices([5, 2, 7, 7, 3, 7], 7))  # Output: [2, 3, 5]
```

#### Find Element with Maximum Frequency

```python
def first_greater_than(arr, x):
    for i in range(len(arr)):
        if arr[i] > x:
            return arr[i]
    return None

# Example
print(first_greater_than([4, 5, 1, 3, 6], 4))  # Output: 5
```

#### Find the Missing Number in a Range

```python
def find_missing(arr, n):
    for i in range(1, n+1):
        if i not in arr:
            return i

# Example
print(find_missing([1, 2, 4, 5], 5))  # Output: 3
```

#### Search String Matching a Pattern

```python
def search_pattern(strings, pattern):
    pattern = pattern.lower()
    return [s for s in strings if pattern in s.lower()]

# Example
names = ["Alice", "Bob", "Alicia", "Charlie", "al"]
print(search_pattern(names, "ali"))  # Output: ['Alice', 'Alicia', 'al']
```

#### Find Second Occurrence

```python
def second_occurrence(arr, target):
    count = 0
    for i in range(len(arr)):
        if arr[i] == target:
            count += 1
            if count == 2:
                return i
    return -1

# Example
print(second_occurrence([1, 4, 3, 4, 5], 4))  # Output: 3
```
