### Merge Sort 

```python
def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2
        left = merge_sort(arr[:mid])
        right = merge_sort(arr[mid:])
        return merge(left, right)
    return arr

def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result
```

```python
# Time Complexity Analysis:
# The merge_sort function divides the array into halves recursively (log n levels),
# and at each level, merging takes O(n) time.
# So, the overall time complexity is O(n log n).

# Auxiliary arrays refer to the temporary left and right arrays created during the merge process.
# When merging, the algorithm copies elements into these new arrays before combining them back into the original array.
# This extra storage is what contributes to the O(n) space complexity.

# Space Complexity Analysis:
# The algorithm uses additional space for the left and right subarrays at each recursion level.
# In the worst case, the space complexity is O(n) due to the auxiliary arrays.

print("Time Complexity: O(n log n)")
print("Space Complexity: O(n)")
```

```python
# Explanation of log n in Merge Sort:

# In merge sort, the array is divided into two halves at each recursive step.
# This process continues until each subarray has only one element.
# The number of times we can divide the array in half is log₂(n), where n is the number of elements.
# This is because:
# - After 1st division: n/2 subarrays
# - After 2nd division: n/4 subarrays
# - ...
# - After k divisions: n/(2^k) subarrays
# We stop dividing when n/(2^k) = 1, so 2^k = n, which means k = log₂(n).

# Therefore, the depth of the recursion tree (number of division levels) is log₂(n).
# At each level, merging all subarrays takes O(n) time.
# So, total time = O(n) * O(log n) = O(n log n)
```
