```python
import heapq

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

    # Optional for debugging
    def __repr__(self):
        return f"{self.val} -> {self.next}"

class Solution:
    def mergeKLists(self, lists):
        min_heap = []

        # Step 1: Push initial nodes to heap
        for  l in lists:
            if l:
                heapq.heappush(min_heap,  l)  # (value, unique_id, node)

        dummy = ListNode()
        current = dummy

        # Step 2: Pop from heap and push next of popped node
        while min_heap:
            node = heapq.heappop(min_heap)
            current.next = node
            current = current.next

            if node.next:
                heapq.heappush(min_heap,  node.next)

        return dummy.next
```

```python
from collections import Counter

class Solution:
    def checkInclusion(self, s1: str, s2: str) -> bool:
        n, m = len(s1), len(s2)
        if n > m:
            return False

        s1_count = Counter(s1)
        window_count = Counter(s2[:n])

        if s1_count == window_count:
            return True

        for i in range(n, m):
            window_count[s2[i]] += 1        # Add incoming char
            window_count[s2[i - n]] -= 1    # Remove outgoing char

            if window_count[s2[i - n]] == 0:
                del window_count[s2[i - n]]

            if window_count == s1_count:
                return True

        return False


```

```python
def isMatch(s, p):
        # Create a memoization dictionary to store results of subproblems
        memo = {}
        # Define a helper function for recursion with indices i (for s) and j (for p)
        def dp(i, j):
            # If result is already computed, return from memo
            if (i, j) in memo:
                return memo[(i, j)]
            # If we've reached the end of the pattern
            if j == len(p):
                # Return True if s is also fully matched
                ans = i == len(s)
            else:
                # Check if the current character matches (if s is not exhausted)
                first_match = i < len(s) and (p[j] == s[i] or p[j] == '.')
                # If next character in p is '*', we have two choices:
                if (j + 1) < len(p) and p[j + 1] == '*':
                    # 1. Don't use the star (move p by 2)
                    #2. Use the star if first_match (move s by 1, keep p)
                    ans = dp(i, j + 2) or (first_match and dp(i + 1, j))
                else:
                    # If not '*', proceed only if first_match, move both by 1
                    ans = first_match and dp(i + 1, j + 1)
            # Store result in memo and return
            memo[(i, j)] = ans
            return ans
        # Start recursion from the beginning of s and p
        return dp(0, 0)
```

```python
class Solution(object):
    def isScramble(self, s1, s2, memo={}):
        if s1==s2:
            return True
        if (s1,s2) in memo:
            return memo[(s1,s2)]
        if sorted(s1)!=sorted(s2):
            return False
        n = len(s1)
        for i in range(1, n):
            if (self.isScramble(s1[:i], s2[:i], memo) and
                self.isScramble(s1[i:], s2[i:], memo)) or \
               (self.isScramble(s1[:i], s2[n-i:], memo) and
                self.isScramble(s1[i:], s2[:n-i], memo)):
                memo[(s1,s2)] = True
                return True
        memo[(s1,s2)] = False
        return False
```
