### Int to Binary

```python
def int_to_binary(n):
    """
    Function to convert an integer to its binary representation.

    Parameters:
    n (int): The integer to convert.

    Returns:
    str: The binary representation of the integer.
    """
    # Your code here
    if n == 0:
        return str(0)

    bin = '' if n > 0 else '-'

    while abs(n) > 0:
        remainder = abs(n) % 2
        bin = bin + str(remainder)
        n = abs(n)// 2

    return bin
```

### GCD

```python
def gcd(n, m):
    """
    Function to find the GCD of two integers without using built-in functions and recursion.

    Parameters:
    n (int): The first integer.
    m (int): The second integer.

    Returns:
    int: The GCD of n and m.
    """
    # Your code here
    number = n if n < m else m
    g = []
    for i in range(1,number+1):
        if n % i == 0 and m % i == 0:
            g.append(i)

    inn = g[0]

    for i in g:
        if i > inn:
            inn = i

    return inn

```
