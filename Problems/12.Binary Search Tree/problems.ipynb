# Search in BST

```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def search_bst(root, val):
    """
    Function to search for a node in a Binary Search Tree (BST) whose value equals val.
    :param root: TreeNode -> root of the binary search tree
    :param val: int -> the value to search for
    :return: TreeNode -> the node whose value equals val, or None if it doesn't exist
    """
    # TODO: Implement this function
    if root is None or root.val == val:
        return root
 
    # If the current node's value is greater than val, search in the left subtree
    if val < root.val:
        return search_bst(root.left, val)
    
    # If the current node's value is less than val, search in the right subtree
    return search_bst(root.right, val)
```

# Predescessor and successor 

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def find_predecessor_successor(root, key):
    """
    Function to find the predecessor and successor of a node with the given key in a BST.
    
    :param root: TreeNode -> The root of the binary search tree
    :param key: int -> The value of the node for which to find the predecessor and successor
    :return: Tuple[Optional[int], Optional[int]] -> A tuple containing the predecessor and successor
    """
    # TODO: Implement the logic to find the predecessor and successor
    def find_successor(root, key):
        successor = None
        while root:
            if key < root.val:
                successor = root
                root = root.left
            else:
                root = root.right
        return successor
    
    def find_predecessor(root, key):
        predecessor = None
        while root:
            if key > root.val:
                predecessor = root
                root = root.right
            else:
                root = root.left
        return predecessor


    succ = find_successor(root, key)
    pred = find_predecessor(root, key)
    
    
    return pred.val if pred else None, succ.val if succ else None
```

# Kth smallest element in BST
https://leetcode.com/problems/kth-smallest-element-in-a-bst/description/

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def kth_smallest(root, k):
    # Core logic for the learner to implement
    
    def inorder(node):
        if node:
            # Recursively go to the left child
            yield from inorder(node.left)
            # Yield the current node's value
            yield node.val
            # Recursively go to the right child
            yield from inorder(node.right)
    
    # Traverse the BST and get the k-th smallest element
    gen = inorder(root)
    for _ in range(k):
        result = next(gen)
    
    return result
```

# BST Queries
https://www.naukri.com/code360/problems/bst-queries_1095658

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
 
def count_nodes_in_range(root, queries):
    """
    Count the number of nodes within each range [L, R] in a Binary Search Tree (BST).
    
    Approach:
    - Perform an in-order traversal of the BST to get all nodes in sorted order.
    - For each query [L, R], count the number of nodes that lie within this range using binary search.
    """
    
    # Helper function to perform in-order traversal and collect nodes
    def inorder(node):
        if node:
            yield from inorder(node.left)
            yield node.val
            yield from inorder(node.right)
    
    # Collect all node values in sorted order
    sorted_nodes = list(inorder(root))
    
    def count_in_range(L, R):
        """
        Use binary search to find the number of elements within the range [L, R].
        """
        left_index = binary_search_left(sorted_nodes, L)
        right_index = binary_search_right(sorted_nodes, R)
        return right_index - left_index + 1
    
    def binary_search_left(arr, target):
        """
        Binary search to find the first index where arr[index] >= target.
        """
        low, high = 0, len(arr)
        while low < high:
            mid = (low + high) // 2
            if arr[mid] < target:
                low = mid + 1
            else:
                high = mid
        return low
    
    def binary_search_right(arr, target):
        """
        Binary search to find the last index where arr[index] <= target.
        """
        low, high = 0, len(arr)
        while low < high:
            mid = (low + high) // 2
            if arr[mid] <= target:
                low = mid + 1
            else:
                high = mid
        return low - 1
    
    # Process each query
    return [count_in_range(L, R) for L, R in queries]
 
# Explanation:
# 1. In-order traversal is used to get all node values in a sorted list.
# 2. Binary search functions find the range boundaries for each query.
# 3. count_in_range determines how many values fall within the specified range for each query.
```
