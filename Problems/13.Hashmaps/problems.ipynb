# Count even Odd

```python
def count_odd_even_occurrences(ARR):
    """
    Function to count the number of elements that occur an odd number of times
    and the number of elements that occur an even number of times.
    
    :param ARR: List[int] -> The input list of integers
    :return: Tuple[int, int] -> A tuple containing two values:
             - The number of elements with odd occurrences
             - The number of elements with even occurrences
    """
    # TODO: Implement the logic using a hashmap (dictionary)
    v ={}
    for x in ARR:
        
        if x in v :
            v[x]+=1
        else:
            v[x] = 1
    
    
    odd_count = 0
    even_count = 0
    
    # Iterate through the frequency map and count odd/even occurrences
    for frequency in v.values():
        if frequency % 2 == 0:
            even_count += 1
        else:
            odd_count += 1
    
    # Return the result as a tuple (odd_count, even_count)
    return (odd_count, even_count) 
```

# Group anagrams

```python
def group_anagrams(strs):
    """
    This function groups a list of strings into lists of anagrams.
    
    Approach:
    - We use a dictionary to map each sorted version of a string to a list of all its anagrams.
    - The sorted string acts as a unique key, and the anagrams are stored in lists as values.
 
    Example:
    strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
    After sorting, "eat", "tea", and "ate" will all map to the key "aet".
    "tan" and "nat" will map to "ant", and "bat" will map to "abt".
    """
    # Dictionary to store the sorted strings as keys and their anagrams as values
    anagram_groups = {}
 
    for word in strs:
        # Sort the word to get the key (sorted string)
        sorted_word = ''.join(sorted(word))
        
        # Add the word to the corresponding anagram group
        if sorted_word in anagram_groups:
            anagram_groups[sorted_word].append(word)
        else:
            anagram_groups[sorted_word] = [word]
 
    # Return all the anagram groups
    return list(anagram_groups.values())
 
# Explanation:
# 1. We create a dictionary where the key is the sorted version of the string.
# 2. We traverse through the list of strings, sort each string, and group them into lists based on their sorted key.
# 3. The final output is the list of all these anagram groups.
```

# Maximum Frequency Number


```python
def most_frequent_number(arr):
    """
    Function to find and return the number that occurs the most times in the array.
    In case of ties, return the number that appears first in the array.
    
    :param arr: List[int] -> The input list of integers
    :return: int -> The most frequent number in the array
    """
    # Create a hashmap (dictionary) to store the frequency of each element
    frequency_map = {}
    # Create another dictionary to store the first occurrence index of each element
    first_occurrence_map = {}
    
    # Iterate through the list and populate the frequency and first occurrence maps
    for index, num in enumerate(arr):
        if num in frequency_map:
            frequency_map[num] += 1
        else:
            frequency_map[num] = 1
            first_occurrence_map[num] = index
    
    # Initialize variables to find the most frequent number
    max_freq = 0
    most_frequent = None
    
    # Iterate through the frequency map to find the most frequent number
    for num, freq in frequency_map.items():
        # Check if this number has a higher frequency or if it's tied but appears first
        if freq > max_freq or (freq == max_freq and first_occurrence_map[num] < first_occurrence_map[most_frequent]):
            max_freq = freq
            most_frequent = num
    
    return most_frequent
```

# Greatest Number Product

```python
def greatest_product_equal_to_element(arr):
    num_map = {}
    for num in arr:
        num_map[num] = True
    max_product = -1
    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            product = arr[i] * arr[j]
            if product in num_map:
                max_product = max(max_product, product)
    
    return max_product
```

# Two sum

```python
def two_sum(nums, target):
    """
    Function to return indices of the two numbers such that they add up to the target.
    
    :param nums: List[int] -> The input list of integers
    :param target: int -> The target sum
    :return: List[int] -> A list of two indices whose corresponding elements add up to the target
    """
    # TODO: Implement the logic using a hashmap (dictionary)
    num_to_index = {}
    
    # Iterate through the list of numbers
    for i, num in enumerate(nums):
        # Calculate the difference needed to reach the target
        difference = target - num
        
        # Check if the difference is already in the hashmap
        if difference in num_to_index:
            # If found, return the indices of the two numbers
            return [num_to_index[difference], i]
        
        # Otherwise, store the current number and its index in the hashmap
        num_to_index[num] = i
    
    # If no solution is found (though the problem guarantees one solution exists)
    return []

print(two_sum([2, 7, 11, 15],9))
```

# Duplicate Number

```python
def contains_duplicate(nums):
    """
    Function to check if the array contains any duplicate elements.
    :param nums: List[int] -> The input list of integers
    :return: bool -> True if duplicates exist, False otherwise
    """
    # TODO: Implement this function using a hashmap
    b= {}
    for num in nums:
        if num in b:
            return True
        b[num] = True

    return False
        
```
