# Search algorithm
  - Linear
  - Binary
  

```python
def binary_serach(arr,target):
    start = 0 
    end = len(arr) - 1

    while start <= end:
        mid = (start + end) // 2

        if arr[mid] == target :
            return True
        elif arr[mid] < target :
            start = mid +1
        elif arr[mid] > target :
            end = mid -1
    
    return False


```

```python
binary_serach([1,2,3,4,5,6],1)
```
