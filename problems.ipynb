```python
def rotate(l,k):
    if not l:
        return l
    v = len(l)
    k = k%v
    for _ in range(v):
        las = l.pop()
        l.insert(0,las)
    
    return l
```

# Palindromic Tuple

```python
def pallindromical_tuple(strng):
    left = 0
    right = len(strng) -1
    while left < right:
        if strng[left] != strng[right]:
            return False
        left += 1
        right -=1

    return True
```

```python
pallindromical_tuple('racecar')
```
